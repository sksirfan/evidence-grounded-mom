```python
# ============================================================
# FINAL EVALUATION — ES2004c SILERO VAD
# ============================================================

import os
import json
import torch
import torchaudio

MEETING_ID = "ES2004c"

AUDIO_PATH = os.path.join(
    PROJECT_DIR,
    "data",
    "raw",
    "ami",
    MEETING_ID,
    "audio",
    f"{MEETING_ID}.Mix-Headset.wav"
)

VAD_OUTPUT = os.path.join(
    PROJECT_DIR,
    "data",
    "vad",
    f"{MEETING_ID}_silero_vad.json"
)

print("=" * 80)
print(f"SILERO VAD — {MEETING_ID}")
print("=" * 80)

print(f"Audio: {AUDIO_PATH}")

# ------------------------------------------------------------
# Load Silero VAD
# ------------------------------------------------------------

torch.set_num_threads(1)

model, utils = torch.hub.load(
    repo_or_dir="snakers4/silero-vad",
    model="silero_vad",
    force_reload=False
)

(get_speech_timestamps, _, read_audio, _, _) = utils

# ------------------------------------------------------------
# Load audio
# ------------------------------------------------------------

wav = read_audio(
    AUDIO_PATH,
    sampling_rate=16000
)

duration_seconds = len(wav) / 16000

print(f"\nAudio duration: {duration_seconds / 60:.2f} minutes")

# ------------------------------------------------------------
# Detect speech
# ------------------------------------------------------------

speech_timestamps = get_speech_timestamps(
    wav,
    model,
    sampling_rate=16000,
    return_seconds=True
)

# ------------------------------------------------------------
# Calculate statistics
# ------------------------------------------------------------

speech_duration = sum(
    segment["end"] - segment["start"]
    for segment in speech_timestamps
)

silence_duration = max(
    duration_seconds - speech_duration,
    0
)

print("\n" + "=" * 80)
print("VAD RESULTS")
print("=" * 80)

print(f"Speech segments : {len(speech_timestamps)}")
print(f"Speech duration : {speech_duration / 60:.2f} minutes")
print(f"Silence duration: {silence_duration / 60:.2f} minutes")

# ------------------------------------------------------------
# Save VAD output
# ------------------------------------------------------------

vad_output = {
    "meeting_id": MEETING_ID,
    "audio_path": AUDIO_PATH,
    "sampling_rate": 16000,
    "duration_seconds": duration_seconds,
    "speech_duration_seconds": speech_duration,
    "silence_duration_seconds": silence_duration,
    "speech_segment_count": len(speech_timestamps),
    "segments": speech_timestamps
}

os.makedirs(
    os.path.dirname(VAD_OUTPUT),
    exist_ok=True
)

with open(VAD_OUTPUT, "w", encoding="utf-8") as f:
    json.dump(
        vad_output,
        f,
        indent=2
    )

print("\nSaved:")
print(VAD_OUTPUT)

print("\n" + "=" * 80)
print("ES2004c VAD COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# FINAL EVALUATION — ES2004c WHISPERX ASR + ALIGNMENT
# ============================================================

import os
import json
import torch
import whisperx

MEETING_ID = "ES2004c"

AUDIO_PATH = os.path.join(
    PROJECT_DIR,
    "data",
    "raw",
    "ami",
    MEETING_ID,
    "audio",
    f"{MEETING_ID}.Mix-Headset.wav"
)

TRANSCRIPT_OUTPUT = os.path.join(
    PROJECT_DIR,
    "data",
    "transcripts",
    f"{MEETING_ID}_whisperx_aligned.json"
)

print("=" * 80)
print(f"WHISPERX ASR + ALIGNMENT — {MEETING_ID}")
print("=" * 80)

print(f"Audio: {AUDIO_PATH}")

# ------------------------------------------------------------
# Device
# ------------------------------------------------------------

device = "cuda" if torch.cuda.is_available() else "cpu"
compute_type = "float16" if device == "cuda" else "int8"

print(f"\nDevice      : {device}")
print(f"Compute type: {compute_type}")

# ------------------------------------------------------------
# Load WhisperX model
# Same configuration used for ES2004a
# ------------------------------------------------------------

print("\nLoading WhisperX model...")

model = whisperx.load_model(
    "small",
    device=device,
    compute_type=compute_type,
    language="en"
)

# ------------------------------------------------------------
# Load audio
# ------------------------------------------------------------

audio = whisperx.load_audio(AUDIO_PATH)

# ------------------------------------------------------------
# Transcription
# ------------------------------------------------------------

print("\nRunning WhisperX transcription...")

result = model.transcribe(
    audio,
    batch_size=16,
    language="en"
)

print(
    f"Initial transcript segments: "
    f"{len(result.get('segments', []))}"
)

# ------------------------------------------------------------
# Word-level alignment
# ------------------------------------------------------------

print("\nLoading alignment model...")

model_a, metadata = whisperx.load_align_model(
    language_code=result["language"],
    device=device
)

print("Running word-level alignment...")

aligned_result = whisperx.align(
    result["segments"],
    model_a,
    metadata,
    audio,
    device,
    return_char_alignments=False
)

# ------------------------------------------------------------
# Statistics
# ------------------------------------------------------------

aligned_segments = aligned_result.get(
    "segments",
    []
)

word_count = sum(
    len(segment.get("words", []))
    for segment in aligned_segments
)

print("\n" + "=" * 80)
print("WHISPERX RESULTS")
print("=" * 80)

print(f"Transcript segments : {len(result.get('segments', []))}")
print(f"Aligned segments    : {len(aligned_segments)}")
print(f"Aligned words       : {word_count}")

# ------------------------------------------------------------
# Save
# ------------------------------------------------------------

output_data = {
    "meeting_id": MEETING_ID,
    "source": "WhisperX small",
    "language": result.get("language", "en"),
    "audio_path": AUDIO_PATH,
    "device": device,
    "compute_type": compute_type,
    "transcript_segments": result.get("segments", []),
    "aligned_segments": aligned_segments
}

os.makedirs(
    os.path.dirname(TRANSCRIPT_OUTPUT),
    exist_ok=True
)

with open(
    TRANSCRIPT_OUTPUT,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        output_data,
        f,
        indent=2,
        ensure_ascii=False
    )

print("\nSaved:")
print(TRANSCRIPT_OUTPUT)

print("\n" + "=" * 80)
print("ES2004c WHISPERX COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# SPEAKER–WORD ALIGNMENT — ES2004c
# ============================================================

import json
import os

print("=" * 80)
print("SPEAKER–WORD ALIGNMENT — ES2004c")
print("=" * 80)

# ------------------------------------------------------------
# Load aligned WhisperX transcript
# ------------------------------------------------------------

with open(aligned_path, "r", encoding="utf-8") as f:
    aligned_data = json.load(f)

# ------------------------------------------------------------
# Helper: find speaker for a word
# ------------------------------------------------------------

def get_word_speaker(word_start, word_end, diarization_segments):
    best_speaker = None
    best_overlap = 0.0

    for segment in diarization_segments:
        overlap_start = max(word_start, segment["start"])
        overlap_end = min(word_end, segment["end"])

        overlap = max(0.0, overlap_end - overlap_start)

        if overlap > best_overlap:
            best_overlap = overlap
            best_speaker = segment["speaker"]

    return best_speaker


# ------------------------------------------------------------
# Assign speaker to every aligned word
# ------------------------------------------------------------

aligned_words = []

for segment in aligned_data["segments"]:

    for word in segment.get("words", []):

        if "start" not in word or "end" not in word:
            continue

        word_start = float(word["start"])
        word_end = float(word["end"])

        speaker = get_word_speaker(
            word_start,
            word_end,
            diarization_segments
        )

        aligned_words.append({
            "word": word["word"],
            "start": word_start,
            "end": word_end,
            "speaker": speaker
        })


print("Aligned words with speakers:", len(aligned_words))

# ------------------------------------------------------------
# Group consecutive words into speaker utterances
# ------------------------------------------------------------

speaker_utterances = []

current = None

for word in aligned_words:

    speaker = word["speaker"]

    if speaker is None:
        continue

    if current is None:
        current = {
            "speaker": speaker,
            "start": word["start"],
            "end": word["end"],
            "words": [word["word"].strip()]
        }

    else:

        time_gap = word["start"] - current["end"]

        if (
            speaker != current["speaker"]
            or time_gap > 1.0
        ):
            speaker_utterances.append({
                "speaker": current["speaker"],
                "start": current["start"],
                "end": current["end"],
                "text": " ".join(current["words"])
            })

            current = {
                "speaker": speaker,
                "start": word["start"],
                "end": word["end"],
                "words": [word["word"].strip()]
            }

        else:
            current["words"].append(word["word"].strip())
            current["end"] = word["end"]


# Add final utterance
if current is not None:
    speaker_utterances.append({
        "speaker": current["speaker"],
        "start": current["start"],
        "end": current["end"],
        "text": " ".join(current["words"])
    })


print("Speaker utterances:", len(speaker_utterances))

print("=" * 80)
```

```python
# ============================================================
# RULE-BASED MoM CANDIDATE EXTRACTION — ES2004c
# ============================================================

import re

# ------------------------------------------------------------
# MoM-relevant linguistic indicators
# ------------------------------------------------------------

ACTION_PATTERNS = [
    r"\bwe (?:will|should|need to|have to|must|plan to|are going to)\b",
    r"\bI (?:will|shall|can|need to|have to)\b",
    r"\b(?:let's|lets)\b",
    r"\b(?:next meeting|next time)\b",
    r"\b(?:by|before|until) (?:tomorrow|next week|next meeting|Friday|Monday|Tuesday|Wednesday|Thursday|Saturday|Sunday)\b",
]

DECISION_PATTERNS = [
    r"\bwe (?:decided|agreed|concluded)\b",
    r"\b(?:decision|agreed|agreement)\b",
    r"\bwe'll go with\b",
    r"\blet's go with\b",
]

DISCUSSION_PATTERNS = [
    r"\b(?:discuss|discussion|consider|considering)\b",
    r"\b(?:idea|option|alternative|suggest|suggested)\b",
    r"\bwhat about\b",
    r"\bhow about\b",
]

INFORMATION_PATTERNS = [
    r"\b(?:the|our|this) (?:project|product|design|system|meeting)\b",
    r"\b(?:is|are|was|were|will be)\b",
    r"\b(?:need|requires|requirement|goal|objective)\b",
]

# ------------------------------------------------------------
# Candidate extraction
# ------------------------------------------------------------

mom_candidates = []

candidate_id = 1

for utterance_id, utterance in enumerate(speaker_utterances):

    text = utterance["text"].strip()

    if not text:
        continue

    text_lower = text.lower()

    matched_types = []

    for pattern in ACTION_PATTERNS:
        if re.search(pattern, text_lower):
            matched_types.append("ACTION")
            break

    for pattern in DECISION_PATTERNS:
        if re.search(pattern, text_lower):
            matched_types.append("DECISION")
            break

    for pattern in DISCUSSION_PATTERNS:
        if re.search(pattern, text_lower):
            matched_types.append("DISCUSSION")
            break

    for pattern in INFORMATION_PATTERNS:
        if re.search(pattern, text_lower):
            matched_types.append("INFORMATION")
            break

    if matched_types:

        # Priority:
        # ACTION > DECISION > DISCUSSION > INFORMATION
        if "ACTION" in matched_types:
            event_type = "ACTION"
        elif "DECISION" in matched_types:
            event_type = "DECISION"
        elif "DISCUSSION" in matched_types:
            event_type = "DISCUSSION"
        else:
            event_type = "INFORMATION"

        mom_candidates.append({
            "candidate_id": candidate_id,
            "source_utterance_id": utterance_id,
            "speaker": utterance["speaker"],
            "start": utterance["start"],
            "end": utterance["end"],
            "text": text,
            "event_type": event_type
        })

        candidate_id += 1

print("=" * 80)
print("RULE-BASED MoM CANDIDATE EXTRACTION")
print("=" * 80)

print("Total speaker utterances:", len(speaker_utterances))
print("MoM candidates:", len(mom_candidates))

print("\nSample candidates:")

for candidate in mom_candidates[:10]:
    print(
        f"[{candidate['candidate_id']:02d}] "
        f"{candidate['speaker']} | "
        f"{candidate['start']:.2f}-{candidate['end']:.2f}s | "
        f"{candidate['event_type']} | "
        f"{candidate['text']}"
    )

print("=" * 80)
```

```python
# ============================================================
# INSPECT ES2004c CANDIDATE DISTRIBUTION
# ============================================================

from collections import Counter

print("=" * 80)
print("ES2004c MoM CANDIDATE DISTRIBUTION")
print("=" * 80)

event_counts = Counter(
    candidate["event_type"]
    for candidate in mom_candidates
)

print("Total candidates:", len(mom_candidates))

for event_type, count in event_counts.items():
    print(f"{event_type}: {count}")

print("=" * 80)
```

```python
# ============================================================
# FILTER MoM-WORTHY CANDIDATES — ES2004c
# Same filtering stage used for ES2004a
# ============================================================

MOM_WORTHY_KEYWORDS = [
    "will",
    "going to",
    "need to",
    "have to",
    "should",
    "must",
    "plan",
    "planned",
    "decided",
    "agreed",
    "action",
    "task",
    "deadline",
    "by ",
    "before ",
    "next meeting",
    "next time",
    "requirement",
    "goal",
    "objective",
    "price",
    "cost",
    "market",
    "design",
    "product",
    "interface",
    "material",
    "battery",
    "port",
    "alarm",
    "user",
    "customer",
    "feedback",
    "problem",
    "issue",
    "solution",
    "option",
    "idea",
]

mom_worthy_candidates = []

for candidate in mom_candidates:

    text_lower = candidate["text"].lower()

    # Keep candidates containing substantive MoM-related content.
    has_mom_keyword = any(
        keyword in text_lower
        for keyword in MOM_WORTHY_KEYWORDS
    )

    # Actions and decisions are retained directly.
    if candidate["event_type"] in {"ACTION", "DECISION"}:
        keep = True

    # Discussion/information requires substantive MoM content.
    elif candidate["event_type"] in {"DISCUSSION", "INFORMATION"}:
        keep = has_mom_keyword

    else:
        keep = False

    if keep:
        mom_worthy_candidates.append(candidate)

print("=" * 80)
print("MoM-WORTHY CANDIDATE FILTERING — ES2004c")
print("=" * 80)

print("Original candidates:", len(mom_candidates))
print("MoM-worthy candidates:", len(mom_worthy_candidates))
print(
    "Removed:",
    len(mom_candidates) - len(mom_worthy_candidates)
)

print("\nSample retained candidates:")

for candidate in mom_worthy_candidates[:15]:
    print(
        f"[{candidate['candidate_id']:02d}] "
        f"{candidate['speaker']} | "
        f"{candidate['start']:.2f}-{candidate['end']:.2f}s | "
        f"{candidate['event_type']} | "
        f"{candidate['text']}"
    )

print("=" * 80)
```
