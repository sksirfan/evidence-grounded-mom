````python
import json
import os

# File paths
claims_path = os.path.join(
    PROJECT_ROOT,
    "data/transcripts/ES2004a_structured_mom_claims.json"
)

transcript_path = os.path.join(
    PROJECT_ROOT,
    "data/transcripts/ES2004a_speaker_transcript.json"
)

evidence_path = os.path.join(
    PROJECT_ROOT,
    "data/evidence/ES2004a_evidence_documents.json"
)

# Load files
with open(claims_path, "r", encoding="utf-8") as f:
    claims_data = json.load(f)

with open(transcript_path, "r", encoding="utf-8") as f:
    transcript_data = json.load(f)

with open(evidence_path, "r", encoding="utf-8") as f:
    evidence_data = json.load(f)

# Inspect structure
print("STRUCTURED MoM")
print(type(claims_data))
print("Keys:", claims_data.keys())

print("\nTRANSCRIPT")
print(type(transcript_data))
if isinstance(transcript_data, dict):
    print("Keys:", transcript_data.keys())
else:
    print("Number of records:", len(transcript_data))

print("\nEVIDENCE")
print(type(evidence_data))
if isinstance(evidence_data, dict):
    print("Keys:", evidence_data.keys())
else:
    print("Number of records:", len(evidence_data))
````

````python
import os

# ============================================================
# CREATE A COMPLETELY NEW DASHBOARD APP
# ============================================================

PROJECT_ROOT = "/content/drive/MyDrive/MTechIndProj/MoM_Project"

DASHBOARD_DIR = os.path.join(
    PROJECT_ROOT,
    "07_dashboard"
)

APP_PATH = os.path.join(
    DASHBOARD_DIR,
    "app.py"
)

os.makedirs(
    DASHBOARD_DIR,
    exist_ok=True
)


APP_CODE = r'''
# ============================================================
# VERIFIABLE MEETING MINUTES
# Multi-Attribute Evidence Consistency
#
# Clean Streamlit Dashboard
# ============================================================

import os
import json
import html

import streamlit as st

from reportlab.lib.pagesizes import A4
from reportlab.platypus import (
    SimpleDocTemplate,
    Paragraph,
    Spacer,
    Table,
    TableStyle
)
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet


# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

PROJECT_ROOT = (
    "/content/drive/MyDrive/MTechIndProj/MoM_Project"
)

MEETING_ID = "ES2004a"


# ============================================================
# 2. DATA PATHS
# ============================================================

CLAIMS_PATH = os.path.join(
    PROJECT_ROOT,
    "data",
    "transcripts",
    f"{MEETING_ID}_structured_mom_claims.json"
)

TRANSCRIPT_PATH = os.path.join(
    PROJECT_ROOT,
    "data",
    "transcripts",
    f"{MEETING_ID}_speaker_transcript.json"
)

EVIDENCE_PATH = os.path.join(
    PROJECT_ROOT,
    "data",
    "evidence",
    f"{MEETING_ID}_evidence_documents.json"
)

VERIFICATION_PATH = os.path.join(
    PROJECT_ROOT,
    "outputs",
    "verified",
    f"{MEETING_ID}_M9_full_multi_attribute_verification.json"
)

PDF_DIR = os.path.join(
    PROJECT_ROOT,
    "outputs",
    "pdf"
)

PDF_PATH = os.path.join(
    PDF_DIR,
    f"{MEETING_ID}_Verifiable_Meeting_Minutes.pdf"
)


# ============================================================
# 3. HELPER FUNCTIONS
# ============================================================

def load_json(path):
    """Load a JSON file."""

    with open(
        path,
        "r",
        encoding="utf-8"
    ) as file:

        return json.load(file)


def format_timestamp(seconds):
    """
    Convert seconds to MM:SS.

    Example:
        84.66 -> 01:24
        105.24 -> 01:45
    """

    if seconds is None:
        return "N/A"

    seconds = float(seconds)

    minutes = int(seconds // 60)
    secs = int(seconds % 60)

    return f"{minutes:02d}:{secs:02d}"


def safe_html(text):
    """Escape text before inserting into HTML/PDF."""

    if text is None:
        return ""

    return html.escape(
        str(text)
    )


# ============================================================
# 4. LOAD ES2004a DATA
# ============================================================

claims_data = load_json(
    CLAIMS_PATH
)

transcript_data = load_json(
    TRANSCRIPT_PATH
)

evidence_data = load_json(
    EVIDENCE_PATH
)

verification_data = load_json(
    VERIFICATION_PATH
)

results = verification_data[
    "results"
]


# ============================================================
# 5. STREAMLIT CONFIGURATION
# ============================================================

st.set_page_config(
    page_title="Verifiable Meeting Minutes",
    page_icon="📝",
    layout="wide"
)


# ============================================================
# 6. CUSTOM UI STYLE
# ============================================================

st.markdown(
    """
    <style>

    /* -------------------------------------------------------
       Page
       ------------------------------------------------------- */

    .stApp {
        background:
            linear-gradient(
                135deg,
                #f8faff 0%,
                #f4f3ff 100%
            );
    }


    /* -------------------------------------------------------
       Main header
       ------------------------------------------------------- */

    .hero {
        background:
            linear-gradient(
                135deg,
                #312e81,
                #4f46e5,
                #7c3aed
            );

        padding: 30px 34px;

        border-radius: 22px;

        color: white;

        margin-bottom: 25px;

        box-shadow:
            0 12px 30px
            rgba(79, 70, 229, 0.22);
    }

    .hero-title {
        font-size: 36px;
        font-weight: 800;
        margin-bottom: 8px;
    }

    .hero-subtitle {
        font-size: 16px;
        opacity: 0.92;
    }


    /* -------------------------------------------------------
       Mode cards
       ------------------------------------------------------- */

    .mode-card {
        background: white;

        border-radius: 20px;

        padding: 24px;

        min-height: 245px;

        border: 1px solid #e5e7eb;

        box-shadow:
            0 8px 24px
            rgba(15, 23, 42, 0.07);
    }

    .mode-title {
        font-size: 22px;
        font-weight: 750;
        color: #1e293b;
    }

    .mode-description {
        color: #64748b;
        font-size: 14px;
        margin-top: 5px;
        margin-bottom: 18px;
    }


    /* -------------------------------------------------------
       Metrics
       ------------------------------------------------------- */

    .metric-card {
        background: white;

        border-radius: 16px;

        padding: 16px;

        text-align: center;

        border: 1px solid #e5e7eb;

        box-shadow:
            0 5px 16px
            rgba(15, 23, 42, 0.06);
    }

    .metric-value {
        font-size: 27px;
        font-weight: 800;
    }

    .metric-label {
        color: #64748b;
        font-size: 12px;
        margin-top: 3px;
    }


    /* -------------------------------------------------------
       Upload information
       ------------------------------------------------------- */

    .upload-box {
        background: #eef2ff;

        border: 2px dashed #6366f1;

        border-radius: 16px;

        padding: 18px;

        color: #3730a3;

        margin-bottom: 15px;
    }


    /* -------------------------------------------------------
       Section titles
       ------------------------------------------------------- */

    .section-title {
        font-size: 25px;

        font-weight: 800;

        color: #1e293b;

        margin-top: 10px;

        margin-bottom: 15px;
    }


    /* -------------------------------------------------------
       Evidence card
       ------------------------------------------------------- */

    .evidence-card {
        background: #f8fafc;

        border-left: 4px solid #6366f1;

        border-radius: 10px;

        padding: 13px 16px;

        margin: 8px 0;
    }


    /* -------------------------------------------------------
       Status
       ------------------------------------------------------- */

    .status-verified {
        color: #166534;

        background: #dcfce7;

        padding: 6px 13px;

        border-radius: 20px;

        font-weight: 700;
    }

    .status-flagged {
        color: #991b1b;

        background: #fee2e2;

        padding: 6px 13px;

        border-radius: 20px;

        font-weight: 700;
    }

    </style>
    """,
    unsafe_allow_html=True
)


# ============================================================
# 7. HERO HEADER
# ============================================================

st.markdown(
    """
    <div class="hero">

        <div class="hero-title">
            📝 Verifiable Meeting Minutes
        </div>

        <div class="hero-subtitle">
            Multi-Attribute Evidence Consistency
            &nbsp; • &nbsp;
            Evidence-Grounded Meeting Intelligence
        </div>

    </div>
    """,
    unsafe_allow_html=True
)


# ============================================================
# 8. TWO OPERATING MODES
# ============================================================

implemented_col, upload_col = st.columns(
    2,
    gap="large"
)


# ============================================================
# LEFT MODE
# IMPLEMENTED MEETINGS
# ============================================================

with implemented_col:

    st.markdown(
        """
        <div class="mode-card">

            <div class="mode-title">
                📚 Implemented Meetings
            </div>

            <div class="mode-description">
                Explore meetings that have already been
                processed through the evidence verification
                pipeline.
            </div>

        </div>
        """,
        unsafe_allow_html=True
    )


    implemented_meetings = [
        "ES2004a"
    ]


    selected_meeting = st.selectbox(
        "Select Meeting",
        implemented_meetings,
        key="implemented_meeting"
    )


    st.markdown(
        f"**Selected:** `{selected_meeting}`"
    )


    # --------------------------------------------------------
    # Summary metrics
    # --------------------------------------------------------

    total_claims = verification_data[
        "total_claims"
    ]

    verified_claims = verification_data[
        "verified_claims"
    ]

    flagged_claims = verification_data[
        "flagged_claims"
    ]

    verification_rate = verification_data[
        "verification_rate"
    ]


    metric1, metric2 = st.columns(2)

    with metric1:

        st.markdown(
            f"""
            <div class="metric-card">

                <div class="metric-value">
                    {total_claims}
                </div>

                <div class="metric-label">
                    Total MoM Claims
                </div>

            </div>
            """,
            unsafe_allow_html=True
        )


    with metric2:

        st.markdown(
            f"""
            <div class="metric-card">

                <div
                    class="metric-value"
                    style="color:#16a34a;"
                >
                    {verified_claims}
                </div>

                <div class="metric-label">
                    Verified
                </div>

            </div>
            """,
            unsafe_allow_html=True
        )


    metric3, metric4 = st.columns(2)

    with metric3:

        st.markdown(
            f"""
            <div class="metric-card">

                <div
                    class="metric-value"
                    style="color:#dc2626;"
                >
                    {flagged_claims}
                </div>

                <div class="metric-label">
                    Flagged
                </div>

            </div>
            """,
            unsafe_allow_html=True
        )


    with metric4:

        st.markdown(
            f"""
            <div class="metric-card">

                <div
                    class="metric-value"
                    style="color:#7c3aed;"
                >
                    {verification_rate:.2f}%
                </div>

                <div class="metric-label">
                    Verification Rate
                </div>

            </div>
            """,
            unsafe_allow_html=True
        )


# ============================================================
# RIGHT MODE
# UPLOAD NEW MEETING
# ============================================================

with upload_col:

    st.markdown(
        """
        <div class="mode-card">

            <div class="mode-title">
                📤 Upload New Meeting
            </div>

            <div class="mode-description">
                Upload a meeting recording and run the
                evidence-grounded MoM pipeline.
            </div>

            <div class="upload-box">

                🎙️ <b>Meeting Recording</b>

                <br><br>

                Supported formats:
                WAV • MP3 • M4A • MP4

                <br><br>

                Processing will include:
                VAD • ASR • Diarization •
                MoM Generation • Evidence Retrieval •
                Verification

            </div>

        </div>
        """,
        unsafe_allow_html=True
    )


    uploaded_file = st.file_uploader(
        "Choose meeting recording",
        type=[
            "wav",
            "mp3",
            "m4a",
            "mp4"
        ],
        key="meeting_upload"
    )


    if uploaded_file is not None:

        st.success(
            f"Uploaded: {uploaded_file.name}"
        )


        process_clicked = st.button(
            "🚀 Process Meeting",
            use_container_width=True,
            key="process_meeting"
        )


        if process_clicked:

            st.info(
                "Upload processing pipeline will be "
                "connected in the next implementation stage."
            )


# ============================================================
# 9. RESULTS SECTION
# ============================================================

st.divider()

st.markdown(
    """
    <div class="section-title">
        📊 Meeting Results
    </div>
    """,
    unsafe_allow_html=True
)


# ============================================================
# 10. CLAIM DISPLAY
# ============================================================

for idx, result in enumerate(
    results,
    start=1
):

    status = result.get(
        "status",
        "UNKNOWN"
    )

    claim = result.get(
        "claim",
        ""
    )

    claim_speaker = result.get(
        "claim_speaker",
        "Unknown"
    )

    claim_type = result.get(
        "claim_type",
        "Unknown"
    )

    reason = result.get(
        "reason",
        ""
    )


    with st.expander(
        f"Claim {idx} — {status}",
        expanded=(idx == 1)
    ):

        st.markdown(
            f"### {claim}"
        )


        # ----------------------------------------------------
        # Metadata
        # ----------------------------------------------------

        meta1, meta2, meta3 = st.columns(3)


        with meta1:

            st.markdown(
                f"**👤 Speaker:** {claim_speaker}"
            )


        with meta2:

            st.markdown(
                f"**🏷 Event Type:** {claim_type}"
            )


        with meta3:

            if result.get(
                "evidence_start"
            ) is not None:

                start = result[
                    "evidence_start"
                ]

                end = result.get(
                    "evidence_end",
                    start
                )

                st.markdown(
                    f"**⏱ Evidence Time:** "
                    f"{format_timestamp(start)} – "
                    f"{format_timestamp(end)}"
                )

            else:

                st.markdown(
                    "**⏱ Evidence Time:** N/A"
                )


        st.divider()


        # ----------------------------------------------------
        # Verification status
        # ----------------------------------------------------

        if status == "VERIFIED":

            st.success(
                "✓ VERIFIED"
            )

        else:

            st.error(
                "✗ FLAGGED"
            )


        st.markdown(
            f"**Reason:** {reason}"
        )


        # ====================================================
        # CONSISTENCY CHECKS
        # ====================================================

        st.markdown(
            "#### 🔍 Multi-Attribute Consistency"
        )


        checks = result.get(
            "checks",
            {}
        )


        check_names = [
            (
                "content_consistency",
                "Content"
            ),
            (
                "speaker_consistency",
                "Speaker"
            ),
            (
                "timestamp_consistency",
                "Timestamp"
            ),
            (
                "event_type_consistency",
                "Event Type"
            )
        ]


        check_columns = st.columns(4)


        for col, (
            check_key,
            display_name
        ) in zip(
            check_columns,
            check_names
        ):

            check = checks.get(
                check_key,
                {}
            )


            with col:

                if check_key == (
                    "timestamp_consistency"
                ):

                    st.info(
                        f"ℹ **{display_name}**\n\n"
                        "Not independently evaluated"
                    )


                elif not check.get(
                    "applicable",
                    True
                ):

                    st.info(
                        f"— **{display_name}**\n\n"
                        "Not applicable"
                    )


                elif check.get(
                    "passed",
                    False
                ):

                    st.success(
                        f"✓ **{display_name}**\n\n"
                        "Passed"
                    )


                else:

                    st.error(
                        f"✗ **{display_name}**\n\n"
                        "Failed"
                    )


        # ====================================================
        # SUPPORTING EVIDENCE
        # ====================================================

        st.markdown(
            "#### 🔎 Supporting Evidence"
        )


        content_check = checks.get(
            "content_consistency",
            {}
        )


        details = content_check.get(
            "details",
            {}
        )


        supporting_candidates = details.get(
            "supporting_candidates",
            []
        )


        if not supporting_candidates:

            st.info(
                "No supporting evidence available."
            )


        else:

            for evidence in supporting_candidates:

                evidence_id = evidence.get(
                    "evidence_id",
                    "N/A"
                )

                speaker = evidence.get(
                    "speaker",
                    "Unknown"
                )

                start = evidence.get(
                    "start",
                    0
                )

                end = evidence.get(
                    "end",
                    0
                )

                text = evidence.get(
                    "text",
                    ""
                )


                st.markdown(
                    f"""
                    <div class="evidence-card">

                        <b>
                            Evidence {evidence_id}
                        </b>

                        &nbsp; • &nbsp;

                        {speaker}

                        &nbsp; • &nbsp;

                        ⏱ {format_timestamp(start)}
                        –
                        {format_timestamp(end)}

                        <br><br>

                        {safe_html(text)}

                    </div>
                    """,
                    unsafe_allow_html=True
                )


                st.caption(
                    f"NLI: "
                    f"{evidence.get('nli_label', 'N/A')}"
                    f"  |  "
                    f"Retrieval similarity: "
                    f"{evidence.get('retrieval_similarity', 0):.3f}"
                )


# ============================================================
# 11. PDF GENERATION
# ============================================================

def create_mom_pdf(
    results,
    verification_data
):
    """
    Create the clean project-deliverable PDF.

    The PDF intentionally excludes the detailed
    four-attribute consistency table because those
    diagnostics are already visible in the dashboard.
    """

    os.makedirs(
        PDF_DIR,
        exist_ok=True
    )


    styles = getSampleStyleSheet()

    title_style = styles[
        "Title"
    ]

    heading_style = styles[
        "Heading2"
    ]

    body_style = styles[
        "BodyText"
    ]


    document = SimpleDocTemplate(
        PDF_PATH,
        pagesize=A4,
        rightMargin=40,
        leftMargin=40,
        topMargin=40,
        bottomMargin=40
    )


    story = []


    # --------------------------------------------------------
    # PDF header
    # --------------------------------------------------------

    story.append(
        Paragraph(
            "Verifiable Meeting Minutes",
            title_style
        )
    )


    story.append(
        Paragraph(
            "Multi-Attribute Evidence Consistency",
            heading_style
        )
    )


    story.append(
        Paragraph(
            f"<b>Meeting:</b> {MEETING_ID}",
            body_style
        )
    )


    story.append(
        Spacer(
            1,
            15
        )
    )


    # --------------------------------------------------------
    # Summary table
    # --------------------------------------------------------

    summary_data = [
        [
            "Total Claims",
            "Verified",
            "Flagged",
            "Verification Rate"
        ],
        [
            str(
                verification_data[
                    "total_claims"
                ]
            ),
            str(
                verification_data[
                    "verified_claims"
                ]
            ),
            str(
                verification_data[
                    "flagged_claims"
                ]
            ),
            (
                f"{verification_data['verification_rate']:.2f}%"
            )
        ]
    ]


    summary_table = Table(
        summary_data
    )


    summary_table.setStyle(
        TableStyle([
            (
                "BACKGROUND",
                (0, 0),
                (-1, 0),
                colors.lightgrey
            ),
            (
                "GRID",
                (0, 0),
                (-1, -1),
                0.5,
                colors.grey
            ),
            (
                "ALIGN",
                (0, 0),
                (-1, -1),
                "CENTER"
            ),
            (
                "FONTNAME",
                (0, 0),
                (-1, 0),
                "Helvetica-Bold"
            ),
            (
                "TOPPADDING",
                (0, 0),
                (-1, 0),
                8
            ),
            (
                "BOTTOMPADDING",
                (0, 0),
                (-1, 0),
                8
            )
        ])
    )


    story.append(
        summary_table
    )


    story.append(
        Spacer(
            1,
            20
        )
    )


    # --------------------------------------------------------
    # Claims
    # --------------------------------------------------------

    for idx, result in enumerate(
        results,
        start=1
    ):

        status = result.get(
            "status",
            "UNKNOWN"
        )

        claim = safe_html(
            result.get(
                "claim",
                ""
            )
        )

        speaker = safe_html(
            result.get(
                "claim_speaker",
                "Unknown"
            )
        )

        event_type = safe_html(
            result.get(
                "claim_type",
                "Unknown"
            )
        )

        reason = safe_html(
            result.get(
                "reason",
                ""
            )
        )


        story.append(
            Paragraph(
                f"Claim {idx} — "
                f"{safe_html(status)}",
                heading_style
            )
        )


        story.append(
            Paragraph(
                f"<b>Claim:</b> {claim}",
                body_style
            )
        )


        story.append(
            Paragraph(
                f"<b>Speaker:</b> {speaker}",
                body_style
            )
        )


        story.append(
            Paragraph(
                f"<b>Event Type:</b> {event_type}",
                body_style
            )
        )


        if result.get(
            "evidence_start"
        ) is not None:

            start = result[
                "evidence_start"
            ]

            end = result.get(
                "evidence_end",
                start
            )

            story.append(
                Paragraph(
                    f"<b>Evidence Time:</b> "
                    f"{format_timestamp(start)} – "
                    f"{format_timestamp(end)}",
                    body_style
                )
            )


        story.append(
            Paragraph(
                f"<b>Verification:</b> "
                f"{safe_html(status)}",
                body_style
            )
        )


        story.append(
            Paragraph(
                f"<b>Reason:</b> {reason}",
                body_style
            )
        )


        story.append(
            Spacer(
                1,
                10
            )
        )


        # ----------------------------------------------------
        # Supporting evidence
        # ----------------------------------------------------

        story.append(
            Paragraph(
                "<b>Supporting Evidence</b>",
                body_style
            )
        )


        checks = result.get(
            "checks",
            {}
        )


        content_check = checks.get(
            "content_consistency",
            {}
        )


        details = content_check.get(
            "details",
            {}
        )


        supporting_candidates = details.get(
            "supporting_candidates",
            []
        )


        for evidence in supporting_candidates:

            evidence_id = evidence.get(
                "evidence_id",
                "N/A"
            )

            evidence_speaker = safe_html(
                evidence.get(
                    "speaker",
                    "Unknown"
                )
            )

            evidence_start = evidence.get(
                "start",
                0
            )

            evidence_end = evidence.get(
                "end",
                0
            )

            evidence_text = safe_html(
                evidence.get(
                    "text",
                    ""
                )
            )


            story.append(
                Paragraph(
                    f"<b>Evidence "
                    f"{safe_html(evidence_id)}</b> "
                    f"| {evidence_speaker} | "
                    f"{format_timestamp(evidence_start)}"
                    f"–"
                    f"{format_timestamp(evidence_end)}",
                    body_style
                )
            )


            story.append(
                Paragraph(
                    evidence_text,
                    body_style
                )
            )


            story.append(
                Spacer(
                    1,
                    6
                )
            )


        story.append(
            Spacer(
                1,
                15
            )
        )


    document.build(
        story
    )


    return PDF_PATH


# ============================================================
# 12. PDF EXPORT
# ============================================================

st.divider()

st.subheader(
    "📥 Export"
)


if st.button(
    "📥 Export Verified MoM as PDF",
    use_container_width=True
):

    try:

        generated_pdf = create_mom_pdf(
            results,
            verification_data
        )


        with open(
            generated_pdf,
            "rb"
        ) as pdf_file:

            pdf_bytes = pdf_file.read()


        st.download_button(
            label="Download PDF",
            data=pdf_bytes,
            file_name=(
                f"{MEETING_ID}_"
                "Verifiable_Meeting_Minutes.pdf"
            ),
            mime="application/pdf",
            use_container_width=True
        )


        st.success(
            "PDF generated successfully."
        )


    except Exception as error:

        st.error(
            f"PDF generation failed: {error}"
        )

'''


# ============================================================
# WRITE NEW APP
# ============================================================

with open(
    APP_PATH,
    "w",
    encoding="utf-8"
) as file:

    file.write(
        APP_CODE
    )


print("✓ New app.py created from scratch")
print(APP_PATH)
print(
    "Size:",
    os.path.getsize(APP_PATH),
    "bytes"
)
````

````python
import os

PROJECT_ROOT = "/content/drive/MyDrive/MTechIndProj/MoM_Project"

DASHBOARD_DIR = os.path.join(
    PROJECT_ROOT,
    "07_dashboard"
)

UTILS_PATH = os.path.join(
    DASHBOARD_DIR,
    "dashboard_utils.py"
)

UTILS_CODE = r'''
# ============================================================
# DASHBOARD UTILITIES
# Verifiable Meeting Minutes
# ============================================================

import json
import html


# ============================================================
# 1. JSON UTILITIES
# ============================================================

def load_json(path):
    """
    Load a JSON file from disk.
    """

    with open(
        path,
        "r",
        encoding="utf-8"
    ) as file:

        return json.load(file)


# ============================================================
# 2. TIMESTAMP UTILITIES
# ============================================================

def format_timestamp(seconds):
    """
    Convert seconds to MM:SS format.

    Examples:
        84.66  -> 01:24
        105.24 -> 01:45
    """

    if seconds is None:
        return "N/A"

    seconds = float(seconds)

    minutes = int(
        seconds // 60
    )

    secs = int(
        seconds % 60
    )

    return f"{minutes:02d}:{secs:02d}"


# ============================================================
# 3. TEXT UTILITIES
# ============================================================

def safe_html(text):
    """
    Escape text before displaying it as HTML.
    """

    if text is None:
        return ""

    return html.escape(
        str(text)
    )


# ============================================================
# 4. VERIFICATION UTILITIES
# ============================================================

def get_verification_status(result):
    """
    Return the verification status of a claim.
    """

    return result.get(
        "status",
        "UNKNOWN"
    )


def get_verification_reason(result):
    """
    Return the verification reason.
    """

    return result.get(
        "reason",
        ""
    )


def get_claim_metadata(result):
    """
    Extract commonly used claim metadata.
    """

    return {
        "claim": result.get(
            "claim",
            ""
        ),
        "speaker": result.get(
            "claim_speaker",
            "Unknown"
        ),
        "event_type": result.get(
            "claim_type",
            "Unknown"
        ),
        "status": result.get(
            "status",
            "UNKNOWN"
        ),
        "reason": result.get(
            "reason",
            ""
        )
    }


# ============================================================
# 5. EVIDENCE UTILITIES
# ============================================================

def get_supporting_evidence(result):
    """
    Extract supporting evidence from the
    content-consistency check.
    """

    checks = result.get(
        "checks",
        {}
    )

    content_check = checks.get(
        "content_consistency",
        {}
    )

    details = content_check.get(
        "details",
        {}
    )

    return details.get(
        "supporting_candidates",
        []
    )


def format_evidence_timestamp(evidence):
    """
    Return an evidence timestamp in MM:SS format.
    """

    start = evidence.get(
        "start",
        0
    )

    end = evidence.get(
        "end",
        0
    )

    return (
        f"{format_timestamp(start)}"
        f"–"
        f"{format_timestamp(end)}"
    )


# ============================================================
# 6. CONSISTENCY CHECK UTILITIES
# ============================================================

CONSISTENCY_CHECKS = [
    (
        "content_consistency",
        "Content"
    ),
    (
        "speaker_consistency",
        "Speaker"
    ),
    (
        "timestamp_consistency",
        "Timestamp"
    ),
    (
        "event_type_consistency",
        "Event Type"
    )
]


def get_consistency_checks(result):
    """
    Return the four multi-attribute checks.
    """

    checks = result.get(
        "checks",
        {}
    )

    return [
        (
            key,
            display_name,
            checks.get(
                key,
                {}
            )
        )
        for key, display_name
        in CONSISTENCY_CHECKS
    ]


def is_timestamp_check(check_key):
    """
    Identify the timestamp consistency check.
    """

    return (
        check_key ==
        "timestamp_consistency"
    )


# ============================================================
# 7. SUMMARY UTILITIES
# ============================================================

def get_verification_summary(
    verification_data
):
    """
    Extract dashboard summary metrics.
    """

    return {
        "total_claims": verification_data.get(
            "total_claims",
            0
        ),
        "verified_claims": verification_data.get(
            "verified_claims",
            0
        ),
        "flagged_claims": verification_data.get(
            "flagged_claims",
            0
        ),
        "verification_rate": verification_data.get(
            "verification_rate",
            0.0
        )
    }


# ============================================================
# 8. PDF UTILITIES
# ============================================================

def get_pdf_claim_data(result):
    """
    Extract the information required by the
    PDF generator.
    """

    evidence_start = result.get(
        "evidence_start"
    )

    evidence_end = result.get(
        "evidence_end"
    )

    return {
        "claim": result.get(
            "claim",
            ""
        ),
        "speaker": result.get(
            "claim_speaker",
            "Unknown"
        ),
        "event_type": result.get(
            "claim_type",
            "Unknown"
        ),
        "status": result.get(
            "status",
            "UNKNOWN"
        ),
        "reason": result.get(
            "reason",
            ""
        ),
        "evidence_start": evidence_start,
        "evidence_end": evidence_end,
        "supporting_evidence": get_supporting_evidence(
            result
        )
    }
'''


with open(
    UTILS_PATH,
    "w",
    encoding="utf-8"
) as file:

    file.write(
        UTILS_CODE
    )


print(
    "✓ dashboard_utils.py created"
)

print(
    UTILS_PATH
)

print(
    "Size:",
    os.path.getsize(
        UTILS_PATH
    ),
    "bytes"
)
````

````python
# ============================================================
# STEP 1 — INSPECT EXISTING PIPELINE FUNCTIONS
# ============================================================

import json
import os
import re
from pathlib import Path

PROJECT_DIR = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project"
)

NOTEBOOKS = {
    "VAD": PROJECT_DIR / "01_audio_vad" / "01_silero_vad.ipynb",
    "ASR_DIARIZATION": PROJECT_DIR / "02_asr_diarization" / "02_asr_diarization.ipynb",
    "MOM_GENERATION": PROJECT_DIR / "04_mom_generation" / "04_mom_generation.ipynb",
}

print("=" * 90)
print("EXISTING PIPELINE FUNCTION INSPECTION")
print("=" * 90)

for name, path in NOTEBOOKS.items():

    print(f"\n{name}")
    print("-" * 90)
    print("Path:", path)
    print("Exists:", path.exists())

    if not path.exists():
        continue

    with open(path, "r", encoding="utf-8") as f:
        nb = json.load(f)

    print("Number of cells:", len(nb["cells"]))

    # Extract function definitions
    functions = []

    for cell_idx, cell in enumerate(nb["cells"]):

        if cell.get("cell_type") != "code":
            continue

        source = "".join(cell.get("source", []))

        matches = re.findall(
            r"^\s*(?:async\s+)?def\s+([A-Za-z_][A-Za-z0-9_]*)\s*\(",
            source,
            flags=re.MULTILINE
        )

        for func in matches:
            functions.append((func, cell_idx))

    print("\nFunctions found:")

    for func, cell_idx in functions:
        print(f"  {func:<45} cell {cell_idx}")

print("\n" + "=" * 90)
print("INSPECTION COMPLETE")
print("=" * 90)
````

````python
from pathlib import Path

PROJECT_DIR = Path("/content/drive/MyDrive/MTechIndProj/MoM_Project")
PIPELINE_FILE = PROJECT_DIR / "07_dashboard" / "meeting_pipeline.py"

pipeline_code = r'''
# ============================================================
# meeting_pipeline.py
# Multi-Attribute Evidence Consistency for Verifiable MoM
# ============================================================

from pathlib import Path
import json
import os
import torch

PROJECT_DIR = Path("/content/drive/MyDrive/MTechIndProj/MoM_Project")

DATA_DIR = PROJECT_DIR / "data"
MODEL_DIR = PROJECT_DIR / "models"
OUTPUT_DIR = PROJECT_DIR / "outputs"

RAW_DIR = DATA_DIR / "raw"
VAD_DIR = DATA_DIR / "vad"
TRANSCRIPT_DIR = DATA_DIR / "transcripts"
EVIDENCE_DIR = DATA_DIR / "evidence"
EMBEDDING_DIR = DATA_DIR / "embeddings"
FAISS_DIR = DATA_DIR / "faiss"

ASR_MODEL_DIR = MODEL_DIR / "asr"
DIARIZATION_MODEL_DIR = MODEL_DIR / "diarization"
BGE_MODEL_DIR = MODEL_DIR / "bge"
LLM_MODEL_DIR = MODEL_DIR / "llm"

VAD_THRESHOLD = 0.5
GAP_THRESHOLD_SECONDS = 1.0

WHISPER_MODEL_NAME = "small"
WHISPER_LANGUAGE = "en"

BGE_MODEL_NAME = "BAAI/bge-base-en-v1.5"
NLI_MODEL_NAME = "cross-encoder/nli-deberta-v3-small"

NLI_DEVICE = "cuda"

DEFAULT_TOP_K = 10
DEFAULT_CONTEXT_WINDOW = 2


def ensure_directories():
    directories = [
        DATA_DIR,
        MODEL_DIR,
        OUTPUT_DIR,
        RAW_DIR,
        VAD_DIR,
        TRANSCRIPT_DIR,
        EVIDENCE_DIR,
        EMBEDDING_DIR,
        FAISS_DIR,
        ASR_MODEL_DIR,
        DIARIZATION_MODEL_DIR,
        BGE_MODEL_DIR,
        LLM_MODEL_DIR,
    ]

    for directory in directories:
        directory.mkdir(parents=True, exist_ok=True)


def save_json(data, path):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)

    with open(path, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=2, ensure_ascii=False)


def load_json(path):
    path = Path(path)

    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)


ensure_directories()


# ============================================================
# Silero VAD
# ============================================================

def load_silero_vad():

    model, utils = torch.hub.load(
        repo_or_dir="snakers4/silero-vad",
        model="silero_vad",
        force_reload=False
    )

    return model, utils


def run_silero_vad(
    audio_path,
    output_path=None,
    threshold=VAD_THRESHOLD
):

    audio_path = Path(audio_path)

    if not audio_path.exists():
        raise FileNotFoundError(
            f"Audio file not found: {audio_path}"
        )

    model, utils = load_silero_vad()

    (
        get_speech_timestamps,
        save_audio,
        read_audio,
        VADIterator,
        collect_chunks
    ) = utils

    wav = read_audio(
        str(audio_path),
        sampling_rate=16000
    )

    speech_timestamps = get_speech_timestamps(
        wav,
        model,
        threshold=threshold,
        sampling_rate=16000
    )

    segments = []

    for segment in speech_timestamps:

        segments.append({
            "start": round(
                segment["start"] / 16000,
                3
            ),
            "end": round(
                segment["end"] / 16000,
                3
            )
        })

    result = {
        "audio_path": str(audio_path),
        "sampling_rate": 16000,
        "threshold": threshold,
        "num_segments": len(segments),
        "segments": segments
    }

    if output_path is not None:
        save_json(result, output_path)

    return result


# ============================================================
# WhisperX
# ============================================================

def load_whisperx_model(
    model_name=WHISPER_MODEL_NAME,
    device=None
):

    import whisperx

    if device is None:
        device = (
            "cuda"
            if torch.cuda.is_available()
            else "cpu"
        )

    if device == "cuda" and torch.cuda.is_available():
        compute_type = "float16"
    else:
        compute_type = "int8"

    print("Loading WhisperX ASR model...")
    print("Model:", model_name)
    print("Device:", device)
    print("Compute type:", compute_type)

    model = whisperx.load_model(
        model_name,
        device,
        compute_type=compute_type,
        language=WHISPER_LANGUAGE
    )

    print("WhisperX model loaded successfully.")

    return model


def transcribe_with_whisperx(
    audio_path,
    model=None
):

    import whisperx

    audio_path = Path(audio_path)

    if not audio_path.exists():
        raise FileNotFoundError(
            f"Audio file not found: {audio_path}"
        )

    if model is None:
        model = load_whisperx_model()

    audio = whisperx.load_audio(
        str(audio_path)
    )

    result = model.transcribe(
        audio,
        batch_size=16,
        language=WHISPER_LANGUAGE
    )

    return result, audio


def align_whisperx_transcription(
    transcription_result,
    audio,
    device=None
):

    import whisperx

    if device is None:
        device = (
            "cuda"
            if torch.cuda.is_available()
            else "cpu"
        )

    language_code = transcription_result.get(
        "language",
        WHISPER_LANGUAGE
    )

    print("Loading WhisperX alignment model...")
    print("Language:", language_code)

    align_model, metadata = whisperx.load_align_model(
        language_code=language_code,
        device=device
    )

    aligned_result = whisperx.align(
        transcription_result["segments"],
        align_model,
        metadata,
        audio,
        device,
        return_char_alignments=False
    )

    print("WhisperX alignment completed.")

    return aligned_result
'''

PIPELINE_FILE.write_text(
    pipeline_code,
    encoding="utf-8"
)

print("meeting_pipeline.py rewritten successfully.")
print("File:", PIPELINE_FILE)
print("Size:", PIPELINE_FILE.stat().st_size, "bytes")
````

````python
from pathlib import Path

PIPELINE_FILE = Path(
    "/content/drive/MyDrive/MTechIndProj/"
    "MoM_Project/07_dashboard/meeting_pipeline.py"
)

whisperx_code = r'''

# ============================================================
# WHISPERX ASR + WORD-LEVEL ALIGNMENT
# ============================================================

def load_whisperx_model(model_name=WHISPER_MODEL_NAME, device=None):
    import torch
    import whisperx

    if device is None:
        device = "cuda" if torch.cuda.is_available() else "cpu"

    compute_type = "float16" if device == "cuda" else "int8"

    model = whisperx.load_model(
        model_name,
        device=device,
        compute_type=compute_type,
        language="en"
    )

    print(f"WhisperX model loaded: {model_name}")
    print(f"Device: {device}")
    print(f"Compute type: {compute_type}")

    return model


def transcribe_with_whisperx(audio_path, model=None):
    import whisperx

    if model is None:
        model = load_whisperx_model()

    audio = whisperx.load_audio(str(audio_path))

    result = model.transcribe(
        audio,
        batch_size=8
    )

    print("WhisperX transcription completed.")
    print("Language:", result.get("language"))
    print("Transcript segments:", len(result.get("segments", [])))

    return result, audio


def align_whisperx_transcription(transcription_result, audio, device=None):
    import torch
    import whisperx

    if device is None:
        device = "cuda" if torch.cuda.is_available() else "cpu"

    language = transcription_result.get("language", "en")

    align_model, metadata = whisperx.load_align_model(
        language_code=language,
        device=device
    )

    aligned_result = whisperx.align(
        transcription_result["segments"],
        align_model,
        metadata,
        audio,
        device,
        return_char_alignments=False
    )

    print("WhisperX word-level alignment completed.")
    print(
        "Aligned segments:",
        len(aligned_result.get("segments", []))
    )

    return aligned_result


def run_whisperx_pipeline(audio_path, output_path=None):
    import torch

    device = "cuda" if torch.cuda.is_available() else "cpu"

    model = load_whisperx_model(
        model_name=WHISPER_MODEL_NAME,
        device=device
    )

    transcription_result, audio = transcribe_with_whisperx(
        audio_path,
        model=model
    )

    aligned_result = align_whisperx_transcription(
        transcription_result,
        audio,
        device=device
    )

    if output_path is not None:
        save_json(aligned_result, output_path)
        print(f"WhisperX output saved to: {output_path}")

    return aligned_result
'''

with open(PIPELINE_FILE, "a", encoding="utf-8") as f:
    f.write(whisperx_code)

print("WhisperX functions appended successfully.")
print("File exists:", PIPELINE_FILE.exists())
print("File size:", PIPELINE_FILE.stat().st_size, "bytes")
````

````python
# Generalized Pyannote function:
# ============================================================
# Add reusable Pyannote speaker diarization functions
# Uses in-memory audio loading to avoid TorchCodec dependency
# and supports Pyannote Audio 4.x DiarizeOutput format.
# ============================================================
from pathlib import Path
import os
import json
import torch
import soundfile as sf

def load_pyannote_pipeline(hf_token=None):
    """Load Pyannote speaker diarization pipeline."""
    from pyannote.audio import Pipeline

    if hf_token is None:
        hf_token = os.environ.get("HF_TOKEN")

    pipeline = Pipeline.from_pretrained(
        "pyannote/speaker-diarization-3.1",
        token=hf_token
    )

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    pipeline.to(device)

    return pipeline


def run_pyannote_diarization(
    audio_path,
    output_path=None,
    hf_token=None
):
    """
    Run Pyannote diarization using an in-memory waveform.

    Returns:
        list of dictionaries containing start, end and speaker.
    """

    pipeline = load_pyannote_pipeline(hf_token)

    waveform, sample_rate = sf.read(str(audio_path))

    if waveform.ndim == 1:
        waveform = waveform[None, :]
    else:
        waveform = waveform.T

    waveform = torch.tensor(
        waveform,
        dtype=torch.float32
    )

    diarization_output = pipeline({
        "waveform": waveform,
        "sample_rate": sample_rate
    })

    # Pyannote 4.x returns DiarizeOutput.
    # Exclusive diarization gives one speaker per time region.
    annotation = diarization_output.exclusive_speaker_diarization

    segments = []

    for turn, _, speaker in annotation.itertracks(
        yield_label=True
    ):
        segments.append({
            "start": round(float(turn.start), 3),
            "end": round(float(turn.end), 3),
            "speaker": speaker
        })

    result = {
        "audio_path": str(audio_path),
        "sample_rate": sample_rate,
        "num_segments": len(segments),
        "speakers": sorted(
            list(set(x["speaker"] for x in segments))
        ),
        "segments": segments
    }

    if output_path is not None:
        output_path = Path(output_path)
        output_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        with open(output_path, "w", encoding="utf-8") as f:
            json.dump(
                result,
                f,
                indent=2,
                ensure_ascii=False
            )

    return result
````

````python
# ============================================================
# Append reusable Pyannote speaker diarization functions
# Uses in-memory audio loading to avoid TorchCodec dependency
# and supports Pyannote Audio 4.x DiarizeOutput format.
# ============================================================

from pathlib import Path

pipeline_file = Path(
    "/content/drive/MyDrive/"
    "MTechIndProj/MoM_Project/07_dashboard/"
    "meeting_pipeline.py"
)

pyannote_code = r'''

# ============================================================
# Pyannote Speaker Diarization
# ============================================================

def load_pyannote_pipeline(hf_token=None):
    """Load Pyannote speaker diarization pipeline."""
    import os
    import torch
    from pyannote.audio import Pipeline

    if hf_token is None:
        hf_token = os.environ.get("HF_TOKEN")

    pipeline = Pipeline.from_pretrained(
        "pyannote/speaker-diarization-3.1",
        token=hf_token
    )

    device = torch.device(
        "cuda" if torch.cuda.is_available() else "cpu"
    )

    pipeline.to(device)

    return pipeline


def run_pyannote_diarization(
    audio_path,
    output_path=None,
    hf_token=None
):
    """
    Run Pyannote speaker diarization using
    an in-memory waveform.

    Returns:
        Dictionary containing diarization segments
        and detected speakers.
    """

    import json
    import torch
    import soundfile as sf

    pipeline = load_pyannote_pipeline(hf_token)

    waveform, sample_rate = sf.read(str(audio_path))

    # Convert audio to [channels, time].
    if waveform.ndim == 1:
        waveform = waveform[None, :]
    else:
        waveform = waveform.T

    waveform = torch.tensor(
        waveform,
        dtype=torch.float32
    )

    diarization_output = pipeline({
        "waveform": waveform,
        "sample_rate": sample_rate
    })

    # Pyannote Audio 4.x returns DiarizeOutput.
    # Exclusive diarization provides one speaker
    # for each time region.
    annotation = (
        diarization_output.exclusive_speaker_diarization
    )

    segments = []

    for turn, _, speaker in annotation.itertracks(
        yield_label=True
    ):
        segments.append({
            "start": round(float(turn.start), 3),
            "end": round(float(turn.end), 3),
            "speaker": speaker
        })

    result = {
        "audio_path": str(audio_path),
        "sample_rate": sample_rate,
        "num_segments": len(segments),
        "speakers": sorted(
            list(set(
                item["speaker"] for item in segments
            ))
        ),
        "segments": segments
    }

    if output_path is not None:
        output_path = Path(output_path)

        output_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        with open(
            output_path,
            "w",
            encoding="utf-8"
        ) as f:
            json.dump(
                result,
                f,
                indent=2,
                ensure_ascii=False
            )

        print(
            f"Pyannote output saved to: {output_path}"
        )

    return result
'''

with open(
    pipeline_file,
    "a",
    encoding="utf-8"
) as f:
    f.write("\n" + pyannote_code)

print("Pyannote functions appended successfully.")
print("File size:", pipeline_file.stat().st_size, "bytes")

# Confirm the functions are physically present in the file.
text = pipeline_file.read_text(encoding="utf-8")

print(
    "load_pyannote_pipeline saved:",
    "def load_pyannote_pipeline" in text
)

print(
    "run_pyannote_diarization saved:",
    "def run_pyannote_diarization" in text
)
````

````python
# ============================================================
# Add reusable speaker-word alignment functions
# Aligns WhisperX word timestamps with Pyannote speaker
# segments using temporal overlap and groups words into
# speaker-attributed utterances.
# ============================================================

from pathlib import Path

pipeline_file = Path(
    "/content/drive/MyDrive/"
    "MTechIndProj/MoM_Project/07_dashboard/"
    "meeting_pipeline.py"
)

speaker_alignment_code = r'''

# ============================================================
# Speaker–Word Alignment
# ============================================================

def calculate_temporal_overlap(
    word_start,
    word_end,
    speaker_start,
    speaker_end
):
    """Calculate temporal overlap between a word and speaker segment."""

    overlap_start = max(word_start, speaker_start)
    overlap_end = min(word_end, speaker_end)

    return max(0.0, overlap_end - overlap_start)


def assign_speaker_to_word(
    word_start,
    word_end,
    diarization_segments
):
    """Assign the speaker with the largest temporal overlap."""

    best_speaker = None
    best_overlap = 0.0

    for segment in diarization_segments:

        overlap = calculate_temporal_overlap(
            word_start,
            word_end,
            segment["start"],
            segment["end"]
        )

        if overlap > best_overlap:
            best_overlap = overlap
            best_speaker = segment["speaker"]

    return best_speaker, round(best_overlap, 3)


def align_words_with_speakers(
    whisper_data,
    pyannote_data
):
    """
    Align WhisperX word-level timestamps with Pyannote
    speaker diarization.
    """

    words = whisper_data.get("word_segments", [])

    if not words:
        for segment in whisper_data.get("segments", []):
            words.extend(segment.get("words", []))

    diarization_segments = pyannote_data.get(
        "segments",
        []
    )

    aligned_words = []

    for word in words:

        if "start" not in word or "end" not in word:
            continue

        speaker, overlap = assign_speaker_to_word(
            float(word["start"]),
            float(word["end"]),
            diarization_segments
        )

        aligned_words.append({
            "word": word["word"],
            "start": round(float(word["start"]), 3),
            "end": round(float(word["end"]), 3),
            "score": word.get("score"),
            "speaker": speaker,
            "speaker_overlap": overlap
        })

    return aligned_words


def group_words_into_speaker_utterances(
    aligned_words
):
    """Group consecutive speaker-attributed words into utterances."""

    if not aligned_words:
        return []

    utterances = []

    current_speaker = aligned_words[0]["speaker"]
    current_words = [aligned_words[0]]

    for word in aligned_words[1:]:

        if word["speaker"] == current_speaker:
            current_words.append(word)

        else:
            utterances.append({
                "speaker": current_speaker,
                "start": current_words[0]["start"],
                "end": current_words[-1]["end"],
                "text": " ".join(
                    item["word"]
                    for item in current_words
                ),
                "words": current_words
            })

            current_speaker = word["speaker"]
            current_words = [word]

    utterances.append({
        "speaker": current_speaker,
        "start": current_words[0]["start"],
        "end": current_words[-1]["end"],
        "text": " ".join(
            item["word"]
            for item in current_words
        ),
        "words": current_words
    })

    return utterances


def run_speaker_word_alignment(
    whisper_data,
    pyannote_data,
    output_path=None
):
    """
    Run complete speaker-word alignment.

    Returns:
        Dictionary containing aligned words and
        speaker-attributed utterances.
    """

    import json

    aligned_words = align_words_with_speakers(
        whisper_data,
        pyannote_data
    )

    utterances = group_words_into_speaker_utterances(
        aligned_words
    )

    result = {
        "num_words": len(aligned_words),
        "num_utterances": len(utterances),
        "speakers": sorted(
            list(set(
                word["speaker"]
                for word in aligned_words
                if word["speaker"] is not None
            ))
        ),
        "words": aligned_words,
        "utterances": utterances
    }

    if output_path is not None:

        output_path = Path(output_path)
        output_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        with open(
            output_path,
            "w",
            encoding="utf-8"
        ) as f:
            json.dump(
                result,
                f,
                indent=2,
                ensure_ascii=False
            )

        print(
            f"Speaker-word alignment saved to: {output_path}"
        )

    return result
'''

with open(
    pipeline_file,
    "a",
    encoding="utf-8"
) as f:
    f.write("\n" + speaker_alignment_code)

print("Speaker-word alignment functions appended successfully.")
print("File size:", pipeline_file.stat().st_size, "bytes")

text = pipeline_file.read_text(encoding="utf-8")

print(
    "run_speaker_word_alignment saved:",
    "def run_speaker_word_alignment" in text
)
````

````python
# ============================================================
# STEP 3 — Generalized Rule-Based MoM Candidate Filtering
# Purpose: Identify potentially important meeting utterances
#          for downstream structured MoM generation.
# ============================================================

import re
from pathlib import Path
import json

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

MIN_TEXT_LENGTH = 20

# Strong indicators of potentially important meeting content
ACTION_PATTERNS = [
    r"\bwe need to\b",
    r"\bwe have to\b",
    r"\bwe should\b",
    r"\bwe must\b",
    r"\bneed to\b",
    r"\bhas to\b",
    r"\bhave to\b",
    r"\bshould\b",
    r"\bmust\b",
    r"\bwill\b",
    r"\bgoing to\b",
    r"\blet's\b",
    r"\blet us\b",
    r"\bagree\b",
    r"\bagreed\b",
    r"\bdecide\b",
    r"\bdecided\b",
    r"\bdecision\b",
    r"\bassign\b",
    r"\bassigned\b",
    r"\bresponsible\b",
    r"\bdeadline\b",
    r"\bby next\b",
    r"\bby the\b",
    r"\bdue\b",
    r"\bcomplete\b",
    r"\bfinish\b",
    r"\bdeliver\b",
]

# Proposal / recommendation indicators
PROPOSAL_PATTERNS = [
    r"\bi suggest\b",
    r"\bi propose\b",
    r"\bwe could\b",
    r"\bwe can\b",
    r"\bmaybe we\b",
    r"\bhow about\b",
    r"\bwhy don't we\b",
    r"\bwhat if\b",
    r"\bperhaps\b",
    r"\brecommend\b",
    r"\bsuggest\b",
    r"\bpropose\b",
]

# Requirement / project-content indicators
REQUIREMENT_PATTERNS = [
    r"\brequirement\b",
    r"\bfeature\b",
    r"\bfunction\b",
    r"\bdesign\b",
    r"\bproject\b",
    r"\bproduct\b",
    r"\buser\b",
    r"\bcustomer\b",
    r"\bgoal\b",
    r"\bobjective\b",
    r"\btarget\b",
    r"\bplan\b",
    r"\bstrategy\b",
    r"\bproblem\b",
    r"\bsolution\b",
]

# Question patterns that may contain important decisions/issues
QUESTION_PATTERNS = [
    r"\bshould we\b",
    r"\bcan we\b",
    r"\bdo we\b",
    r"\bhow can we\b",
    r"\bwhat should\b",
    r"\bwhat do we\b",
]

# Conversational content that is unlikely to become an MoM item
LOW_VALUE_PATTERNS = [
    r"^\s*(okay|ok|yeah|yes|no|right|sure|thanks|thank you)\s*[.!?]*\s*$",
    r"\bhello everybody\b",
    r"\bgood morning\b",
    r"\bthank you\b",
    r"\bthanks\b",
    r"\bwho would like to\b",
    r"\bmy favourite\b",
    r"\bmy favorite\b",
    r"\bguess what\b",
]

# ------------------------------------------------------------
# Helper functions
# ------------------------------------------------------------

def normalize_text(text):
    """Normalize whitespace while preserving the original wording."""
    if not text:
        return ""
    return re.sub(r"\s+", " ", str(text)).strip()


def pattern_matches(text, patterns):
    """Return matching patterns for a text."""
    matches = []

    for pattern in patterns:
        if re.search(pattern, text, flags=re.IGNORECASE):
            matches.append(pattern)

    return matches


def classify_mom_candidate(utterance):
    """
    Assign a rule-based MoM candidate score.

    The score is only used for candidate selection.
    It is NOT a verification score.
    """

    text = normalize_text(utterance.get("text", ""))
    speaker = utterance.get("speaker")

    if speaker is None:
        speaker = "UNKNOWN"

    score = 0
    reasons = []

    # --------------------------------------------------------
    # Basic length filtering
    # --------------------------------------------------------

    if len(text) < MIN_TEXT_LENGTH:
        return {
            **utterance,
            "speaker": speaker,
            "mom_candidate": False,
            "candidate_score": 0,
            "candidate_reasons": ["too_short"],
        }

    # --------------------------------------------------------
    # Low-value conversational content
    # --------------------------------------------------------

    low_value_matches = pattern_matches(text, LOW_VALUE_PATTERNS)

    if low_value_matches:
        # Do not immediately discard long utterances.
        # Strong meeting-content signals can still override this.
        reasons.append("low_value_conversational_pattern")

    # --------------------------------------------------------
    # Strong action / decision signals
    # --------------------------------------------------------

    action_matches = pattern_matches(text, ACTION_PATTERNS)

    if action_matches:
        score += 3
        reasons.append("action_or_decision_signal")

    # --------------------------------------------------------
    # Proposal / recommendation signals
    # --------------------------------------------------------

    proposal_matches = pattern_matches(text, PROPOSAL_PATTERNS)

    if proposal_matches:
        score += 2
        reasons.append("proposal_or_recommendation_signal")

    # --------------------------------------------------------
    # Requirement / project signals
    # --------------------------------------------------------

    requirement_matches = pattern_matches(text, REQUIREMENT_PATTERNS)

    if requirement_matches:
        score += 2
        reasons.append("requirement_or_project_signal")

    # --------------------------------------------------------
    # Question / issue signals
    # --------------------------------------------------------

    question_matches = pattern_matches(text, QUESTION_PATTERNS)

    if question_matches:
        score += 1
        reasons.append("decision_question_signal")

    # --------------------------------------------------------
    # Longer utterances can contain multiple meeting points
    # --------------------------------------------------------

    word_count = len(text.split())

    if word_count >= 40:
        score += 1
        reasons.append("substantive_utterance")

    # --------------------------------------------------------
    # Penalize purely conversational content
    # --------------------------------------------------------

    if low_value_matches and score == 0:
        score = 0

    # --------------------------------------------------------
    # Candidate decision
    # --------------------------------------------------------

    # Score >= 2 → potentially useful for MoM generation
    is_candidate = score >= 2

    if not reasons:
        reasons.append("no_strong_mom_signal")

    return {
        **utterance,
        "speaker": speaker,
        "text": text,
        "mom_candidate": is_candidate,
        "candidate_score": score,
        "candidate_reasons": reasons,
    }


# ------------------------------------------------------------
# Apply filtering to all speaker-attributed utterances
# ------------------------------------------------------------

classified_utterances = [
    classify_mom_candidate(utt)
    for utt in utterances
]

mom_candidates = [
    utt
    for utt in classified_utterances
    if utt["mom_candidate"]
]

non_candidates = [
    utt
    for utt in classified_utterances
    if not utt["mom_candidate"]
]

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("=" * 80)
print("RULE-BASED MOM CANDIDATE FILTERING")
print("=" * 80)

print(f"Total utterances       : {len(classified_utterances)}")
print(f"MoM candidates         : {len(mom_candidates)}")
print(f"Filtered utterances    : {len(non_candidates)}")

if classified_utterances:
    candidate_percentage = (
        len(mom_candidates) / len(classified_utterances)
    ) * 100
else:
    candidate_percentage = 0

print(f"Candidate percentage   : {candidate_percentage:.2f}%")

# ------------------------------------------------------------
# Candidate score distribution
# ------------------------------------------------------------

score_distribution = {}

for utt in classified_utterances:
    score = utt["candidate_score"]
    score_distribution[score] = score_distribution.get(score, 0) + 1

print("\nCandidate score distribution:")

for score in sorted(score_distribution):
    print(f"Score {score}: {score_distribution[score]} utterances")

# ------------------------------------------------------------
# Display selected candidates
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("FIRST 20 MOM CANDIDATES")
print("=" * 80)

for i, utt in enumerate(mom_candidates[:20], start=1):

    speaker = utt["speaker"]
    start = utt.get("start", 0)
    end = utt.get("end", 0)
    score = utt["candidate_score"]
    reasons = ", ".join(utt["candidate_reasons"])
    text = utt["text"]

    print(
        f"\n[{i:02d}] "
        f"{speaker} | "
        f"{start:.2f}s → {end:.2f}s | "
        f"score={score}"
    )

    print(f"     {text}")
    print(f"     Reasons: {reasons}")
````

````python
# ============================================================
# STEP 4 — Refined MoM Candidate Filtering
# Purpose: Reduce conversational/activity false positives while
#          retaining decisions, actions, requirements, proposals,
#          assignments, and substantive project discussion.
# ============================================================

import re


# ------------------------------------------------------------
# Strong MoM signals
# ------------------------------------------------------------

STRONG_ACTION_PATTERNS = [
    r"\bwe need to\b",
    r"\bwe have to\b",
    r"\bwe should\b",
    r"\bwe must\b",
    r"\bneed to\b",
    r"\bhas to\b",
    r"\bhave to\b",
    r"\bshould\b",
    r"\bmust\b",
    r"\bagree\b",
    r"\bagreed\b",
    r"\bdecide\b",
    r"\bdecided\b",
    r"\bdecision\b",
    r"\bassign\b",
    r"\bassigned\b",
    r"\bresponsible for\b",
    r"\bdeadline\b",
    r"\bdue\b",
    r"\bby next\b",
    r"\bby the end\b",
]

PROPOSAL_PATTERNS = [
    r"\bi suggest\b",
    r"\bi propose\b",
    r"\bwe could\b",
    r"\bwe can\b",
    r"\bhow about\b",
    r"\bwhy don't we\b",
    r"\bwhat if\b",
    r"\brecommend\b",
    r"\bsuggest\b",
    r"\bpropose\b",
]

DECISION_PATTERNS = [
    r"\bshall we\b",
    r"\bshould we\b",
    r"\bdo we agree\b",
    r"\bwe agree\b",
    r"\blet's go with\b",
    r"\blet's use\b",
    r"\blet's do\b",
    r"\bwe'll use\b",
    r"\bwe will use\b",
]

REQUIREMENT_PATTERNS = [
    r"\brequirement\b",
    r"\bfeature\b",
    r"\bfunction\b",
    r"\bfunctional\b",
    r"\bdesign\b",
    r"\bproduct\b",
    r"\buser-friendly\b",
    r"\btarget group\b",
    r"\btarget market\b",
    r"\bgoal\b",
    r"\bobjective\b",
]

SUBSTANTIVE_PROJECT_PATTERNS = [
    r"\bremote control\b",
    r"\buser\b",
    r"\bcustomer\b",
    r"\bmarket\b",
    r"\bprice\b",
    r"\bcost\b",
    r"\bscreen\b",
    r"\blcd\b",
    r"\bbutton\b",
    r"\bmenu\b",
    r"\bdesign\b",
    r"\bfunction\b",
    r"\bfeature\b",
]


# ------------------------------------------------------------
# Activity / conversational false-positive patterns
# ------------------------------------------------------------

ACTIVITY_PATTERNS = [
    r"\bdrawing\b",
    r"\bdraw\b",
    r"\bdrawing skill\b",
    r"\bwhiteboard\b",
    r"\bpen\b",
    r"\bguess\b",
    r"\bguess what\b",
    r"\bmy favourite animal\b",
    r"\bmy favorite animal\b",
    r"\btiger\b",
    r"\bdog\b",
    r"\bcat\b",
    r"\beagle\b",
    r"\beagle\b",
    r"\bbat\b",
    r"\beagle\b",
    r"\bseagull\b",
    r"\banimal\b",
]

LOW_VALUE_PATTERNS = [
    r"^\s*(okay|ok|yeah|yes|no|right|sure|thanks|thank you)\s*[.!?]*\s*$",
    r"\bhello everybody\b",
    r"\bgood morning\b",
    r"\bwho would like to\b",
    r"\bspace for everyone\b",
]


def find_matches(text, patterns):
    """Return matched patterns."""
    matches = []

    for pattern in patterns:
        if re.search(pattern, text, flags=re.IGNORECASE):
            matches.append(pattern)

    return matches


def refined_mom_candidate(utterance):

    text = re.sub(
        r"\s+",
        " ",
        str(utterance.get("text", ""))
    ).strip()

    speaker = utterance.get("speaker") or "UNKNOWN"

    # Basic metadata
    result = {
        **utterance,
        "speaker": speaker,
        "text": text,
        "mom_candidate": False,
        "candidate_score": 0,
        "candidate_reasons": [],
    }

    if len(text) < 25:
        result["candidate_reasons"] = ["too_short"]
        return result

    # --------------------------------------------------------
    # Detect signal categories
    # --------------------------------------------------------

    action = find_matches(text, STRONG_ACTION_PATTERNS)
    proposal = find_matches(text, PROPOSAL_PATTERNS)
    decision = find_matches(text, DECISION_PATTERNS)
    requirement = find_matches(text, REQUIREMENT_PATTERNS)
    project = find_matches(text, SUBSTANTIVE_PROJECT_PATTERNS)

    activity = find_matches(text, ACTIVITY_PATTERNS)
    low_value = find_matches(text, LOW_VALUE_PATTERNS)

    score = 0
    reasons = []

    # Strong action / decision
    if action:
        score += 3
        reasons.append("action_or_decision")

    # Explicit proposal
    if proposal:
        score += 3
        reasons.append("proposal")

    # Explicit decision language
    if decision:
        score += 4
        reasons.append("explicit_decision")

    # Requirement
    if requirement:
        score += 2
        reasons.append("requirement")

    # Substantive project content
    if project:
        score += 1
        reasons.append("project_content")

    # --------------------------------------------------------
    # Activity penalty
    # --------------------------------------------------------

    if activity:
        score -= 3
        reasons.append("activity_or_small_talk")

    # --------------------------------------------------------
    # Very low-value conversational utterance
    # --------------------------------------------------------

    if low_value:
        score -= 5
        reasons.append("low_value_conversation")

    # --------------------------------------------------------
    # Long substantive utterance
    # --------------------------------------------------------

    word_count = len(text.split())

    if word_count >= 50 and project:
        score += 1
        reasons.append("substantive_length")

    # --------------------------------------------------------
    # Final candidate decision
    # --------------------------------------------------------

    # Explicit decisions/actions/proposals are allowed through
    # even when they contain some conversational wording.
    strong_signal = bool(action or proposal or decision)

    # Require either a strong signal + sufficient score,
    # or substantive project/requirement content.
    if score >= 3 and (strong_signal or requirement or project):
        result["mom_candidate"] = True

    result["candidate_score"] = max(score, 0)
    result["candidate_reasons"] = reasons

    return result


# ------------------------------------------------------------
# Re-run classification
# ------------------------------------------------------------

refined_classified_utterances = [
    refined_mom_candidate(utt)
    for utt in utterances
]

refined_candidates = [
    utt
    for utt in refined_classified_utterances
    if utt["mom_candidate"]
]

refined_non_candidates = [
    utt
    for utt in refined_classified_utterances
    if not utt["mom_candidate"]
]


# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("=" * 80)
print("REFINED MOM CANDIDATE FILTERING")
print("=" * 80)

print(f"Total utterances       : {len(refined_classified_utterances)}")
print(f"MoM candidates         : {len(refined_candidates)}")
print(f"Filtered utterances    : {len(refined_non_candidates)}")

percentage = (
    len(refined_candidates) / len(refined_classified_utterances) * 100
    if refined_classified_utterances else 0
)

print(f"Candidate percentage   : {percentage:.2f}%")


# ------------------------------------------------------------
# Display candidates
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("REFINED MOM CANDIDATES")
print("=" * 80)

for i, utt in enumerate(refined_candidates, start=1):

    print(
        f"\n[{i:02d}] "
        f"{utt['speaker']} | "
        f"{utt.get('start', 0):.2f}s → "
        f"{utt.get('end', 0):.2f}s | "
        f"score={utt['candidate_score']}"
    )

    print(f"     {utt['text']}")
    print(
        "     Reasons: "
        + ", ".join(utt["candidate_reasons"])
    )
````

````python
# ============================================================
# STEP 5 — Meeting-Relevance Refinement
# Purpose: Remove obvious activity/small-talk candidates and
#          retain propositions that can become MoM items.
# ============================================================

import re


# ------------------------------------------------------------
# Explicit activity / icebreaker vocabulary
# ------------------------------------------------------------

ACTIVITY_TERMS = [
    "draw",
    "drawing",
    "whiteboard",
    "pen",
    "guess",
    "tiger",
    "dog",
    "cat",
    "bat",
    "eagle",
    "seagull",
    "animal",
    "favourite animal",
    "favorite animal",
    "drawing skill",
]


# ------------------------------------------------------------
# Meeting outcome / task signals
# ------------------------------------------------------------

DECISION_PATTERNS = [
    r"\bwe decided\b",
    r"\bwe agreed\b",
    r"\bagreed to\b",
    r"\blet's go with\b",
    r"\blet's use\b",
    r"\blet's do\b",
    r"\bwe'll\b",
    r"\bwe will\b",
    r"\bwe're going to\b",
    r"\bthe decision is\b",
]


ACTION_PATTERNS = [
    r"\bwe need to\b",
    r"\bwe have to\b",
    r"\bwe should\b",
    r"\bwe must\b",
    r"\bneed to\b",
    r"\bhave to\b",
    r"\bhas to\b",
    r"\bassigned to\b",
    r"\bresponsible for\b",
    r"\bdeadline\b",
    r"\bdue\b",
    r"\bby next\b",
]


PROPOSAL_PATTERNS = [
    r"\bi suggest\b",
    r"\bi propose\b",
    r"\bwe could\b",
    r"\bwe can\b",
    r"\bhow about\b",
    r"\bwhy don't we\b",
    r"\bwhat if\b",
    r"\bmight be useful\b",
    r"\bcould have\b",
    r"\bcould be\b",
]


REQUIREMENT_PATTERNS = [
    r"\bit has to\b",
    r"\bit needs to\b",
    r"\bmust be\b",
    r"\bshould be\b",
    r"\bneed to be\b",
    r"\bhas to be\b",
    r"\baccessible\b",
    r"\busable\b",
    r"\buser-friendly\b",
    r"\btarget group\b",
    r"\btarget market\b",
    r"\brequirement\b",
    r"\bfeature\b",
]


# ------------------------------------------------------------
# Project-context vocabulary
# ------------------------------------------------------------

PROJECT_TERMS = [
    "remote control",
    "product",
    "design",
    "lcd",
    "screen",
    "menu",
    "button",
    "price",
    "cost",
    "market",
    "user",
    "customer",
    "feature",
    "function",
    "material",
    "plastic",
    "metal",
]


def contains_pattern(text, patterns):
    """Check whether any pattern occurs."""
    return any(
        re.search(pattern, text, flags=re.IGNORECASE)
        for pattern in patterns
    )


def contains_project_term(text):
    """Check whether text contains substantive project vocabulary."""
    text_lower = text.lower()

    return any(
        term in text_lower
        for term in PROJECT_TERMS
    )


def activity_only(text):
    """
    Identify utterances that are predominantly about the
    meeting activity/icebreaker rather than project content.
    """

    text_lower = text.lower()

    activity_hits = sum(
        1
        for term in ACTIVITY_TERMS
        if term in text_lower
    )

    project_content = contains_project_term(text)

    # Strong suppression when activity vocabulary dominates
    # and there is no project context.
    if activity_hits >= 1 and not project_content:
        return True

    return False


def classify_meeting_relevance(utterance):

    text = re.sub(
        r"\s+",
        " ",
        str(utterance.get("text", ""))
    ).strip()

    speaker = utterance.get("speaker") or "UNKNOWN"

    result = {
        **utterance,
        "speaker": speaker,
        "text": text,
        "mom_candidate": False,
        "candidate_type": None,
        "candidate_score": 0,
        "candidate_reasons": [],
    }

    if len(text.split()) < 6:
        result["candidate_reasons"] = ["too_short"]
        return result

    # --------------------------------------------------------
    # First suppress obvious non-meeting activity
    # --------------------------------------------------------

    if activity_only(text):
        result["candidate_reasons"] = [
            "activity_or_icebreaker_content"
        ]
        return result

    # --------------------------------------------------------
    # Detect meeting proposition type
    # --------------------------------------------------------

    is_decision = contains_pattern(
        text,
        DECISION_PATTERNS
    )

    is_action = contains_pattern(
        text,
        ACTION_PATTERNS
    )

    is_proposal = contains_pattern(
        text,
        PROPOSAL_PATTERNS
    )

    is_requirement = contains_pattern(
        text,
        REQUIREMENT_PATTERNS
    )

    has_project_context = contains_project_term(text)

    # --------------------------------------------------------
    # Priority:
    # Decision > Action > Requirement > Proposal
    # --------------------------------------------------------

    if is_decision:
        candidate_type = "DECISION"
        score = 5

    elif is_action and has_project_context:
        candidate_type = "ACTION"
        score = 4

    elif is_requirement and has_project_context:
        candidate_type = "REQUIREMENT"
        score = 3

    elif is_proposal and has_project_context:
        candidate_type = "PROPOSAL"
        score = 3

    elif has_project_context and len(text.split()) >= 30:
        # Substantive project discussion can still be useful.
        candidate_type = "DISCUSSION_POINT"
        score = 2

    else:
        candidate_type = None
        score = 0

    # --------------------------------------------------------
    # Store result
    # --------------------------------------------------------

    if candidate_type is not None:
        result["mom_candidate"] = True
        result["candidate_type"] = candidate_type
        result["candidate_score"] = score

        reasons = []

        if is_decision:
            reasons.append("decision_signal")

        if is_action:
            reasons.append("action_signal")

        if is_requirement:
            reasons.append("requirement_signal")

        if is_proposal:
            reasons.append("proposal_signal")

        if has_project_context:
            reasons.append("project_context")

        result["candidate_reasons"] = reasons

    return result


# ------------------------------------------------------------
# Run refined meeting-understanding filter
# ------------------------------------------------------------

meeting_understanding = [
    classify_meeting_relevance(utt)
    for utt in utterances
]

final_candidates = [
    utt
    for utt in meeting_understanding
    if utt["mom_candidate"]
]

print("=" * 80)
print("MEETING-RELEVANCE CANDIDATE FILTER")
print("=" * 80)

print(f"Total utterances : {len(meeting_understanding)}")
print(f"MoM candidates   : {len(final_candidates)}")

percentage = (
    len(final_candidates) / len(meeting_understanding) * 100
    if meeting_understanding else 0
)

print(f"Candidate rate   : {percentage:.2f}%")


# ------------------------------------------------------------
# Candidate-type distribution
# ------------------------------------------------------------

type_counts = {}

for utt in final_candidates:
    candidate_type = utt["candidate_type"]

    type_counts[candidate_type] = (
        type_counts.get(candidate_type, 0) + 1
    )

print("\nCandidate types:")

for candidate_type, count in sorted(type_counts.items()):
    print(f"{candidate_type}: {count}")


# ------------------------------------------------------------
# Display candidates
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("FINAL CANDIDATE REVIEW")
print("=" * 80)

for i, utt in enumerate(final_candidates, start=1):

    print(
        f"\n[{i:02d}] "
        f"{utt['speaker']} | "
        f"{utt.get('start', 0):.2f}s → "
        f"{utt.get('end', 0):.2f}s"
    )

    print(
        f"     Type: {utt['candidate_type']} | "
        f"Score: {utt['candidate_score']}"
    )

    print(f"     {utt['text']}")

    print(
        "     Reasons: "
        + ", ".join(utt["candidate_reasons"])
    )
````

````python
# ============================================================
# STEP 6 — Candidate Topic Grouping
# Purpose: Group related MoM candidate utterances into coherent
#          discussion topics before structured MoM generation.
#
# Method:
#   - Use BGE embeddings already used by the project
#   - Compare candidate utterances by cosine similarity
#   - Group nearby/related discussion points
#   - Preserve speaker and timestamp provenance
# ============================================================

import numpy as np
import torch
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity


# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

EMBEDDING_MODEL_NAME = "BAAI/bge-base-en-v1.5"

# Similarity threshold for grouping related utterances.
# We start conservatively and inspect the result.
TOPIC_SIMILARITY_THRESHOLD = 0.55


# ------------------------------------------------------------
# Load embedding model
# ------------------------------------------------------------

print("=" * 80)
print("LOADING BGE EMBEDDING MODEL")
print("=" * 80)

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Device:", device)
print("Model :", EMBEDDING_MODEL_NAME)

topic_embedding_model = SentenceTransformer(
    EMBEDDING_MODEL_NAME,
    device=device
)


# ------------------------------------------------------------
# Prepare candidate texts
# ------------------------------------------------------------

candidate_texts = [
    utt["text"]
    for utt in final_candidates
]

print("\nCandidates to group:", len(candidate_texts))


# ------------------------------------------------------------
# Generate embeddings
# ------------------------------------------------------------

candidate_embeddings = topic_embedding_model.encode(
    candidate_texts,
    normalize_embeddings=True,
    convert_to_numpy=True,
    show_progress_bar=True
)

print(
    "Embedding matrix:",
    candidate_embeddings.shape
)


# ------------------------------------------------------------
# Calculate pairwise similarity
# ------------------------------------------------------------

similarity_matrix = cosine_similarity(
    candidate_embeddings
)


# ------------------------------------------------------------
# Simple connected-component grouping
#
# If two candidates are sufficiently similar, they belong
# to the same discussion group. Transitive connections are
# retained so that a discussion can span several utterances.
# ------------------------------------------------------------

n = len(final_candidates)

visited = set()
topic_groups = []


for i in range(n):

    if i in visited:
        continue

    group_indices = []
    queue = [i]
    visited.add(i)

    while queue:

        current = queue.pop(0)
        group_indices.append(current)

        for j in range(n):

            if j in visited:
                continue

            if similarity_matrix[current, j] >= TOPIC_SIMILARITY_THRESHOLD:
                visited.add(j)
                queue.append(j)

    topic_groups.append(group_indices)


# ------------------------------------------------------------
# Convert groups into structured objects
# ------------------------------------------------------------

grouped_candidates = []

for group_id, indices in enumerate(topic_groups, start=1):

    group_items = []

    for idx in sorted(indices):

        utt = final_candidates[idx]

        group_items.append({
            "candidate_index": idx + 1,
            "speaker": utt.get("speaker", "UNKNOWN"),
            "start": utt.get("start"),
            "end": utt.get("end"),
            "candidate_type": utt.get("candidate_type"),
            "score": utt.get("candidate_score"),
            "text": utt.get("text", ""),
        })

    # Group time span
    valid_starts = [
        x["start"]
        for x in group_items
        if x["start"] is not None
    ]

    valid_ends = [
        x["end"]
        for x in group_items
        if x["end"] is not None
    ]

    grouped_candidates.append({
        "group_id": group_id,
        "num_candidates": len(group_items),
        "start": min(valid_starts) if valid_starts else None,
        "end": max(valid_ends) if valid_ends else None,
        "candidates": group_items,
    })


# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("TOPIC GROUPING SUMMARY")
print("=" * 80)

print("Input candidates :", len(final_candidates))
print("Topic groups     :", len(grouped_candidates))

group_sizes = [
    group["num_candidates"]
    for group in grouped_candidates
]

print("Group sizes      :", group_sizes)


# ------------------------------------------------------------
# Display grouped candidates
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("GROUPED DISCUSSION TOPICS")
print("=" * 80)

for group in grouped_candidates:

    print(
        f"\n{'-' * 80}\n"
        f"GROUP {group['group_id']} | "
        f"{group['num_candidates']} candidate(s) | "
        f"{group['start']:.2f}s → {group['end']:.2f}s"
    )

    for item in group["candidates"]:

        print(
            f"\n  [{item['candidate_index']:02d}] "
            f"{item['speaker']} | "
            f"{item['candidate_type']}"
        )

        print(f"       {item['text']}")
````

````python
# ============================================================
# STEP 7 — Temporal + Semantic Candidate Grouping
# Purpose: Prevent distant meeting topics from being merged
#          solely because they have moderate semantic similarity.
# ============================================================

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

# Maximum temporal gap between candidate utterances that can
# directly belong to the same discussion group.
MAX_TIME_GAP = 120.0  # seconds

# Semantic similarity threshold.
SEMANTIC_THRESHOLD = 0.60


# ------------------------------------------------------------
# Build pairwise grouping relationships
# ------------------------------------------------------------

n = len(final_candidates)

adjacency = {
    i: set()
    for i in range(n)
}


for i in range(n):

    start_i = final_candidates[i].get("start", 0)
    end_i = final_candidates[i].get("end", start_i)

    for j in range(i + 1, n):

        start_j = final_candidates[j].get("start", 0)
        end_j = final_candidates[j].get("end", start_j)

        # ----------------------------------------------------
        # Temporal distance
        # ----------------------------------------------------

        if start_j >= end_i:
            temporal_gap = start_j - end_i

        elif start_i >= end_j:
            temporal_gap = start_i - end_j

        else:
            temporal_gap = 0.0

        # ----------------------------------------------------
        # Semantic similarity
        # ----------------------------------------------------

        semantic_similarity = similarity_matrix[i, j]

        # ----------------------------------------------------
        # Group only when BOTH conditions hold
        # ----------------------------------------------------

        if (
            temporal_gap <= MAX_TIME_GAP
            and semantic_similarity >= SEMANTIC_THRESHOLD
        ):
            adjacency[i].add(j)
            adjacency[j].add(i)


# ------------------------------------------------------------
# Connected components
# ------------------------------------------------------------

visited = set()
temporal_topic_groups = []


for i in range(n):

    if i in visited:
        continue

    queue = [i]
    visited.add(i)
    group = []

    while queue:

        current = queue.pop(0)
        group.append(current)

        for neighbour in adjacency[current]:

            if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighbour)

    temporal_topic_groups.append(sorted(group))


# ------------------------------------------------------------
# Convert to structured groups
# ------------------------------------------------------------

temporal_grouped_candidates = []

for group_id, indices in enumerate(
    temporal_topic_groups,
    start=1
):

    items = []

    for idx in indices:

        utt = final_candidates[idx]

        items.append({
            "candidate_index": idx + 1,
            "speaker": utt.get("speaker", "UNKNOWN"),
            "start": utt.get("start"),
            "end": utt.get("end"),
            "candidate_type": utt.get("candidate_type"),
            "score": utt.get("candidate_score"),
            "text": utt.get("text", ""),
        })

    starts = [
        x["start"]
        for x in items
        if x["start"] is not None
    ]

    ends = [
        x["end"]
        for x in items
        if x["end"] is not None
    ]

    temporal_grouped_candidates.append({
        "group_id": group_id,
        "num_candidates": len(items),
        "start": min(starts) if starts else None,
        "end": max(ends) if ends else None,
        "candidates": items,
    })


# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("=" * 80)
print("TEMPORAL + SEMANTIC TOPIC GROUPING")
print("=" * 80)

print("Input candidates :", len(final_candidates))
print("Topic groups     :", len(temporal_grouped_candidates))

print(
    "Group sizes      :",
    [
        g["num_candidates"]
        for g in temporal_grouped_candidates
    ]
)


# ------------------------------------------------------------
# Display groups
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("GROUPED TOPICS")
print("=" * 80)

for group in temporal_grouped_candidates:

    print(
        f"\n{'-' * 80}\n"
        f"GROUP {group['group_id']} | "
        f"{group['num_candidates']} candidate(s) | "
        f"{group['start']:.2f}s → {group['end']:.2f}s"
    )

    for item in group["candidates"]:

        print(
            f"\n  [{item['candidate_index']:02d}] "
            f"{item['speaker']} | "
            f"{item['candidate_type']}"
        )

        print(f"       {item['text']}")
````

````python
# ============================================================
# STEP 8 — Fine Semantic Topic Grouping
# Purpose: Split broad temporal discussion windows into
#          semantically coherent MoM topics.
#
# Important:
#   - Does NOT modify any files.
#   - Preserves candidate provenance.
#   - Uses semantic similarity only WITHIN each temporal group.
# ============================================================

from sklearn.metrics.pairwise import cosine_similarity
import numpy as np


# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

FINE_SIMILARITY_THRESHOLD = 0.62


# ------------------------------------------------------------
# Fine grouping function
# ------------------------------------------------------------

def fine_semantic_group(items, embeddings, indices):
    """
    Group candidates within a coarse temporal window using
    semantic similarity.

    indices contains the original candidate indices.
    """

    if len(indices) <= 1:
        return [indices]

    local_embeddings = embeddings[indices]

    local_similarity = cosine_similarity(
        local_embeddings
    )

    n_local = len(indices)

    adjacency = {
        i: set()
        for i in range(n_local)
    }

    for i in range(n_local):
        for j in range(i + 1, n_local):

            if (
                local_similarity[i, j]
                >= FINE_SIMILARITY_THRESHOLD
            ):
                adjacency[i].add(j)
                adjacency[j].add(i)

    # --------------------------------------------------------
    # Connected components
    # --------------------------------------------------------

    visited = set()
    groups = []

    for i in range(n_local):

        if i in visited:
            continue

        queue = [i]
        visited.add(i)
        local_group = []

        while queue:

            current = queue.pop(0)
            local_group.append(
                indices[current]
            )

            for neighbour in adjacency[current]:

                if neighbour not in visited:
                    visited.add(neighbour)
                    queue.append(neighbour)

        groups.append(sorted(local_group))

    return groups


# ------------------------------------------------------------
# Convert coarse groups into candidate-index groups
# ------------------------------------------------------------

fine_topic_groups = []

for coarse_group in temporal_grouped_candidates:

    candidate_indices = [
        item["candidate_index"] - 1
        for item in coarse_group["candidates"]
    ]

    subgroups = fine_semantic_group(
        coarse_group["candidates"],
        candidate_embeddings,
        candidate_indices
    )

    fine_topic_groups.extend(subgroups)


# ------------------------------------------------------------
# Build final fine-grained topic objects
# ------------------------------------------------------------

fine_grouped_candidates = []

for group_id, indices in enumerate(
    fine_topic_groups,
    start=1
):

    items = []

    for idx in indices:

        utt = final_candidates[idx]

        items.append({
            "candidate_index": idx + 1,
            "speaker": utt.get(
                "speaker",
                "UNKNOWN"
            ),
            "start": utt.get("start"),
            "end": utt.get("end"),
            "candidate_type": utt.get(
                "candidate_type"
            ),
            "score": utt.get(
                "candidate_score"
            ),
            "text": utt.get("text", ""),
        })

    starts = [
        x["start"]
        for x in items
        if x["start"] is not None
    ]

    ends = [
        x["end"]
        for x in items
        if x["end"] is not None
    ]

    fine_grouped_candidates.append({
        "group_id": group_id,
        "num_candidates": len(items),
        "start": min(starts) if starts else None,
        "end": max(ends) if ends else None,
        "candidates": items,
    })


# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("=" * 80)
print("FINE SEMANTIC TOPIC GROUPING")
print("=" * 80)

print(
    "Input candidates :",
    len(final_candidates)
)

print(
    "Fine topic groups:",
    len(fine_grouped_candidates)
)

print(
    "Group sizes      :",
    [
        group["num_candidates"]
        for group in fine_grouped_candidates
    ]
)


# ------------------------------------------------------------
# Display fine-grained groups
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("FINE-GRAINED TOPIC GROUPS")
print("=" * 80)

for group in fine_grouped_candidates:

    print(
        f"\n{'-' * 80}\n"
        f"GROUP {group['group_id']} | "
        f"{group['num_candidates']} candidate(s) | "
        f"{group['start']:.2f}s → "
        f"{group['end']:.2f}s"
    )

    for item in group["candidates"]:

        print(
            f"\n  [{item['candidate_index']:02d}] "
            f"{item['speaker']} | "
            f"{item['candidate_type']}"
        )

        print(
            f"       {item['text']}"
        )
````

````python
# ============================================================
# STEP 9 — Candidate Filtering v2
# Purpose: Keep project-relevant meeting propositions while
#          suppressing icebreakers, activity narration, and
#          generic meeting administration.
# ============================================================

import re


# ------------------------------------------------------------
# Project-domain vocabulary
# ------------------------------------------------------------

PROJECT_TERMS = [
    "remote control",
    "product",
    "design",
    "functional",
    "conceptual",
    "feature",
    "button",
    "screen",
    "lcd",
    "menu",
    "display",
    "plastic",
    "metal",
    "material",
    "price",
    "cost",
    "production",
    "market",
    "customer",
    "user",
    "user-friendly",
    "accessible",
    "usable",
    "international",
    "target",
    "recording",
    "volume",
    "channel",
    "programming",
    "program",
]


# ------------------------------------------------------------
# Explicit non-project/activity vocabulary
# ------------------------------------------------------------

ACTIVITY_TERMS = [
    "draw",
    "drawing",
    "whiteboard",
    "pen",
    "favourite animal",
    "favorite animal",
    "tiger",
    "dog",
    "cat",
    "eagle",
    "bat",
    "seagull",
]


# ------------------------------------------------------------
# Strong project proposition patterns
# ------------------------------------------------------------

DECISION_PATTERNS = [
    r"\bwe decided\b",
    r"\bwe agreed\b",
    r"\bagreed to\b",
    r"\blet's use\b",
    r"\blet's go with\b",
    r"\bwe will use\b",
    r"\bwe'll use\b",
]

ACTION_PATTERNS = [
    r"\bwe need to\b",
    r"\bwe have to\b",
    r"\bwe should\b",
    r"\bwe must\b",
    r"\bneed to be\b",
    r"\bhas to be\b",
    r"\bhave to be\b",
    r"\bassigned to\b",
    r"\bresponsible for\b",
    r"\bdeadline\b",
    r"\bdue\b",
]

PROPOSAL_PATTERNS = [
    r"\bi suggest\b",
    r"\bi propose\b",
    r"\bwe could\b",
    r"\bwe can\b",
    r"\bhow about\b",
    r"\bwhat if\b",
    r"\bmight be useful\b",
    r"\bcould have\b",
    r"\bcould be\b",
]


def contains_any(text, terms):
    """Check whether any term occurs in the text."""
    text_lower = text.lower()

    return any(
        term.lower() in text_lower
        for term in terms
    )


def pattern_match(text, patterns):
    """Check whether any regex pattern occurs."""
    return any(
        re.search(
            pattern,
            text,
            flags=re.IGNORECASE
        )
        for pattern in patterns
    )


def project_term_count(text):
    """Count distinct project-domain terms present."""
    text_lower = text.lower()

    return sum(
        1
        for term in PROJECT_TERMS
        if term.lower() in text_lower
    )


def refine_candidate_v2(utterance):

    text = re.sub(
        r"\s+",
        " ",
        str(utterance.get("text", ""))
    ).strip()

    speaker = utterance.get("speaker") or "UNKNOWN"

    result = {
        **utterance,
        "speaker": speaker,
        "text": text,
        "mom_candidate": False,
        "candidate_type": None,
        "candidate_score": 0,
        "candidate_reasons": [],
    }

    # --------------------------------------------------------
    # Basic filtering
    # --------------------------------------------------------

    if len(text.split()) < 6:
        result["candidate_reasons"] = ["too_short"]
        return result

    # --------------------------------------------------------
    # Detect project relevance
    # --------------------------------------------------------

    project_count = project_term_count(text)
    has_project_context = project_count >= 1

    # --------------------------------------------------------
    # Detect activity / icebreaker content
    # --------------------------------------------------------

    activity_present = contains_any(
        text,
        ACTIVITY_TERMS
    )

    # If activity language is present and there is no
    # meaningful project context, reject it.
    if activity_present and project_count == 0:

        result["candidate_reasons"] = [
            "non_project_activity"
        ]

        return result

    # --------------------------------------------------------
    # Detect proposition types
    # --------------------------------------------------------

    is_decision = pattern_match(
        text,
        DECISION_PATTERNS
    )

    is_action = pattern_match(
        text,
        ACTION_PATTERNS
    )

    is_proposal = pattern_match(
        text,
        PROPOSAL_PATTERNS
    )

    # --------------------------------------------------------
    # Meeting administration
    # --------------------------------------------------------

    admin_terms = [
        "next meeting",
        "before the next meeting",
        "go off",
        "come back",
        "presentation",
        "agenda",
        "first meeting",
        "tool training",
        "individual work",
    ]

    administrative = contains_any(
        text,
        admin_terms
    )

    # Administrative content is retained only when there
    # is also a clear project/task signal.
    if administrative and not (
        has_project_context
        and (is_action or is_decision or is_proposal)
    ):

        result["candidate_reasons"] = [
            "meeting_administration"
        ]

        return result

    # --------------------------------------------------------
    # Classify project-relevant content
    # --------------------------------------------------------

    candidate_type = None
    score = 0
    reasons = []

    if is_decision and has_project_context:

        candidate_type = "DECISION"
        score = 5
        reasons.append("project_decision")

    elif is_action and has_project_context:

        candidate_type = "ACTION"
        score = 4
        reasons.append("project_action")

    elif is_proposal and has_project_context:

        candidate_type = "PROPOSAL"
        score = 4
        reasons.append("project_proposal")

    elif has_project_context and project_count >= 2:

        candidate_type = "DISCUSSION_POINT"
        score = 3
        reasons.append("substantive_project_discussion")

    elif has_project_context:

        candidate_type = "PROJECT_CONTEXT"
        score = 2
        reasons.append("project_context")

    # --------------------------------------------------------
    # Additional evidence of substantive discussion
    # --------------------------------------------------------

    word_count = len(text.split())

    if word_count >= 25 and has_project_context:

        score += 1
        reasons.append("substantive_length")

    # --------------------------------------------------------
    # Final decision
    # --------------------------------------------------------

    if candidate_type is not None and score >= 3:

        result["mom_candidate"] = True
        result["candidate_type"] = candidate_type
        result["candidate_score"] = score
        result["candidate_reasons"] = reasons

    else:

        result["candidate_reasons"] = (
            reasons
            if reasons
            else ["insufficient_project_relevance"]
        )

    return result


# ------------------------------------------------------------
# Apply Candidate Filtering v2
# ------------------------------------------------------------

candidates_v2 = [
    refine_candidate_v2(utt)
    for utt in utterances
]

final_candidates_v2 = [
    utt
    for utt in candidates_v2
    if utt["mom_candidate"]
]


# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("=" * 80)
print("CANDIDATE FILTERING V2")
print("=" * 80)

print(
    "Total utterances :",
    len(candidates_v2)
)

print(
    "MoM candidates   :",
    len(final_candidates_v2)
)

candidate_rate = (
    len(final_candidates_v2)
    / len(candidates_v2)
    * 100
    if candidates_v2
    else 0
)

print(
    f"Candidate rate   : {candidate_rate:.2f}%"
)


# ------------------------------------------------------------
# Candidate type distribution
# ------------------------------------------------------------

type_counts = {}

for utt in final_candidates_v2:

    candidate_type = utt["candidate_type"]

    type_counts[candidate_type] = (
        type_counts.get(candidate_type, 0)
        + 1
    )

print("\nCandidate types:")

for candidate_type, count in sorted(
    type_counts.items()
):
    print(
        f"{candidate_type}: {count}"
    )


# ------------------------------------------------------------
# Display candidates
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CANDIDATES V2")
print("=" * 80)

for i, utt in enumerate(
    final_candidates_v2,
    start=1
):

    print(
        f"\n[{i:02d}] "
        f"{utt['speaker']} | "
        f"{utt.get('start', 0):.2f}s → "
        f"{utt.get('end', 0):.2f}s"
    )

    print(
        f"     Type: {utt['candidate_type']} | "
        f"Score: {utt['candidate_score']}"
    )

    print(
        f"     {utt['text']}"
    )

    print(
        "     Reasons: "
        + ", ".join(
            utt["candidate_reasons"]
        )
    )
````

````python
# ============================================================
# STEP 10 — Final Candidate Filtering V3
# Purpose: Finalize the meeting-relevant candidate pool by
#          explicitly removing icebreakers, activity narration,
#          and generic meeting administration.
#
# This remains a candidate-extraction stage only.
# It is NOT evidence verification.
# ============================================================

import re


# ------------------------------------------------------------
# Explicit non-MoM activity/context terms
# ------------------------------------------------------------

NON_MOM_ACTIVITY_TERMS = [
    "draw",
    "drawing",
    "whiteboard",
    "pen",
    "favourite animal",
    "favorite animal",
    "tiger",
    "dog",
    "cat",
    "eagle",
    "bat",
    "seagull",
    "guess",
    "drawing skill",
    "sum up the characteristics of that animal",
]


# ------------------------------------------------------------
# Meeting-introduction / agenda indicators
# ------------------------------------------------------------

MEETING_INTRO_TERMS = [
    "first meeting",
    "hello everybody",
    "our agenda",
    "get to know each other",
    "feel more comfortable",
    "tool training",
    "we've got 25 minutes",
    "we're going to practice",
    "space for everyone else",
]


# ------------------------------------------------------------
# Explicit project proposition patterns
# ------------------------------------------------------------

DECISION_PATTERNS = [
    r"\bwe decided\b",
    r"\bwe agreed\b",
    r"\bagreed to\b",
    r"\blet's use\b",
    r"\blet's go with\b",
    r"\bwe will use\b",
    r"\bwe'll use\b",
]


ACTION_PATTERNS = [
    r"\bwe need to\b",
    r"\bwe have to\b",
    r"\bwe should\b",
    r"\bwe must\b",
    r"\bneed to be\b",
    r"\bhas to be\b",
    r"\bhave to be\b",
    r"\bassigned to\b",
    r"\bresponsible for\b",
    r"\bdeadline\b",
    r"\bdue\b",
]


PROPOSAL_PATTERNS = [
    r"\bi suggest\b",
    r"\bi propose\b",
    r"\bwe could\b",
    r"\bwe can\b",
    r"\bhow about\b",
    r"\bwhat if\b",
    r"\bmight be useful\b",
    r"\bcould have\b",
    r"\bcould be\b",
]


# ------------------------------------------------------------
# Project vocabulary
# ------------------------------------------------------------

PROJECT_TERMS = [
    "remote control",
    "product",
    "design",
    "functional",
    "conceptual",
    "feature",
    "button",
    "screen",
    "lcd",
    "menu",
    "display",
    "plastic",
    "metal",
    "material",
    "price",
    "cost",
    "production",
    "market",
    "customer",
    "user",
    "user-friendly",
    "accessible",
    "usable",
    "international",
    "target",
    "recording",
    "volume",
    "channel",
    "programming",
]


def has_term(text, terms):
    """Return True when any term occurs."""
    text_lower = text.lower()

    return any(
        term.lower() in text_lower
        for term in terms
    )


def matches_pattern(text, patterns):
    """Return True when any regex pattern matches."""
    return any(
        re.search(
            pattern,
            text,
            flags=re.IGNORECASE
        )
        for pattern in patterns
    )


def count_project_terms(text):
    """Count distinct project terms."""
    text_lower = text.lower()

    return sum(
        1
        for term in PROJECT_TERMS
        if term.lower() in text_lower
    )


def classify_candidate_v3(utterance):

    text = re.sub(
        r"\s+",
        " ",
        str(utterance.get("text", ""))
    ).strip()

    speaker = utterance.get("speaker") or "UNKNOWN"

    result = {
        **utterance,
        "speaker": speaker,
        "text": text,
        "mom_candidate": False,
        "candidate_type": None,
        "candidate_score": 0,
        "candidate_reasons": [],
    }

    # --------------------------------------------------------
    # Basic length
    # --------------------------------------------------------

    if len(text.split()) < 6:
        result["candidate_reasons"] = ["too_short"]
        return result

    # --------------------------------------------------------
    # Explicit activity rejection
    # --------------------------------------------------------

    activity_present = has_term(
        text,
        NON_MOM_ACTIVITY_TERMS
    )

    project_count = count_project_terms(text)

    if activity_present and project_count == 0:

        result["candidate_reasons"] = [
            "non_mom_activity"
        ]

        return result

    # --------------------------------------------------------
    # Explicit meeting-introduction rejection
    # --------------------------------------------------------

    intro_present = has_term(
        text,
        MEETING_INTRO_TERMS
    )

    decision = matches_pattern(
        text,
        DECISION_PATTERNS
    )

    action = matches_pattern(
        text,
        ACTION_PATTERNS
    )

    proposal = matches_pattern(
        text,
        PROPOSAL_PATTERNS
    )

    # Long introduction containing project vocabulary should
    # not automatically become an MoM candidate.
    #
    # Only retain it if it contains an explicit project
    # decision/action/proposal AND is not predominantly
    # introductory/activity content.
    if intro_present and not (
        decision or action or proposal
    ):

        result["candidate_reasons"] = [
            "meeting_introduction"
        ]

        return result

    # --------------------------------------------------------
    # Project relevance
    # --------------------------------------------------------

    has_project_context = project_count >= 1

    if not has_project_context:

        result["candidate_reasons"] = [
            "no_project_context"
        ]

        return result

    # --------------------------------------------------------
    # Classify proposition
    # --------------------------------------------------------

    candidate_type = None
    score = 0
    reasons = []

    if decision:

        candidate_type = "DECISION"
        score = 5
        reasons.append("decision_signal")

    elif action:

        candidate_type = "ACTION"
        score = 4
        reasons.append("action_signal")

    elif proposal:

        candidate_type = "PROPOSAL"
        score = 4
        reasons.append("proposal_signal")

    elif project_count >= 2:

        candidate_type = "DISCUSSION_POINT"
        score = 3
        reasons.append("substantive_project_content")

    else:

        candidate_type = "PROJECT_CONTEXT"
        score = 2
        reasons.append("project_context")

    # --------------------------------------------------------
    # Length supports substantive discussion
    # --------------------------------------------------------

    if len(text.split()) >= 25:

        score += 1
        reasons.append("substantive_length")

    # --------------------------------------------------------
    # Final threshold
    # --------------------------------------------------------

    if score >= 3:

        result["mom_candidate"] = True
        result["candidate_type"] = candidate_type
        result["candidate_score"] = score
        result["candidate_reasons"] = reasons

    else:

        result["candidate_reasons"] = [
            "insufficient_mom_signal"
        ]

    return result


# ------------------------------------------------------------
# Apply V3
# ------------------------------------------------------------

candidates_v3 = [
    classify_candidate_v3(utt)
    for utt in utterances
]

final_candidates_v3 = [
    utt
    for utt in candidates_v3
    if utt["mom_candidate"]
]


# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

print("=" * 80)
print("FINAL CANDIDATE FILTERING V3")
print("=" * 80)

print(
    "Total utterances :",
    len(candidates_v3)
)

print(
    "MoM candidates   :",
    len(final_candidates_v3)
)

rate = (
    len(final_candidates_v3)
    / len(candidates_v3)
    * 100
    if candidates_v3
    else 0
)

print(
    f"Candidate rate   : {rate:.2f}%"
)


# ------------------------------------------------------------
# Candidate type distribution
# ------------------------------------------------------------

type_counts = {}

for utt in final_candidates_v3:

    candidate_type = utt["candidate_type"]

    type_counts[candidate_type] = (
        type_counts.get(candidate_type, 0)
        + 1
    )

print("\nCandidate types:")

for candidate_type, count in sorted(
    type_counts.items()
):
    print(
        f"{candidate_type}: {count}"
    )


# ------------------------------------------------------------
# Candidate review
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("FINAL CANDIDATE REVIEW")
print("=" * 80)

for i, utt in enumerate(
    final_candidates_v3,
    start=1
):

    print(
        f"\n[{i:02d}] "
        f"{utt['speaker']} | "
        f"{utt.get('start', 0):.2f}s → "
        f"{utt.get('end', 0):.2f}s"
    )

    print(
        f"     Type: {utt['candidate_type']} | "
        f"Score: {utt['candidate_score']}"
    )

    print(
        f"     {utt['text']}"
    )

    print(
        "     Reasons: "
        + ", ".join(
            utt["candidate_reasons"]
        )
    )
````

````python
# ============================================================
# STEP 11 — STRUCTURED MoM GENERATION INTERFACE
# Supports:
#   1. BART baseline
#   2. Qwen2.5-3B-Instruct final generator
# ============================================================

import json
import os
import re
from pathlib import Path


# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

GENERATOR_CONFIG = {
    "bart": {
        "model_name": "facebook/bart-large-cnn",
        "role": "baseline",
    },
    "qwen": {
        "model_name": "Qwen/Qwen2.5-3B-Instruct",
        "role": "final",
    },
}


# ------------------------------------------------------------
# Structured claim format
# ------------------------------------------------------------

def create_structured_claim(
    text,
    speaker=None,
    start=None,
    end=None,
    event_type=None,
    source_utterances=None,
):
    """
    Create a standardized MoM claim structure.

    This format will be used by both BART and Qwen
    generation paths and later by evidence verification.
    """

    return {
        "claim": str(text).strip(),

        "speaker": speaker,

        "start": start,

        "end": end,

        "event_type": event_type,

        "source_utterances": (
            source_utterances
            if source_utterances is not None
            else []
        ),

        "generation_model": None,

        "verification": None,
    }


# ------------------------------------------------------------
# Candidate preparation
# ------------------------------------------------------------

def prepare_candidate_context(candidates):
    """
    Convert filtered MoM candidates into a compact context
    suitable for structured generation.
    """

    context_blocks = []

    for idx, candidate in enumerate(
        candidates,
        start=1
    ):

        text = str(
            candidate.get("text", "")
        ).strip()

        if not text:
            continue

        speaker = (
            candidate.get("speaker")
            or "UNKNOWN"
        )

        start = candidate.get(
            "start",
            None
        )

        end = candidate.get(
            "end",
            None
        )

        candidate_type = (
            candidate.get(
                "candidate_type"
            )
            or "DISCUSSION_POINT"
        )

        block = (
            f"[Candidate {idx}]\n"
            f"Speaker: {speaker}\n"
            f"Time: {start} - {end}\n"
            f"Type: {candidate_type}\n"
            f"Text: {text}"
        )

        context_blocks.append(block)

    return "\n\n".join(
        context_blocks
    )


# ------------------------------------------------------------
# BART baseline generation
# ------------------------------------------------------------

def generate_mom_bart(
    candidates,
    model=None,
    tokenizer=None,
):
    """
    BART baseline interface.

    The existing BART implementation can be connected here
    without changing the downstream pipeline.
    """

    context = prepare_candidate_context(
        candidates
    )

    if not context:
        return []

    # --------------------------------------------------------
    # If model/tokenizer are not supplied, return the prepared
    # context rather than silently loading a large model.
    # --------------------------------------------------------

    if model is None or tokenizer is None:

        return {
            "generator": "bart",
            "model_name": GENERATOR_CONFIG[
                "bart"
            ]["model_name"],
            "status": "context_ready",
            "context": context,
        }

    # --------------------------------------------------------
    # Actual BART generation
    # --------------------------------------------------------

    inputs = tokenizer(
        context,
        return_tensors="pt",
        truncation=True,
        max_length=1024,
    )

    output_ids = model.generate(
        **inputs,
        max_length=256,
        min_length=40,
        num_beams=4,
        early_stopping=True,
    )

    generated_text = tokenizer.decode(
        output_ids[0],
        skip_special_tokens=True,
    )

    return {
        "generator": "bart",
        "model_name": GENERATOR_CONFIG[
            "bart"
        ]["model_name"],
        "text": generated_text,
        "context": context,
    }


# ------------------------------------------------------------
# Qwen generation interface
# ------------------------------------------------------------

def generate_mom_qwen(
    candidates,
    model=None,
    tokenizer=None,
):
    """
    Qwen2.5-3B-Instruct generation interface.

    The model will be connected after the interface is
    validated. This prevents unnecessary model loading.
    """

    context = prepare_candidate_context(
        candidates
    )

    if not context:
        return []

    # --------------------------------------------------------
    # Do not load Qwen automatically at this stage.
    # --------------------------------------------------------

    if model is None or tokenizer is None:

        return {
            "generator": "qwen",
            "model_name": GENERATOR_CONFIG[
                "qwen"
            ]["model_name"],
            "status": "context_ready",
            "context": context,
        }

    # --------------------------------------------------------
    # Structured generation prompt
    # --------------------------------------------------------

    system_prompt = """
You generate structured meeting minutes from
speaker-attributed meeting evidence.

For each meaningful meeting claim:

1. Preserve the factual meaning.
2. Identify the event type:
   ACTION, DECISION, PROPOSAL, REQUIREMENT,
   or DISCUSSION_POINT.
3. Preserve the responsible speaker when supported.
4. Do not invent information.
5. Return concise, evidence-grounded claims.
6. Do not merge unrelated topics.

Return JSON only.
""".strip()

    user_prompt = f"""
Meeting candidate evidence:

{context}

Return a JSON array using this structure:

[
  {{
    "claim": "...",
    "speaker": "...",
    "event_type": "...",
    "start": null,
    "end": null
  }}
]
""".strip()

    messages = [
        {
            "role": "system",
            "content": system_prompt,
        },
        {
            "role": "user",
            "content": user_prompt,
        },
    ]

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=4096,
    )

    output_ids = model.generate(
        **inputs,
        max_new_tokens=512,
        do_sample=False,
    )

    generated_ids = output_ids[
        0
    ][
        inputs["input_ids"].shape[1]:
    ]

    generated_text = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True,
    )

    return {
        "generator": "qwen",
        "model_name": GENERATOR_CONFIG[
            "qwen"
        ]["model_name"],
        "text": generated_text,
        "context": context,
    }


# ------------------------------------------------------------
# Unified generator
# ------------------------------------------------------------

def generate_structured_mom(
    candidates,
    generator="bart",
    model=None,
    tokenizer=None,
):
    """
    Unified generation entry point.

    generator:
        "bart" → baseline
        "qwen" → final
    """

    generator = generator.lower().strip()

    if generator == "bart":

        return generate_mom_bart(
            candidates,
            model=model,
            tokenizer=tokenizer,
        )

    if generator == "qwen":

        return generate_mom_qwen(
            candidates,
            model=model,
            tokenizer=tokenizer,
        )

    raise ValueError(
        f"Unsupported generator: {generator}. "
        f"Use 'bart' or 'qwen'."
    )


print("Structured MoM generation interface loaded.")
print("Available generators:")
print("  - bart : baseline")
print("  - qwen : final generator")
````

````python
# ============================================================
# STEP 11E — INTEGRATE MoM CANDIDATE FILTERING + GENERATION
# INTO meeting_pipeline.py
#
# This copies the validated logic into the generalized
# processing engine. Existing pipeline code is preserved.
# ============================================================

from pathlib import Path

PROJECT_DIR = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project"
)

PIPELINE_FILE = (
    PROJECT_DIR
    / "07_dashboard"
    / "meeting_pipeline.py"
)

print("=" * 80)
print("STEP 11E — PIPELINE INTEGRATION")
print("=" * 80)

print("Pipeline file:")
print(PIPELINE_FILE)

if not PIPELINE_FILE.exists():
    raise FileNotFoundError(
        f"meeting_pipeline.py not found:\n{PIPELINE_FILE}"
    )

pipeline_code = PIPELINE_FILE.read_text(
    encoding="utf-8"
)

print(
    "\nCurrent pipeline size:",
    len(pipeline_code),
    "characters"
)


# ============================================================
# 1. Candidate filtering V3
# ============================================================

candidate_filter_code = r'''

# ============================================================
# MEETING UNDERSTANDING — MoM CANDIDATE FILTERING V3
#
# Purpose:
# Select meeting-relevant utterances before structured
# MoM generation.
#
# Important:
# - This is candidate filtering, NOT final verification.
# - Explicit activity/icebreaker content is rejected.
# - Generic meeting introduction is rejected.
# - Project discussion, proposals, actions and decisions
#   are retained.
# ============================================================

import re


NON_MOM_ACTIVITY_TERMS = [
    "draw",
    "drawing",
    "whiteboard",
    "pen",
    "favourite animal",
    "favorite animal",
    "tiger",
    "dog",
    "cat",
    "eagle",
    "bat",
    "seagull",
    "guess",
    "drawing skill",
    "sum up the characteristics of that animal",
]


MEETING_INTRO_TERMS = [
    "first meeting",
    "hello everybody",
    "our agenda",
    "get to know each other",
    "feel more comfortable",
    "tool training",
    "we've got 25 minutes",
    "we're going to practice",
    "space for everyone else",
]


MOM_DECISION_PATTERNS = [
    r"\bwe decided\b",
    r"\bwe agreed\b",
    r"\bagreed to\b",
    r"\blet's use\b",
    r"\blet's go with\b",
    r"\bwe will use\b",
    r"\bwe'll use\b",
]


MOM_ACTION_PATTERNS = [
    r"\bwe need to\b",
    r"\bwe have to\b",
    r"\bwe should\b",
    r"\bwe must\b",
    r"\bneed to be\b",
    r"\bhas to be\b",
    r"\bhave to be\b",
    r"\bassigned to\b",
    r"\bresponsible for\b",
    r"\bdeadline\b",
    r"\bdue\b",
]


MOM_PROPOSAL_PATTERNS = [
    r"\bi suggest\b",
    r"\bi propose\b",
    r"\bwe could\b",
    r"\bwe can\b",
    r"\bhow about\b",
    r"\bwhat if\b",
    r"\bmight be useful\b",
    r"\bcould have\b",
    r"\bcould be\b",
]


MOM_PROJECT_TERMS = [
    "remote control",
    "product",
    "design",
    "functional",
    "conceptual",
    "feature",
    "button",
    "screen",
    "lcd",
    "menu",
    "display",
    "plastic",
    "metal",
    "material",
    "price",
    "cost",
    "production",
    "market",
    "customer",
    "user",
    "user-friendly",
    "accessible",
    "usable",
    "international",
    "target",
    "recording",
    "volume",
    "channel",
    "programming",
]


def _mom_has_term(text, terms):
    text_lower = str(text).lower()

    return any(
        term.lower() in text_lower
        for term in terms
    )


def _mom_matches_pattern(text, patterns):
    return any(
        re.search(
            pattern,
            str(text),
            flags=re.IGNORECASE
        )
        for pattern in patterns
    )


def _mom_count_project_terms(text):
    text_lower = str(text).lower()

    return sum(
        1
        for term in MOM_PROJECT_TERMS
        if term.lower() in text_lower
    )


def classify_mom_candidate_v3(utterance):

    text = re.sub(
        r"\s+",
        " ",
        str(utterance.get("text", ""))
    ).strip()

    speaker = (
        utterance.get("speaker")
        or "UNKNOWN"
    )

    result = {
        **utterance,
        "speaker": speaker,
        "text": text,
        "mom_candidate": False,
        "candidate_type": None,
        "candidate_score": 0,
        "candidate_reasons": [],
    }

    if len(text.split()) < 6:

        result["candidate_reasons"] = [
            "too_short"
        ]

        return result

    activity_present = _mom_has_term(
        text,
        NON_MOM_ACTIVITY_TERMS
    )

    project_count = _mom_count_project_terms(
        text
    )

    if activity_present and project_count == 0:

        result["candidate_reasons"] = [
            "non_mom_activity"
        ]

        return result

    intro_present = _mom_has_term(
        text,
        MEETING_INTRO_TERMS
    )

    decision = _mom_matches_pattern(
        text,
        MOM_DECISION_PATTERNS
    )

    action = _mom_matches_pattern(
        text,
        MOM_ACTION_PATTERNS
    )

    proposal = _mom_matches_pattern(
        text,
        MOM_PROPOSAL_PATTERNS
    )

    if intro_present and not (
        decision
        or action
        or proposal
    ):

        result["candidate_reasons"] = [
            "meeting_introduction"
        ]

        return result

    if project_count == 0:

        result["candidate_reasons"] = [
            "no_project_context"
        ]

        return result

    candidate_type = None
    score = 0
    reasons = []

    if decision:

        candidate_type = "DECISION"
        score = 5
        reasons.append("decision_signal")

    elif action:

        candidate_type = "ACTION"
        score = 4
        reasons.append("action_signal")

    elif proposal:

        candidate_type = "PROPOSAL"
        score = 4
        reasons.append("proposal_signal")

    elif project_count >= 2:

        candidate_type = "DISCUSSION_POINT"
        score = 3
        reasons.append(
            "substantive_project_content"
        )

    else:

        candidate_type = "PROJECT_CONTEXT"
        score = 2
        reasons.append(
            "project_context"
        )

    if len(text.split()) >= 25:

        score += 1
        reasons.append(
            "substantive_length"
        )

    if score >= 3:

        result["mom_candidate"] = True
        result["candidate_type"] = candidate_type
        result["candidate_score"] = score
        result["candidate_reasons"] = reasons

    else:

        result["candidate_reasons"] = [
            "insufficient_mom_signal"
        ]

    return result


def filter_mom_candidates_v3(
    utterances,
    return_all=False
):
    """
    Apply V3 meeting-relevance filtering.

    Parameters
    ----------
    utterances : list
        Speaker-attributed meeting utterances.

    return_all : bool
        If True, return both filtered and rejected items.

    Returns
    -------
    list or dict
        Final MoM candidates, optionally with all results.
    """

    classified = [
        classify_mom_candidate_v3(
            utterance
        )
        for utterance in utterances
    ]

    candidates = [
        item
        for item in classified
        if item["mom_candidate"]
    ]

    if return_all:

        return {
            "candidates": candidates,
            "all_items": classified,
        }

    return candidates


# ============================================================
# STRUCTURED MoM GENERATION INTERFACE
#
# BART = baseline
# Qwen = final generator
# ============================================================

GENERATOR_CONFIG = {
    "bart": {
        "model_name": "facebook/bart-large-cnn",
        "role": "baseline",
    },

    "qwen": {
        "model_name": "Qwen/Qwen2.5-3B-Instruct",
        "role": "final",
    },
}


def create_structured_claim(
    text,
    speaker=None,
    start=None,
    end=None,
    event_type=None,
    source_utterances=None,
):
    """
    Standard claim representation used by downstream
    evidence retrieval and verification.
    """

    return {
        "claim": str(text).strip(),
        "speaker": speaker,
        "start": start,
        "end": end,
        "event_type": event_type,
        "source_utterances": (
            source_utterances
            if source_utterances is not None
            else []
        ),
        "generation_model": None,
        "verification": None,
    }


def prepare_candidate_context(candidates):
    """
    Convert candidate utterances into a compact context
    for a generation model.
    """

    context_blocks = []

    for idx, candidate in enumerate(
        candidates,
        start=1
    ):

        text = str(
            candidate.get("text", "")
        ).strip()

        if not text:
            continue

        speaker = (
            candidate.get("speaker")
            or "UNKNOWN"
        )

        start = candidate.get(
            "start",
            None
        )

        end = candidate.get(
            "end",
            None
        )

        candidate_type = (
            candidate.get(
                "candidate_type"
            )
            or "DISCUSSION_POINT"
        )

        context_blocks.append(
            f"[Candidate {idx}]\n"
            f"Speaker: {speaker}\n"
            f"Time: {start} - {end}\n"
            f"Type: {candidate_type}\n"
            f"Text: {text}"
        )

    return "\n\n".join(
        context_blocks
    )


def generate_mom_bart(
    candidates,
    model=None,
    tokenizer=None,
):
    """
    BART baseline generation.

    The model is loaded only when explicitly supplied.
    """

    context = prepare_candidate_context(
        candidates
    )

    if not context:
        return []

    if model is None or tokenizer is None:

        return {
            "generator": "bart",
            "model_name": GENERATOR_CONFIG[
                "bart"
            ]["model_name"],
            "status": "context_ready",
            "context": context,
        }

    inputs = tokenizer(
        context,
        return_tensors="pt",
        truncation=True,
        max_length=1024,
    )

    # Keep tensors on the same device as the model.
    device = next(
        model.parameters()
    ).device

    inputs = {
        key: value.to(device)
        for key, value in inputs.items()
    }

    output_ids = model.generate(
        **inputs,
        max_length=256,
        min_length=40,
        num_beams=4,
        early_stopping=True,
    )

    generated_text = tokenizer.decode(
        output_ids[0],
        skip_special_tokens=True,
    )

    return {
        "generator": "bart",
        "model_name": GENERATOR_CONFIG[
            "bart"
        ]["model_name"],
        "text": generated_text,
        "context": context,
    }


def generate_mom_qwen(
    candidates,
    model=None,
    tokenizer=None,
):
    """
    Qwen2.5-3B-Instruct structured generation interface.

    Model loading is intentionally external so the dashboard
    does not load a large model during module import.
    """

    context = prepare_candidate_context(
        candidates
    )

    if not context:
        return []

    if model is None or tokenizer is None:

        return {
            "generator": "qwen",
            "model_name": GENERATOR_CONFIG[
                "qwen"
            ]["model_name"],
            "status": "context_ready",
            "context": context,
        }

    system_prompt = """
You generate structured meeting minutes from
speaker-attributed meeting evidence.

For each meaningful meeting claim:

1. Preserve the factual meaning.
2. Identify the event type:
   ACTION, DECISION, PROPOSAL, REQUIREMENT,
   or DISCUSSION_POINT.
3. Preserve the responsible speaker when supported.
4. Do not invent information.
5. Return concise, evidence-grounded claims.
6. Do not merge unrelated topics.

Return JSON only.
""".strip()

    user_prompt = f"""
Meeting candidate evidence:

{context}

Return a JSON array using this structure:

[
  {{
    "claim": "...",
    "speaker": "...",
    "event_type": "...",
    "start": null,
    "end": null
  }}
]
""".strip()

    messages = [
        {
            "role": "system",
            "content": system_prompt,
        },
        {
            "role": "user",
            "content": user_prompt,
        },
    ]

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=4096,
    )

    device = next(
        model.parameters()
    ).device

    inputs = {
        key: value.to(device)
        for key, value in inputs.items()
    }

    output_ids = model.generate(
        **inputs,
        max_new_tokens=512,
        do_sample=False,
    )

    generated_ids = output_ids[
        0
    ][
        inputs["input_ids"].shape[1]:
    ]

    generated_text = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True,
    )

    return {
        "generator": "qwen",
        "model_name": GENERATOR_CONFIG[
            "qwen"
        ]["model_name"],
        "text": generated_text,
        "context": context,
    }


def generate_structured_mom(
    candidates,
    generator="bart",
    model=None,
    tokenizer=None,
):
    """
    Unified MoM generation entry point.

    generator:
        bart -> baseline
        qwen -> final structured generator
    """

    generator = (
        str(generator)
        .lower()
        .strip()
    )

    if generator == "bart":

        return generate_mom_bart(
            candidates,
            model=model,
            tokenizer=tokenizer,
        )

    if generator == "qwen":

        return generate_mom_qwen(
            candidates,
            model=model,
            tokenizer=tokenizer,
        )

    raise ValueError(
        f"Unsupported generator: {generator}. "
        f"Use 'bart' or 'qwen'."
    )

'''

# ============================================================
# 2. Avoid accidental duplicate insertion
# ============================================================

if "def filter_mom_candidates_v3(" in pipeline_code:
    print(
        "\n⚠ V3 candidate filtering already exists "
        "in meeting_pipeline.py."
    )

else:

    with open(
        PIPELINE_FILE,
        "a",
        encoding="utf-8"
    ) as f:

        f.write("\n\n")
        f.write(candidate_filter_code)

    print(
        "\n✓ V3 candidate filtering integrated."
    )
    print(
        "✓ BART/Qwen generation interface integrated."
    )


# ============================================================
# 3. Reload and verify definitions
# ============================================================

updated_code = PIPELINE_FILE.read_text(
    encoding="utf-8"
)

required_functions = [
    "classify_mom_candidate_v3",
    "filter_mom_candidates_v3",
    "prepare_candidate_context",
    "generate_mom_bart",
    "generate_mom_qwen",
    "generate_structured_mom",
]

print("\n" + "=" * 80)
print("INTEGRATION CHECK")
print("=" * 80)

for function_name in required_functions:

    present = (
        f"def {function_name}"
        in updated_code
    )

    print(
        f"{'✓' if present else '✗'} "
        f"{function_name}"
    )

print(
    "\nUpdated pipeline size:",
    len(updated_code),
    "characters"
)

print("\nStep 11E integration complete.")
````

````python
# ============================================================
# STEP 11M — QWEN STRUCTURED GENERATION TEST
# ============================================================

print("=" * 80)
print("STEP 11M — QWEN STRUCTURED GENERATION TEST")
print("=" * 80)

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

QWEN_MODEL_NAME = "Qwen/Qwen2.5-3B-Instruct"

# ------------------------------------------------------------
# Load Qwen
# ------------------------------------------------------------

print("\nLoading Qwen2.5-3B-Instruct...")

qwen_tokenizer = AutoTokenizer.from_pretrained(
    QWEN_MODEL_NAME
)

qwen_model = AutoModelForCausalLM.from_pretrained(
    QWEN_MODEL_NAME,
    torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32,
    device_map="auto" if torch.cuda.is_available() else None
)

qwen_model.eval()

print("\n✓ Qwen loaded")
print("Model :", QWEN_MODEL_NAME)
print(
    "Device:",
    next(qwen_model.parameters()).device
)

# ------------------------------------------------------------
# Prepare candidate context
# ------------------------------------------------------------

qwen_context = meeting_pipeline.prepare_candidate_context(
    pipeline_v3_candidates
)

print("\nCandidate count:", len(pipeline_v3_candidates))
print("Context length :", len(qwen_context))

# ------------------------------------------------------------
# Structured generation prompt
# ------------------------------------------------------------

qwen_prompt = f"""
You are generating structured Minutes of Meeting (MoM)
claims from meeting transcript candidates.

Use ONLY the information present in the candidates.

For each meaningful meeting claim, return a JSON object
with exactly these fields:

- claim
- speaker
- event_type
- start
- end

Allowed event_type values:
ACTION
DECISION
DISCUSSION_POINT
PROPOSAL
REQUIREMENT
PROJECT_CONTEXT

Rules:
1. Do not invent information.
2. Preserve the speaker associated with the evidence.
3. Preserve the original start and end timestamps.
4. Exclude irrelevant conversation such as introductions,
   jokes, personal stories, or unrelated activities.
5. Combine closely related candidate statements when they
   describe the same meeting issue.
6. Keep claims concise and factual.
7. Return ONLY a valid JSON array.
8. Do not include Markdown fences.
9. Do not include explanations outside the JSON array.

Meeting candidates:

{qwen_context}
"""

# ------------------------------------------------------------
# Generate
# ------------------------------------------------------------

messages = [
    {
        "role": "system",
        "content": (
            "You are a precise meeting-minutes extraction "
            "assistant. Output valid JSON only."
        ),
    },
    {
        "role": "user",
        "content": qwen_prompt,
    },
]

formatted_prompt = qwen_tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True,
)

inputs = qwen_tokenizer(
    formatted_prompt,
    return_tensors="pt",
    truncation=True,
    max_length=8192,
)

device = next(qwen_model.parameters()).device

inputs = {
    key: value.to(device)
    for key, value in inputs.items()
}

print("\nGenerating structured MoM...")

with torch.no_grad():
    output_ids = qwen_model.generate(
        **inputs,
        max_new_tokens=1200,
        do_sample=False,
        temperature=0.0,
        pad_token_id=qwen_tokenizer.eos_token_id,
    )

# ------------------------------------------------------------
# Decode only newly generated tokens
# ------------------------------------------------------------

input_length = inputs["input_ids"].shape[1]

generated_ids = output_ids[
    0,
    input_length:
]

qwen_raw_output = qwen_tokenizer.decode(
    generated_ids,
    skip_special_tokens=True,
).strip()

# ------------------------------------------------------------
# Display result
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("QWEN RAW OUTPUT")
print("=" * 80)

print(qwen_raw_output)

print("\n" + "=" * 80)
print("STEP 11M COMPLETE")
print("=" * 80)

print("\nOutput characters:", len(qwen_raw_output))
````

````python
# ============================================================
# STEP 11N — VALIDATE QWEN STRUCTURED OUTPUT
# ============================================================

print("=" * 80)
print("STEP 11N — QWEN STRUCTURED OUTPUT VALIDATION")
print("=" * 80)

import json
import re

# ------------------------------------------------------------
# Parse Qwen JSON
# ------------------------------------------------------------

raw_output = qwen_raw_output.strip()

# Remove accidental markdown fences if the model ever adds them
clean_output = re.sub(
    r"^```(?:json)?\s*|\s*```$",
    "",
    raw_output,
    flags=re.IGNORECASE | re.DOTALL
).strip()

try:
    qwen_claims = json.loads(clean_output)
except json.JSONDecodeError as e:
    print("\n✗ JSON parsing failed")
    print("Error:", e)
    raise

print("\n✓ JSON parsing successful")
print("Generated claims:", len(qwen_claims))

# ------------------------------------------------------------
# Basic schema validation
# ------------------------------------------------------------

required_fields = {
    "claim",
    "speaker",
    "event_type",
    "start",
    "end",
}

allowed_event_types = {
    "ACTION",
    "DECISION",
    "DISCUSSION_POINT",
    "PROPOSAL",
    "REQUIREMENT",
    "PROJECT_CONTEXT",
}

schema_errors = []

for i, claim in enumerate(qwen_claims, start=1):

    if not isinstance(claim, dict):
        schema_errors.append(
            f"Claim {i}: not a JSON object"
        )
        continue

    missing = required_fields - set(claim.keys())

    if missing:
        schema_errors.append(
            f"Claim {i}: missing {sorted(missing)}"
        )

    if claim.get("event_type") not in allowed_event_types:
        schema_errors.append(
            f"Claim {i}: invalid event_type "
            f"{claim.get('event_type')}"
        )

    if not isinstance(claim.get("speaker"), str):
        schema_errors.append(
            f"Claim {i}: invalid speaker"
        )

    if not isinstance(claim.get("start"), (int, float)):
        schema_errors.append(
            f"Claim {i}: invalid start"
        )

    if not isinstance(claim.get("end"), (int, float)):
        schema_errors.append(
            f"Claim {i}: invalid end"
        )

    if (
        isinstance(claim.get("start"), (int, float))
        and isinstance(claim.get("end"), (int, float))
        and claim["end"] < claim["start"]
    ):
        schema_errors.append(
            f"Claim {i}: end < start"
        )

# ------------------------------------------------------------
# Display schema result
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("SCHEMA VALIDATION")
print("=" * 80)

if schema_errors:
    for error in schema_errors:
        print("✗", error)
else:
    print("✓ All generated claims satisfy the required schema")

# ------------------------------------------------------------
# Compare generated metadata against source candidates
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("SOURCE CANDIDATE CONSISTENCY")
print("=" * 80)

metadata_results = []

for i, claim in enumerate(qwen_claims, start=1):

    generated_speaker = claim["speaker"]
    generated_start = float(claim["start"])
    generated_end = float(claim["end"])

    best_candidate = None
    best_overlap = 0.0

    for candidate in pipeline_v3_candidates:

        candidate_speaker = candidate.get("speaker")

        candidate_start = float(
            candidate.get("start", 0)
        )

        candidate_end = float(
            candidate.get("end", 0)
        )

        overlap_start = max(
            generated_start,
            candidate_start
        )

        overlap_end = min(
            generated_end,
            candidate_end
        )

        overlap = max(
            0.0,
            overlap_end - overlap_start
        )

        candidate_duration = max(
            candidate_end - candidate_start,
            1e-9
        )

        overlap_ratio = (
            overlap / candidate_duration
        )

        if overlap_ratio > best_overlap:
            best_overlap = overlap_ratio
            best_candidate = candidate

    if best_candidate is None:

        metadata_results.append({
            "claim_index": i,
            "speaker_match": False,
            "timestamp_match": False,
            "event_type_match": False,
            "overlap_ratio": 0.0,
            "candidate": None,
        })

        continue

    source_speaker = best_candidate.get(
        "speaker"
    )

    source_start = float(
        best_candidate.get("start", 0)
    )

    source_end = float(
        best_candidate.get("end", 0)
    )

    source_event_type = best_candidate.get(
        "type"
    )

    speaker_match = (
        generated_speaker == source_speaker
    )

    timestamp_match = (
        abs(generated_start - source_start) <= 1.0
        and
        abs(generated_end - source_end) <= 1.0
    )

    event_type_match = (
        claim["event_type"] == source_event_type
    )

    metadata_results.append({
        "claim_index": i,
        "speaker_match": speaker_match,
        "timestamp_match": timestamp_match,
        "event_type_match": event_type_match,
        "overlap_ratio": best_overlap,
        "source_speaker": source_speaker,
        "generated_speaker": generated_speaker,
        "source_event_type": source_event_type,
        "generated_event_type": claim["event_type"],
        "source_start": source_start,
        "source_end": source_end,
        "generated_start": generated_start,
        "generated_end": generated_end,
        "candidate": best_candidate,
    })

# ------------------------------------------------------------
# Display consistency results
# ------------------------------------------------------------

for result in metadata_results:

    print(
        f"\nClaim {result['claim_index']}:"
    )

    if result.get("candidate") is None:
        print("  ✗ No matching source candidate")
        continue

    print(
        "  Speaker   :",
        "✓" if result["speaker_match"] else "✗"
    )

    print(
        "  Timestamp :",
        "✓" if result["timestamp_match"] else "✗"
    )

    print(
        "  Event type:",
        "✓" if result["event_type_match"] else "✗"
    )

    print(
        "  Overlap   :",
        f"{result['overlap_ratio']:.3f}"
    )

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

speaker_matches = sum(
    r.get("speaker_match", False)
    for r in metadata_results
)

timestamp_matches = sum(
    r.get("timestamp_match", False)
    for r in metadata_results
)

event_matches = sum(
    r.get("event_type_match", False)
    for r in metadata_results
)

total = len(qwen_claims)

print("\n" + "=" * 80)
print("STEP 11N SUMMARY")
print("=" * 80)

print(
    f"Claims generated       : {total}"
)

print(
    f"Speaker consistency     : "
    f"{speaker_matches}/{total}"
)

print(
    f"Timestamp consistency   : "
    f"{timestamp_matches}/{total}"
)

print(
    f"Event-type consistency  : "
    f"{event_matches}/{total}"
)

print(
    f"Schema errors           : "
    f"{len(schema_errors)}"
)

print("\n✓ STEP 11N COMPLETE")
````

````python
# ============================================================
# STEP 11O — SOURCE-GROUNDED QWEN CLAIM CONSTRUCTION
# ============================================================

print("=" * 80)
print("STEP 11O — SOURCE-GROUNDED QWEN CLAIM CONSTRUCTION")
print("=" * 80)

# ------------------------------------------------------------
# Build source-grounded structured claims
# ------------------------------------------------------------

source_grounded_claims = []

for qwen_claim, validation in zip(
    qwen_claims,
    metadata_results
):

    candidate = validation.get("candidate")

    if candidate is None:
        continue

    grounded_claim = {
        "claim": qwen_claim["claim"],

        # Metadata comes from the source candidate,
        # NOT from Qwen's generated metadata.
        "speaker": candidate["speaker"],
        "event_type": candidate["type"],
        "start": float(candidate["start"]),
        "end": float(candidate["end"]),

        # Preserve provenance for later evidence retrieval
        "source_candidate_id": (
            validation["claim_index"]
        ),
    }

    source_grounded_claims.append(
        grounded_claim
    )

# ------------------------------------------------------------
# Display results
# ------------------------------------------------------------

print("\nGenerated source-grounded claims:")
print("-" * 80)

for i, claim in enumerate(
    source_grounded_claims,
    start=1
):

    print(f"\n[{i}]")
    print("Claim     :", claim["claim"])
    print("Speaker   :", claim["speaker"])
    print("Event type:", claim["event_type"])
    print(
        "Time      : "
        f"{claim['start']:.3f} → "
        f"{claim['end']:.3f}"
    )

# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

assert len(source_grounded_claims) == len(
    qwen_claims
)

for claim in source_grounded_claims:

    assert claim["claim"].strip()
    assert claim["speaker"]
    assert claim["event_type"]
    assert claim["end"] >= claim["start"]

print("\n" + "=" * 80)
print("✓ SOURCE-GROUNDED CLAIM VALIDATION PASSED")
print("=" * 80)

print(
    f"\nClaims retained: "
    f"{len(source_grounded_claims)}"
)
````

````python
# ============================================================
# STEP 11O-B — SOURCE-GROUNDED QWEN CLAIM CONSTRUCTION
# ============================================================

print("=" * 80)
print("STEP 11O-B — SOURCE-GROUNDED QWEN CLAIM CONSTRUCTION")
print("=" * 80)

source_grounded_claims = []

for qwen_claim, validation in zip(
    qwen_claims,
    metadata_results
):

    candidate = validation.get("candidate")

    if candidate is None:
        continue

    grounded_claim = {
        # Qwen supplies only the claim text
        "claim": qwen_claim["claim"],

        # Source candidate supplies metadata
        "speaker": candidate["speaker"],
        "event_type": candidate["candidate_type"],
        "start": float(candidate["start"]),
        "end": float(candidate["end"]),

        # Provenance
        "source_candidate_index": (
            validation["claim_index"]
        ),
    }

    source_grounded_claims.append(
        grounded_claim
    )

# ------------------------------------------------------------
# Display
# ------------------------------------------------------------

print("\nSource-grounded claims:")
print("-" * 80)

for i, claim in enumerate(
    source_grounded_claims,
    start=1
):

    print(f"\n[{i}]")
    print("Claim     :", claim["claim"])
    print("Speaker   :", claim["speaker"])
    print("Event type:", claim["event_type"])
    print(
        "Time      : "
        f"{claim['start']:.3f} → "
        f"{claim['end']:.3f}"
    )
    print(
        "Source    :",
        claim["source_candidate_index"]
    )

# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

assert len(source_grounded_claims) == len(qwen_claims)

for claim in source_grounded_claims:

    assert isinstance(
        claim["claim"],
        str
    )

    assert claim["claim"].strip()

    assert claim["speaker"]

    assert claim["event_type"]

    assert claim["end"] >= claim["start"]

print("\n" + "=" * 80)
print("✓ SOURCE-GROUNDED CLAIM VALIDATION PASSED")
print("=" * 80)

print(
    "\nClaims retained:",
    len(source_grounded_claims)
)
````

````python
# ============================================================
# STEP 11P — INTEGRATE SOURCE-GROUNDED QWEN GENERATION
# ============================================================

from pathlib import Path
import re

pipeline_path = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project/"
    "07_dashboard/meeting_pipeline.py"
)

source = pipeline_path.read_text()

# ------------------------------------------------------------
# Locate current Qwen function
# ------------------------------------------------------------

start_marker = "def generate_mom_qwen("
start = source.find(start_marker)

if start == -1:
    raise RuntimeError(
        "generate_mom_qwen() was not found. "
        "No file was modified."
    )

# Find next top-level function
next_def = source.find(
    "\ndef ",
    start + len(start_marker)
)

if next_def == -1:
    next_def = len(source)

old_function = source[start:next_def]

# ------------------------------------------------------------
# New source-grounded Qwen implementation
# ------------------------------------------------------------

new_function = '''def generate_mom_qwen(
    candidates,
    model=None,
    tokenizer=None,
):
    """
    Qwen structured MoM generation.

    Qwen generates claim text.

    Speaker, event_type, start, and end are inherited
    from the source candidate to preserve provenance.
    """

    context = prepare_candidate_context(
        candidates
    )

    if not context:
        return {
            "generator": "qwen",
            "model_name": GENERATOR_CONFIG[
                "qwen"
            ]["model_name"],
            "status": "empty_context",
            "claims": [],
            "context": context,
        }

    if model is None or tokenizer is None:
        return {
            "generator": "qwen",
            "model_name": GENERATOR_CONFIG[
                "qwen"
            ]["model_name"],
            "status": "context_ready",
            "claims": [],
            "context": context,
        }

    import json
    import re
    import torch

    prompt = f"""
You are generating structured Minutes of Meeting claims.

Use ONLY the information contained in the meeting
candidates below.

For each meaningful meeting claim, return a JSON object
with exactly:

- claim
- speaker
- event_type
- start
- end

Rules:
1. Do not invent information.
2. Keep claims concise and factual.
3. Exclude irrelevant conversation.
4. Return ONLY a JSON array.
5. Do not use Markdown fences.
6. Do not include explanations outside the JSON array.

Meeting candidates:

{context}
"""

    messages = [
        {
            "role": "system",
            "content": (
                "You are a precise meeting-minutes "
                "extraction assistant. "
                "Output valid JSON only."
            ),
        },
        {
            "role": "user",
            "content": prompt,
        },
    ]

    formatted_prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
    )

    inputs = tokenizer(
        formatted_prompt,
        return_tensors="pt",
        truncation=True,
        max_length=8192,
    )

    device = next(
        model.parameters()
    ).device

    inputs = {
        key: value.to(device)
        for key, value in inputs.items()
    }

    with torch.no_grad():

        output_ids = model.generate(
            **inputs,
            max_new_tokens=1200,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id,
        )

    input_length = inputs[
        "input_ids"
    ].shape[1]

    generated_ids = output_ids[
        0,
        input_length:
    ]

    raw_output = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True,
    ).strip()

    clean_output = re.sub(
        r"^```(?:json)?\\s*|\\s*```$",
        "",
        raw_output,
        flags=re.IGNORECASE | re.DOTALL,
    ).strip()

    try:
        generated_claims = json.loads(
            clean_output
        )
    except json.JSONDecodeError as exc:

        return {
            "generator": "qwen",
            "model_name": GENERATOR_CONFIG[
                "qwen"
            ]["model_name"],
            "status": "json_parse_error",
            "claims": [],
            "raw_output": raw_output,
            "error": str(exc),
            "context": context,
        }

    if not isinstance(
        generated_claims,
        list
    ):

        return {
            "generator": "qwen",
            "model_name": GENERATOR_CONFIG[
                "qwen"
            ]["model_name"],
            "status": "invalid_json_structure",
            "claims": [],
            "raw_output": raw_output,
            "context": context,
        }

    # --------------------------------------------------------
    # Ground generated claims to source candidates
    # --------------------------------------------------------

    grounded_claims = []

    for generated in generated_claims:

        if not isinstance(
            generated,
            dict
        ):
            continue

        if not generated.get("claim"):
            continue

        try:
            generated_start = float(
                generated["start"]
            )
            generated_end = float(
                generated["end"]
            )
        except (
            KeyError,
            TypeError,
            ValueError,
        ):
            continue

        best_candidate = None
        best_overlap = 0.0

        for candidate in candidates:

            candidate_start = float(
                candidate["start"]
            )

            candidate_end = float(
                candidate["end"]
            )

            overlap_start = max(
                generated_start,
                candidate_start,
            )

            overlap_end = min(
                generated_end,
                candidate_end,
            )

            overlap = max(
                0.0,
                overlap_end - overlap_start,
            )

            duration = max(
                candidate_end - candidate_start,
                1e-9,
            )

            overlap_ratio = (
                overlap / duration
            )

            if overlap_ratio > best_overlap:

                best_overlap = overlap_ratio
                best_candidate = candidate

        if best_candidate is None:
            continue

        grounded_claims.append(
            {
                "claim": str(
                    generated["claim"]
                ).strip(),

                "speaker": best_candidate[
                    "speaker"
                ],

                "event_type": best_candidate[
                    "candidate_type"
                ],

                "start": float(
                    best_candidate["start"]
                ),

                "end": float(
                    best_candidate["end"]
                ),

                "source_candidate": {
                    "speaker": best_candidate[
                        "speaker"
                    ],
                    "start": float(
                        best_candidate["start"]
                    ),
                    "end": float(
                        best_candidate["end"]
                    ),
                    "event_type": best_candidate[
                        "candidate_type"
                    ],
                },

                "source_overlap": float(
                    best_overlap
                ),
            }
        )

    return {
        "generator": "qwen",
        "model_name": GENERATOR_CONFIG[
            "qwen"
        ]["model_name"],
        "status": "generated",
        "claims": grounded_claims,
        "raw_output": raw_output,
        "context": context,
    }
'''

# ------------------------------------------------------------
# Replace only the Qwen function
# ------------------------------------------------------------

source = (
    source[:start]
    + new_function
    + source[next_def:]
)

pipeline_path.write_text(source)

print("=" * 80)
print("STEP 11P — QWEN PIPELINE INTEGRATION")
print("=" * 80)

print("\n✓ generate_mom_qwen() updated")
print("✓ JSON parsing integrated")
print("✓ Source-candidate matching integrated")
print("✓ Speaker metadata grounded")
print("✓ Event-type metadata grounded")
print("✓ Timestamp metadata grounded")
print("✓ No existing cells/files deleted")
print("\n✓ STEP 11P COMPLETE")
````

````python
# =============================================================================
# STEP 12E — INTEGRATE GENERALIZED BGE + FAISS RETRIEVAL
# =============================================================================

from pathlib import Path

PROJECT_ROOT = Path("/content/drive/MyDrive/MTechIndProj/MoM_Project")
PIPELINE_PATH = PROJECT_ROOT / "07_dashboard" / "meeting_pipeline.py"

print("=" * 80)
print("STEP 12E — INTEGRATING BGE + FAISS RETRIEVAL")
print("=" * 80)

pipeline_text = PIPELINE_PATH.read_text(encoding="utf-8")

# -------------------------------------------------------------------------
# 1. Add imports if they are not already present
# -------------------------------------------------------------------------

imports_to_add = """
import numpy as np
import faiss
from sentence_transformers import SentenceTransformer
"""

if "from sentence_transformers import SentenceTransformer" not in pipeline_text:
    pipeline_text = pipeline_text.rstrip() + "\n\n" + imports_to_add.strip() + "\n"

# -------------------------------------------------------------------------
# 2. Generalized BGE configuration
# -------------------------------------------------------------------------

retrieval_code = r'''

# =============================================================================
# BGE + FAISS EVIDENCE RETRIEVAL
# =============================================================================
# Generalized retrieval layer.
#
# IMPORTANT:
# - Evidence is created from the CURRENT meeting's speaker utterances.
# - No meeting name or transcript file is hard-coded.
# - BGE similarity is used for retrieval only.
# - Evidence verification is performed separately by the NLI stage.
# =============================================================================

BGE_MODEL_NAME = "BAAI/bge-small-en-v1.5"


def load_bge_model(device=None):
    """
    Load the BGE embedding model used for evidence retrieval.
    """

    if device is None:
        device = "cuda" if torch.cuda.is_available() else "cpu"

    return SentenceTransformer(
        BGE_MODEL_NAME,
        device=device
    )


def build_evidence_documents(speaker_utterances):
    """
    Convert the current meeting's speaker utterances into
    evidence documents suitable for BGE + FAISS retrieval.

    Parameters
    ----------
    speaker_utterances : list
        Speaker-level utterances produced by the current meeting pipeline.

    Returns
    -------
    list
        Evidence documents containing speaker, timestamp and text metadata.
    """

    evidence_documents = []

    for i, utterance in enumerate(speaker_utterances, start=1):

        text = str(utterance.get("text", "")).strip()

        if not text:
            continue

        start = float(utterance.get("start", 0.0))
        end = float(utterance.get("end", start))

        evidence_documents.append({
            "evidence_id": i,
            "speaker": utterance.get("speaker"),
            "start": start,
            "end": end,
            "duration": end - start,
            "text": text
        })

    return evidence_documents


def build_faiss_evidence_index(
    evidence_documents,
    bge_model
):
    """
    Build a FAISS cosine-similarity index from the current meeting's
    evidence documents.

    BGE embeddings are normalized, therefore IndexFlatIP provides
    cosine-similarity search.
    """

    if not evidence_documents:
        raise ValueError(
            "Cannot build evidence index: no evidence documents available."
        )

    evidence_texts = [
        document["text"]
        for document in evidence_documents
    ]

    evidence_embeddings = bge_model.encode(
        evidence_texts,
        normalize_embeddings=True,
        convert_to_numpy=True,
        show_progress_bar=False
    ).astype(np.float32)

    embedding_dimension = evidence_embeddings.shape[1]

    faiss_index = faiss.IndexFlatIP(
        embedding_dimension
    )

    faiss_index.add(evidence_embeddings)

    return faiss_index, evidence_embeddings


def retrieve_evidence(
    claim_text,
    bge_model,
    faiss_index,
    evidence_documents,
    top_k=10
):
    """
    Retrieve the top-K semantically relevant evidence documents
    for a generated MoM claim.

    Retrieval similarity identifies candidate evidence.
    It does NOT establish that the evidence supports the claim.
    """

    if not claim_text or not str(claim_text).strip():
        return []

    if faiss_index is None or not evidence_documents:
        return []

    query_embedding = bge_model.encode(
        [str(claim_text)],
        normalize_embeddings=True,
        convert_to_numpy=True
    ).astype(np.float32)

    # Prevent requesting more results than available documents.
    actual_top_k = min(
        int(top_k),
        len(evidence_documents)
    )

    similarities, indices = faiss_index.search(
        query_embedding,
        actual_top_k
    )

    results = []

    for rank, (index, similarity) in enumerate(
        zip(indices[0], similarities[0]),
        start=1
    ):

        if index < 0:
            continue

        evidence = evidence_documents[index].copy()

        evidence["retrieval_rank"] = rank
        evidence["retrieval_similarity"] = float(similarity)

        results.append(evidence)

    return results


def build_evidence_retrieval_resources(
    speaker_utterances,
    device=None
):
    """
    Build all BGE + FAISS resources for the CURRENT meeting.

    This is the main generalized entry point for the retrieval stage.
    """

    evidence_documents = build_evidence_documents(
        speaker_utterances
    )

    bge_model = load_bge_model(
        device=device
    )

    faiss_index, evidence_embeddings = build_faiss_evidence_index(
        evidence_documents,
        bge_model
    )

    return {
        "bge_model": bge_model,
        "faiss_index": faiss_index,
        "evidence_documents": evidence_documents,
        "evidence_embeddings": evidence_embeddings
    }
'''

# -------------------------------------------------------------------------
# 3. Prevent accidental duplicate integration
# -------------------------------------------------------------------------

if "def build_evidence_retrieval_resources(" in pipeline_text:
    print("\n⚠ BGE + FAISS retrieval is already integrated.")
    print("No duplicate code added.")

else:
    pipeline_text = pipeline_text.rstrip() + "\n\n" + retrieval_code.strip() + "\n"

    PIPELINE_PATH.write_text(
        pipeline_text,
        encoding="utf-8"
    )

    print("\n✓ BGE + FAISS retrieval integrated.")

print("\nPipeline file:")
print(PIPELINE_PATH)

print("\nAdded generalized components:")
print("  ✓ load_bge_model()")
print("  ✓ build_evidence_documents()")
print("  ✓ build_faiss_evidence_index()")
print("  ✓ retrieve_evidence()")
print("  ✓ build_evidence_retrieval_resources()")

print("\nImportant:")
print("  ✓ No ES2004a path hard-coded")
print("  ✓ Uses current meeting speaker utterances")
print("  ✓ Retrieval similarity is NOT treated as verification")
print("  ✓ Existing files/cells are not deleted")

print("\n" + "=" * 80)
print("✓ STEP 12E COMPLETE")
print("=" * 80)
````

````python
# STEP 12L — INTEGRATE CLAIM-LEVEL EVIDENCE RETRIEVAL
#
# Adds a new generalized wrapper to meeting_pipeline.py.
# Existing generation and retrieval functions are untouched.

from pathlib import Path

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project"
)

PIPELINE_PATH = (
    PROJECT_ROOT
    / "07_dashboard"
    / "meeting_pipeline.py"
)

with open(
    PIPELINE_PATH,
    "r",
    encoding="utf-8"
) as f:
    pipeline_source = f.read()

function_marker = """
def retrieve_evidence_for_generated_claims(
    generation_result,
    speaker_utterances,
    top_k=10,
    device=None,
):
"""

if function_marker.strip() in pipeline_source:
    print("=" * 80)
    print("STEP 12L — FUNCTION ALREADY PRESENT")
    print("=" * 80)
    print("✓ No duplicate function added")
else:

    new_function = '''

def retrieve_evidence_for_generated_claims(
    generation_result,
    speaker_utterances,
    top_k=10,
    device=None,
):
    """
    Attach BGE + FAISS evidence to generated MoM claims.

    Generation and retrieval remain separate stages.

    Parameters
    ----------
    generation_result : dict
        Output from generate_structured_mom(),
        including the "claims" list.

    speaker_utterances : list
        Speaker-level meeting utterances.

    top_k : int
        Number of evidence items retrieved per claim.

    device : str or None
        Device used by the BGE model.

    Returns
    -------
    dict
        Copy of generation_result with retrieved
        evidence attached to each generated claim.
    """

    if not isinstance(
        generation_result,
        dict
    ):
        raise TypeError(
            "generation_result must be a dictionary."
        )

    claims = generation_result.get(
        "claims",
        []
    )

    if not isinstance(
        claims,
        list
    ):
        claims = []

    retrieval_resources = (
        build_evidence_retrieval_resources(
            speaker_utterances,
            device=device,
        )
    )

    bge_model = retrieval_resources[
        "bge_model"
    ]

    faiss_index = retrieval_resources[
        "faiss_index"
    ]

    evidence_documents = retrieval_resources[
        "evidence_documents"
    ]

    retrieved_claims = []

    for claim in claims:

        if not isinstance(
            claim,
            dict
        ):
            continue

        claim_copy = claim.copy()

        claim_text = str(
            claim.get(
                "claim",
                ""
            )
        ).strip()

        evidence = retrieve_evidence(
            claim_text=claim_text,
            bge_model=bge_model,
            faiss_index=faiss_index,
            evidence_documents=evidence_documents,
            top_k=top_k,
        )

        claim_copy[
            "retrieved_evidence"
        ] = evidence

        retrieved_claims.append(
            claim_copy
        )

    result = generation_result.copy()

    result[
        "claims"
    ] = retrieved_claims

    result[
        "retrieval"
    ] = {
        "model": BGE_MODEL_NAME,
        "top_k": int(top_k),
        "evidence_count": len(
            evidence_documents
        ),
        "status": "completed",
    }

    return result

'''

    with open(
        PIPELINE_PATH,
        "a",
        encoding="utf-8"
    ) as f:
        f.write(new_function)

    print("=" * 80)
    print("STEP 12L — CLAIM-LEVEL RETRIEVAL INTEGRATION")
    print("=" * 80)
    print("✓ New function added:")
    print(
        "  retrieve_evidence_for_generated_claims()"
    )
    print("✓ Existing generation functions untouched")
    print("✓ Existing retrieval functions untouched")
    print("✓ No ES2004a hard-coding")
    print("✓ No files/cells deleted")
    print("=" * 80)
````

````python
# STEP 12O — ACTUAL QWEN CLAIMS → INTEGRATED RETRIEVAL

import os
import torch
import meeting_pipeline

print("=" * 80)
print("STEP 12O — QWEN CLAIMS → BGE + FAISS RETRIEVAL")
print("=" * 80)

PROJECT_ROOT = "/content/drive/MyDrive/MTechIndProj/MoM_Project"

# ------------------------------------------------------------
# 1. Load validated speaker transcript
# ------------------------------------------------------------

speaker_transcript_path = os.path.join(
    PROJECT_ROOT,
    "data",
    "transcripts",
    "ES2004a_pipeline_speaker_transcript.json"
)

speaker_transcript_data = meeting_pipeline.load_json(
    speaker_transcript_path
)

speaker_utterances = speaker_transcript_data["utterances"]

print(
    f"✓ Speaker utterances: "
    f"{len(speaker_utterances)}"
)

# ------------------------------------------------------------
# 2. Load Qwen model and tokenizer
# ------------------------------------------------------------

from transformers import AutoTokenizer, AutoModelForCausalLM

QWEN_MODEL_NAME = (
    meeting_pipeline.GENERATOR_CONFIG[
        "qwen"
    ]["model_name"]
)

print(
    f"\nLoading Qwen: "
    f"{QWEN_MODEL_NAME}"
)

qwen_tokenizer = AutoTokenizer.from_pretrained(
    QWEN_MODEL_NAME
)

qwen_model = AutoModelForCausalLM.from_pretrained(
    QWEN_MODEL_NAME,
    torch_dtype=(
        torch.float16
        if torch.cuda.is_available()
        else torch.float32
    ),
    device_map="auto"
)

print("✓ Qwen model loaded")

# ------------------------------------------------------------
# 3. Generate actual structured MoM claims
# ------------------------------------------------------------

generation_result = (
    meeting_pipeline.generate_structured_mom(
        speaker_utterances,
        generator="qwen",
        model=qwen_model,
        tokenizer=qwen_tokenizer,
    )
)

print(
    f"\n✓ Generation status: "
    f"{generation_result.get('status')}"
)

generated_claims = generation_result.get(
    "claims",
    []
)

print(
    f"✓ Qwen claims generated: "
    f"{len(generated_claims)}"
)

assert len(generated_claims) > 0, (
    "Qwen generated no claims"
)

# ------------------------------------------------------------
# 4. Run integrated retrieval
# ------------------------------------------------------------

retrieval_result = (
    meeting_pipeline
    .retrieve_evidence_for_generated_claims(
        generation_result=generation_result,
        speaker_utterances=speaker_utterances,
        top_k=5,
    )
)

retrieved_claims = retrieval_result.get(
    "claims",
    []
)

print(
    f"\n✓ Claims after retrieval: "
    f"{len(retrieved_claims)}"
)

# ------------------------------------------------------------
# 5. Validate retrieval attachment
# ------------------------------------------------------------

claims_with_evidence = 0

for claim in retrieved_claims:

    evidence = claim.get(
        "retrieved_evidence",
        []
    )

    if len(evidence) > 0:
        claims_with_evidence += 1

assert len(retrieved_claims) == len(
    generated_claims
)

assert claims_with_evidence == len(
    retrieved_claims
)

# ------------------------------------------------------------
# 6. Display summary
# ------------------------------------------------------------

print("\nCLAIM RETRIEVAL SUMMARY")
print("-" * 80)

for i, claim in enumerate(
    retrieved_claims,
    start=1
):

    evidence = claim[
        "retrieved_evidence"
    ]

    top_similarity = (
        evidence[0][
            "retrieval_similarity"
        ]
        if evidence
        else None
    )

    print(
        f"Claim {i:02d} | "
        f"{claim.get('event_type')} | "
        f"{claim.get('speaker')} | "
        f"Evidence: {len(evidence)} | "
        f"Top similarity: "
        f"{top_similarity:.4f}"
    )

    print(
        f"  {claim['claim'][:200]}"
    )

print("\n" + "=" * 80)
print(
    "✓ STEP 12O QWEN + RETRIEVAL TEST PASSED"
)
print("=" * 80)
````

````python
# =============================================================================
# STEP 12O-R5 — BGE SEMANTIC CLAIM-TO-CANDIDATE MATCHING
# =============================================================================

import sys
import os
import numpy as np
import torch

# -------------------------------------------------------------------------
# 1. Add project pipeline directory to Python path
# -------------------------------------------------------------------------
PIPELINE_DIR = (
    "/content/drive/MyDrive/MTechIndProj/MoM_Project/"
    "07_dashboard"
)

if PIPELINE_DIR not in sys.path:
    sys.path.insert(0, PIPELINE_DIR)

import meeting_pipeline

print("=" * 80)
print("STEP 12O-R5 — BGE SEMANTIC CLAIM-TO-CANDIDATE MATCHING")
print("=" * 80)

print(f"✓ meeting_pipeline loaded from:")
print(f"  {meeting_pipeline.__file__}")

# -------------------------------------------------------------------------
# 2. Extract Qwen-generated claims
# -------------------------------------------------------------------------
raw_output = generation_result.get("raw_output", "")

import json
import re

clean_output = re.sub(
    r"```(?:json)?|```",
    "",
    raw_output,
    flags=re.IGNORECASE
).strip()

qwen_claims = json.loads(clean_output)

print(f"✓ Qwen claims extracted: {len(qwen_claims)}")

# -------------------------------------------------------------------------
# 3. Use the actual V3 candidates from 12O-R2
# -------------------------------------------------------------------------
print(f"✓ V3 test candidates: {len(test_candidates)}")

# -------------------------------------------------------------------------
# 4. Load BGE
# -------------------------------------------------------------------------
print("\nLoading BGE...")

bge_model = meeting_pipeline.load_bge_model(
    device="cuda" if torch.cuda.is_available() else "cpu"
)

print("✓ BGE loaded")

# -------------------------------------------------------------------------
# 5. Encode V3 candidate texts
# -------------------------------------------------------------------------
candidate_texts = [
    str(candidate.get("text", "")).strip()
    for candidate in test_candidates
]

candidate_embeddings = bge_model.encode(
    candidate_texts,
    normalize_embeddings=True,
    convert_to_numpy=True,
    show_progress_bar=False,
).astype(np.float32)

# -------------------------------------------------------------------------
# 6. Encode Qwen claims
# -------------------------------------------------------------------------
claim_texts = [
    str(item.get("claim", "")).strip()
    for item in qwen_claims
]

claim_embeddings = bge_model.encode(
    claim_texts,
    normalize_embeddings=True,
    convert_to_numpy=True,
    show_progress_bar=False,
).astype(np.float32)

# -------------------------------------------------------------------------
# 7. Calculate semantic similarity
# -------------------------------------------------------------------------
similarities = np.matmul(
    claim_embeddings,
    candidate_embeddings.T
)

# -------------------------------------------------------------------------
# 8. Display top semantic matches
# -------------------------------------------------------------------------
print("\n" + "=" * 80)
print("QWEN CLAIM → BEST V3 CANDIDATE")
print("=" * 80)

for claim_idx, qwen_item in enumerate(qwen_claims):

    ranked_indices = np.argsort(
        similarities[claim_idx]
    )[::-1]

    print("\n" + "-" * 80)
    print(f"CLAIM {claim_idx + 1}")
    print("-" * 80)

    print("Qwen claim:")
    print(qwen_item.get("claim"))

    print("\nTop 3 candidate matches:")

    for rank, candidate_idx in enumerate(
        ranked_indices[:3],
        start=1
    ):
        candidate = test_candidates[candidate_idx]
        score = similarities[
            claim_idx,
            candidate_idx
        ]

        print(
            f"\nRank {rank} | "
            f"BGE similarity: {score:.4f}"
        )

        print(
            f"Speaker: {candidate.get('speaker')}"
        )

        print(
            f"Type: {candidate.get('candidate_type')}"
        )

        print(
            f"Time: "
            f"{candidate.get('start'):.3f} - "
            f"{candidate.get('end'):.3f}"
        )

        print(
            f"Text: {candidate.get('text', '')[:500]}"
        )

# -------------------------------------------------------------------------
# 9. Completion
# -------------------------------------------------------------------------
print("\n" + "=" * 80)
print("12O-R5 COMPLETE")
print("=" * 80)

print(f"✓ Claims evaluated: {len(qwen_claims)}")
print(
    f"✓ Candidates evaluated per claim: "
    f"{len(test_candidates)}"
)
print("✓ No files modified")
print("✓ meeting_pipeline.py not modified")
````

The goal now is to take the 5 Qwen-generated claims + their BGE-retrieved evidence and run the existing DeBERTa NLI + lexical + speaker/event consistency logic.

````python
# ============================================================
# STEP 12O-R6 — QWEN CLAIM VERIFICATION TEST
# ============================================================
# Purpose:
#   Test the existing multi-attribute verification logic on
#   the 5 Qwen-generated claims.
#
# IMPORTANT:
#   - Test only
#   - No pipeline modification
#   - No files deleted
#   - Qwen is NOT used for verification
#   - DeBERTa NLI performs content verification
# ============================================================

print("=" * 90)
print("STEP 12O-R6 — QWEN CLAIM VERIFICATION TEST")
print("=" * 90)

# ------------------------------------------------------------
# 1. Check required functions
# ------------------------------------------------------------

required_functions = [
    "run_nli",
    "lexical_support",
    "check_speaker_consistency",
    "check_event_type_consistency",
]

missing_functions = [
    name for name in required_functions
    if name not in globals()
]

if missing_functions:
    print("\nMissing functions:")
    for name in missing_functions:
        print(" -", name)

    raise RuntimeError(
        "Required verification functions are not loaded."
    )

print("\n✓ Required verification functions are available.")


# ------------------------------------------------------------
# 2. Check Qwen claims
# ------------------------------------------------------------

print("\nQwen claims available:", len(qwen_claims))

for i, claim in enumerate(qwen_claims, start=1):

    print(
        f"\nClaim {i}: "
        f"{claim.get('claim', '')}"
    )

    print(
        "Speaker:",
        claim.get("speaker")
    )

    print(
        "Type:",
        claim.get("type", claim.get("event_type"))
    )


# ------------------------------------------------------------
# 3. Load DeBERTa NLI model
# ------------------------------------------------------------

from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

NLI_MODEL_NAME = "cross-encoder/nli-deberta-v3-small"

print("\nLoading NLI model:")
print(NLI_MODEL_NAME)

nli_tokenizer = AutoTokenizer.from_pretrained(
    NLI_MODEL_NAME
)

nli_model = AutoModelForSequenceClassification.from_pretrained(
    NLI_MODEL_NAME
)

nli_device = (
    torch.device("cuda")
    if torch.cuda.is_available()
    else torch.device("cpu")
)

nli_model = nli_model.to(nli_device)
nli_model.eval()

print("✓ NLI model loaded on:", nli_device)


# ------------------------------------------------------------
# 4. NLI helper for this test
# ------------------------------------------------------------

def run_qwen_nli_test(
    claim_text,
    evidence_text
):
    """
    Run DeBERTa NLI for one claim/evidence pair.
    """

    inputs = nli_tokenizer(
        claim_text,
        evidence_text,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    inputs = {
        key: value.to(nli_device)
        for key, value in inputs.items()
    }

    with torch.no_grad():

        outputs = nli_model(**inputs)

        probabilities = torch.softmax(
            outputs.logits,
            dim=-1
        )[0]

    contradiction = float(probabilities[0])
    entailment = float(probabilities[1])
    neutral = float(probabilities[2])

    scores = {
        "contradiction": contradiction,
        "entailment": entailment,
        "neutral": neutral
    }

    label = max(
        scores,
        key=scores.get
    )

    return {
        "label": label,
        "contradiction": contradiction,
        "entailment": entailment,
        "neutral": neutral
    }


# ------------------------------------------------------------
# 5. Verify each Qwen claim against top-3 BGE evidence
# ------------------------------------------------------------

qwen_verification_results = []

for claim_index, claim in enumerate(
    qwen_claims,
    start=1
):

    claim_text = str(
        claim.get("claim", "")
    ).strip()

    claim_speaker = claim.get(
        "speaker"
    )

    claim_type = claim.get(
        "event_type",
        claim.get("type")
    )

    # Retrieve top-3 evidence using the
    # already validated BGE + FAISS resources

    retrieved = retrieve_evidence(
        claim_text,
        bge_model=bge_model,
        faiss_index=faiss_index,
        evidence_documents=evidence_documents,
        top_k=3
    )

    candidate_results = []

    for evidence in retrieved:

        evidence_text = str(
            evidence.get("text", "")
        ).strip()

        # -------------------------------
        # DeBERTa NLI
        # -------------------------------

        nli_result = run_qwen_nli_test(
            claim_text,
            evidence_text
        )

        # -------------------------------
        # Lexical support
        # -------------------------------

        try:

            lexical_result = lexical_support(
                claim_text,
                evidence_text
            )

            if isinstance(
                lexical_result,
                dict
            ):
                lexical_overlap = float(
                    lexical_result.get(
                        "overlap",
                        lexical_result.get(
                            "lexical_overlap",
                            0.0
                        )
                    )
                )

            else:
                lexical_overlap = float(
                    lexical_result
                )

        except Exception:

            lexical_overlap = 0.0

        # -------------------------------
        # Relevance-aware status
        # -------------------------------

        if (
            nli_result["label"] == "entailment"
            and nli_result["entailment"] >= 0.80
        ):

            content_status = "SUPPORTED"

        elif (
            nli_result["label"] == "contradiction"
            and nli_result["contradiction"] >= 0.80
            and lexical_overlap >= 0.20
        ):

            content_status = "CONTRADICTED"

        else:

            content_status = "UNCERTAIN"

        # -------------------------------
        # Speaker consistency
        # -------------------------------

        speaker_result = check_speaker_consistency(
            claim_speaker,
            [evidence.get("speaker")]
        )

        # -------------------------------
        # Event-type consistency
        # -------------------------------

        event_result = check_event_type_consistency(
            claim_type,
            [evidence_text]
        )

        candidate_results.append({

            "rank":
                evidence.get(
                    "retrieval_rank"
                ),

            "evidence_id":
                evidence.get(
                    "evidence_id"
                ),

            "retrieval_similarity":
                evidence.get(
                    "retrieval_similarity"
                ),

            "speaker":
                evidence.get(
                    "speaker"
                ),

            "start":
                evidence.get(
                    "start"
                ),

            "end":
                evidence.get(
                    "end"
                ),

            "text":
                evidence_text,

            "nli_label":
                nli_result["label"],

            "nli_contradiction":
                nli_result["contradiction"],

            "nli_entailment":
                nli_result["entailment"],

            "nli_neutral":
                nli_result["neutral"],

            "lexical_overlap":
                lexical_overlap,

            "content_status":
                content_status,

            "speaker_consistent":
                speaker_result.get(
                    "passed",
                    False
                ),

            "event_type_consistent":
                event_result.get(
                    "passed",
                    False
                )
        })

    # --------------------------------------------------------
    # Select strongest content-supported candidate
    # --------------------------------------------------------

    supported = [
        item
        for item in candidate_results
        if item["content_status"] == "SUPPORTED"
    ]

    if supported:

        best = max(
            supported,
            key=lambda x: (
                x["nli_entailment"],
                x["retrieval_similarity"]
            )
        )

        overall_status = "SUPPORTED"

    else:

        best = max(
            candidate_results,
            key=lambda x: (
                x["retrieval_similarity"]
            )
        )

        overall_status = "UNCERTAIN"

    qwen_verification_results.append({

        "claim_index":
            claim_index,

        "claim":
            claim_text,

        "claim_speaker":
            claim_speaker,

        "claim_type":
            claim_type,

        "status":
            overall_status,

        "best_candidate":
            best,

        "candidates":
            candidate_results
    })


# ------------------------------------------------------------
# 6. Display compact results
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("QWEN CLAIM → EVIDENCE VERIFICATION RESULTS")
print("=" * 90)

for result in qwen_verification_results:

    best = result["best_candidate"]

    print(
        f"\nCLAIM {result['claim_index']}"
    )

    print(
        "Claim:",
        result["claim"]
    )

    print(
        "Claim speaker:",
        result["claim_speaker"]
    )

    print(
        "Claim type:",
        result["claim_type"]
    )

    print(
        "Content status:",
        result["status"]
    )

    print(
        "Best evidence rank:",
        best["rank"]
    )

    print(
        "Evidence speaker:",
        best["speaker"]
    )

    print(
        "BGE similarity:",
        round(
            best["retrieval_similarity"],
            4
        )
    )

    print(
        "NLI:",
        best["nli_label"],
        "| entailment:",
        round(
            best["nli_entailment"],
            4
        ),
        "| contradiction:",
        round(
            best["nli_contradiction"],
            4
        )
    )

    print(
        "Lexical overlap:",
        round(
            best["lexical_overlap"],
            4
        )
    )

    print(
        "Speaker consistency:",
        best["speaker_consistent"]
    )

    print(
        "Event consistency:",
        best["event_type_consistent"]
    )

    print(
        "Evidence:",
        best["text"]
    )

print("\n" + "=" * 90)
print("STEP 12O-R6 COMPLETE")
print("=" * 90)
````

````python
# ============================================================
# STEP 12O-R6A — RESTORE EXISTING VERIFICATION FUNCTIONS
# ============================================================
# Purpose:
#   Restore the already-developed M7/M10 verification functions
#   from the saved 04_mom_generation notebook.
#
# IMPORTANT:
#   - No files deleted
#   - No pipeline modification
#   - Uses the existing project implementation
# ============================================================

import json
import os
import re

PROJECT_DIR = "/content/drive/MyDrive/MTechIndProj/MoM_Project"

M7_NOTEBOOK = os.path.join(
    PROJECT_DIR,
    "04_mom_generation",
    "04_mom_generation.ipynb"
)

print("=" * 90)
print("STEP 12O-R6A — RESTORING EXISTING VERIFICATION FUNCTIONS")
print("=" * 90)

print("\nNotebook:")
print(M7_NOTEBOOK)

if not os.path.exists(M7_NOTEBOOK):
    raise FileNotFoundError(
        f"Verification notebook not found:\n{M7_NOTEBOOK}"
    )

with open(
    M7_NOTEBOOK,
    "r",
    encoding="utf-8"
) as f:

    m7_nb = json.load(f)

print(
    "\n✓ Notebook loaded:",
    len(m7_nb["cells"]),
    "cells"
)


# ------------------------------------------------------------
# Functions required for the verification test
# ------------------------------------------------------------

target_functions = [
    "run_nli",
    "lexical_support",
    "check_speaker_consistency",
    "check_event_type_consistency",
    "check_timestamp_consistency",
    "verify_attribute",
    "verify_mom_claim",
]


# ------------------------------------------------------------
# Extract function definitions from notebook
# ------------------------------------------------------------

function_sources = {}

for cell in m7_nb["cells"]:

    source = "".join(
        cell.get("source", [])
    )

    for function_name in target_functions:

        marker = f"def {function_name}("

        if marker in source:

            function_sources[
                function_name
            ] = source


# ------------------------------------------------------------
# Display what was found
# ------------------------------------------------------------

print("\nFunctions found:")

for function_name in target_functions:

    if function_name in function_sources:

        print(
            "✓",
            function_name
        )

    else:

        print(
            "✗",
            function_name
        )


# ------------------------------------------------------------
# Restore definitions
# ------------------------------------------------------------
#
# We execute only cells containing the required definitions.
# This restores the existing functions into the current
# notebook namespace.
# ------------------------------------------------------------

for function_name, source in function_sources.items():

    try:

        exec(
            source,
            globals()
        )

        print(
            "✓ Restored:",
            function_name
        )

    except Exception as e:

        print(
            "\n✗ Failed to restore:",
            function_name
        )

        print(
            type(e).__name__,
            ":",
            e
        )


# ------------------------------------------------------------
# Final availability check
# ------------------------------------------------------------

print("\n" + "-" * 90)
print("FINAL FUNCTION AVAILABILITY")
print("-" * 90)

missing = []

for function_name in target_functions:

    if callable(
        globals().get(function_name)
    ):

        print(
            "✓",
            function_name
        )

    else:

        print(
            "✗",
            function_name
        )

        missing.append(
            function_name
        )


print("\n" + "=" * 90)

if missing:

    print(
        "RESTORATION INCOMPLETE"
    )

    print(
        "Missing:",
        missing
    )

else:

    print(
        "✓ ALL REQUIRED VERIFICATION FUNCTIONS RESTORED"
    )

print("=" * 90)
````

````python
# ============================================================
# STEP 12O-R6 — QWEN CLAIM VERIFICATION TEST
# ============================================================
# Purpose:
#   Test the existing multi-attribute verification logic on
#   the 5 Qwen-generated claims.
#
# IMPORTANT:
#   - Test only
#   - No pipeline modification
#   - No files deleted
#   - Qwen is NOT used for verification
#   - DeBERTa NLI performs content verification
# ============================================================

print("=" * 90)
print("STEP 12O-R6 — QWEN CLAIM VERIFICATION TEST")
print("=" * 90)

# ------------------------------------------------------------
# 1. Check required functions
# ------------------------------------------------------------

required_functions = [
    "run_nli",
    "lexical_support",
    "check_speaker_consistency",
    "check_event_type_consistency",
]

missing_functions = [
    name for name in required_functions
    if name not in globals()
]

if missing_functions:
    print("\nMissing functions:")
    for name in missing_functions:
        print(" -", name)

    raise RuntimeError(
        "Required verification functions are not loaded."
    )

print("\n✓ Required verification functions are available.")


# ------------------------------------------------------------
# 2. Check Qwen claims
# ------------------------------------------------------------

print("\nQwen claims available:", len(qwen_claims))

for i, claim in enumerate(qwen_claims, start=1):

    print(
        f"\nClaim {i}: "
        f"{claim.get('claim', '')}"
    )

    print(
        "Speaker:",
        claim.get("speaker")
    )

    print(
        "Type:",
        claim.get("type", claim.get("event_type"))
    )


# ------------------------------------------------------------
# 3. Load DeBERTa NLI model
# ------------------------------------------------------------

from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

NLI_MODEL_NAME = "cross-encoder/nli-deberta-v3-small"

print("\nLoading NLI model:")
print(NLI_MODEL_NAME)

nli_tokenizer = AutoTokenizer.from_pretrained(
    NLI_MODEL_NAME
)

nli_model = AutoModelForSequenceClassification.from_pretrained(
    NLI_MODEL_NAME
)

nli_device = (
    torch.device("cuda")
    if torch.cuda.is_available()
    else torch.device("cpu")
)

nli_model = nli_model.to(nli_device)
nli_model.eval()

print("✓ NLI model loaded on:", nli_device)


# ------------------------------------------------------------
# 4. NLI helper for this test
# ------------------------------------------------------------

def run_qwen_nli_test(
    claim_text,
    evidence_text
):
    """
    Run DeBERTa NLI for one claim/evidence pair.
    """

    inputs = nli_tokenizer(
        claim_text,
        evidence_text,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    inputs = {
        key: value.to(nli_device)
        for key, value in inputs.items()
    }

    with torch.no_grad():

        outputs = nli_model(**inputs)

        probabilities = torch.softmax(
            outputs.logits,
            dim=-1
        )[0]

    contradiction = float(probabilities[0])
    entailment = float(probabilities[1])
    neutral = float(probabilities[2])

    scores = {
        "contradiction": contradiction,
        "entailment": entailment,
        "neutral": neutral
    }

    label = max(
        scores,
        key=scores.get
    )

    return {
        "label": label,
        "contradiction": contradiction,
        "entailment": entailment,
        "neutral": neutral
    }


# ------------------------------------------------------------
# 5. Verify each Qwen claim against top-3 BGE evidence
# ------------------------------------------------------------

qwen_verification_results = []

for claim_index, claim in enumerate(
    qwen_claims,
    start=1
):

    claim_text = str(
        claim.get("claim", "")
    ).strip()

    claim_speaker = claim.get(
        "speaker"
    )

    claim_type = claim.get(
        "event_type",
        claim.get("type")
    )

    # Retrieve top-3 evidence using the
    # already validated BGE + FAISS resources

    retrieved = retrieve_evidence(
        claim_text,
        bge_model=bge_model,
        faiss_index=faiss_index,
        evidence_documents=evidence_documents,
        top_k=3
    )

    candidate_results = []

    for evidence in retrieved:

        evidence_text = str(
            evidence.get("text", "")
        ).strip()

        # -------------------------------
        # DeBERTa NLI
        # -------------------------------

        nli_result = run_qwen_nli_test(
            claim_text,
            evidence_text
        )

        # -------------------------------
        # Lexical support
        # -------------------------------

        try:

            lexical_result = lexical_support(
                claim_text,
                evidence_text
            )

            if isinstance(
                lexical_result,
                dict
            ):
                lexical_overlap = float(
                    lexical_result.get(
                        "overlap",
                        lexical_result.get(
                            "lexical_overlap",
                            0.0
                        )
                    )
                )

            else:
                lexical_overlap = float(
                    lexical_result
                )

        except Exception:

            lexical_overlap = 0.0

        # -------------------------------
        # Relevance-aware status
        # -------------------------------

        if (
            nli_result["label"] == "entailment"
            and nli_result["entailment"] >= 0.80
        ):

            content_status = "SUPPORTED"

        elif (
            nli_result["label"] == "contradiction"
            and nli_result["contradiction"] >= 0.80
            and lexical_overlap >= 0.20
        ):

            content_status = "CONTRADICTED"

        else:

            content_status = "UNCERTAIN"

        # -------------------------------
        # Speaker consistency
        # -------------------------------

        speaker_result = check_speaker_consistency(
            claim_speaker,
            [evidence.get("speaker")]
        )

        # -------------------------------
        # Event-type consistency
        # -------------------------------

        event_result = check_event_type_consistency(
            claim_type,
            [evidence_text]
        )

        candidate_results.append({

            "rank":
                evidence.get(
                    "retrieval_rank"
                ),

            "evidence_id":
                evidence.get(
                    "evidence_id"
                ),

            "retrieval_similarity":
                evidence.get(
                    "retrieval_similarity"
                ),

            "speaker":
                evidence.get(
                    "speaker"
                ),

            "start":
                evidence.get(
                    "start"
                ),

            "end":
                evidence.get(
                    "end"
                ),

            "text":
                evidence_text,

            "nli_label":
                nli_result["label"],

            "nli_contradiction":
                nli_result["contradiction"],

            "nli_entailment":
                nli_result["entailment"],

            "nli_neutral":
                nli_result["neutral"],

            "lexical_overlap":
                lexical_overlap,

            "content_status":
                content_status,

            "speaker_consistent":
                speaker_result.get(
                    "passed",
                    False
                ),

            "event_type_consistent":
                event_result.get(
                    "passed",
                    False
                )
        })

    # --------------------------------------------------------
    # Select strongest content-supported candidate
    # --------------------------------------------------------

    supported = [
        item
        for item in candidate_results
        if item["content_status"] == "SUPPORTED"
    ]

    if supported:

        best = max(
            supported,
            key=lambda x: (
                x["nli_entailment"],
                x["retrieval_similarity"]
            )
        )

        overall_status = "SUPPORTED"

    else:

        best = max(
            candidate_results,
            key=lambda x: (
                x["retrieval_similarity"]
            )
        )

        overall_status = "UNCERTAIN"

    qwen_verification_results.append({

        "claim_index":
            claim_index,

        "claim":
            claim_text,

        "claim_speaker":
            claim_speaker,

        "claim_type":
            claim_type,

        "status":
            overall_status,

        "best_candidate":
            best,

        "candidates":
            candidate_results
    })


# ------------------------------------------------------------
# 6. Display compact results
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("QWEN CLAIM → EVIDENCE VERIFICATION RESULTS")
print("=" * 90)

for result in qwen_verification_results:

    best = result["best_candidate"]

    print(
        f"\nCLAIM {result['claim_index']}"
    )

    print(
        "Claim:",
        result["claim"]
    )

    print(
        "Claim speaker:",
        result["claim_speaker"]
    )

    print(
        "Claim type:",
        result["claim_type"]
    )

    print(
        "Content status:",
        result["status"]
    )

    print(
        "Best evidence rank:",
        best["rank"]
    )

    print(
        "Evidence speaker:",
        best["speaker"]
    )

    print(
        "BGE similarity:",
        round(
            best["retrieval_similarity"],
            4
        )
    )

    print(
        "NLI:",
        best["nli_label"],
        "| entailment:",
        round(
            best["nli_entailment"],
            4
        ),
        "| contradiction:",
        round(
            best["nli_contradiction"],
            4
        )
    )

    print(
        "Lexical overlap:",
        round(
            best["lexical_overlap"],
            4
        )
    )

    print(
        "Speaker consistency:",
        best["speaker_consistent"]
    )

    print(
        "Event consistency:",
        best["event_type_consistent"]
    )

    print(
        "Evidence:",
        best["text"]
    )

print("\n" + "=" * 90)
print("STEP 12O-R6 COMPLETE")
print("=" * 90)
````

````python
# ============================================================
# STEP 12O-R7 — MULTI-CANDIDATE QWEN CLAIM VERIFICATION
# ============================================================
# Purpose:
#   Verify each Qwen claim against ALL top-3 BGE candidates.
#
# Verification:
#   1. BGE retrieval
#   2. DeBERTa NLI
#   3. Lexical relevance
#   4. M10 relevance-gated classification
#   5. Speaker consistency
#   6. Event-type consistency
#
# IMPORTANT:
#   - TEST ONLY
#   - No meeting_pipeline.py modification
#   - No files deleted
#   - Qwen is NOT used as the verifier
# ============================================================

print("=" * 90)
print("STEP 12O-R7 — MULTI-CANDIDATE QWEN CLAIM VERIFICATION")
print("=" * 90)


# ------------------------------------------------------------
# 1. M10 relevance-gated classification
# ------------------------------------------------------------

def classify_r7_candidate(
    nli_result,
    lexical_overlap,
    entailment_threshold=0.80,
    contradiction_threshold=0.80,
    relevance_threshold=0.20
):
    """
    Existing M10 decision logic.

    SUPPORTED:
        entailment >= 0.80

    CONTRADICTED:
        contradiction >= 0.80
        AND lexical overlap >= 0.20

    UNCERTAIN:
        everything else
    """

    label = str(
        nli_result["label"]
    ).lower()

    entailment = float(
        nli_result["entailment"]
    )

    contradiction = float(
        nli_result["contradiction"]
    )

    lexical_overlap = float(
        lexical_overlap
    )

    if (
        label == "entailment"
        and entailment >= entailment_threshold
    ):
        return "SUPPORTED"

    if (
        label == "contradiction"
        and contradiction >= contradiction_threshold
        and lexical_overlap >= relevance_threshold
    ):
        return "CONTRADICTED"

    return "UNCERTAIN"


# ------------------------------------------------------------
# 2. Verify all top-3 candidates
# ------------------------------------------------------------

r7_results = []


for claim_index, claim in enumerate(
    qwen_claims,
    start=1
):

    claim_text = str(
        claim.get("claim", "")
    ).strip()

    claim_speaker = claim.get(
        "speaker"
    )

    claim_type = claim.get(
        "event_type",
        claim.get("type")
    )

    print("\n" + "=" * 90)
    print(
        f"CLAIM {claim_index}"
    )
    print("=" * 90)

    print(
        "Claim:",
        claim_text
    )

    print(
        "Qwen speaker:",
        claim_speaker
    )

    print(
        "Qwen event type:",
        claim_type
    )


    # --------------------------------------------------------
    # Retrieve TOP-3
    # --------------------------------------------------------

    retrieved = retrieve_evidence(
        claim_text,
        bge_model=bge_model,
        faiss_index=faiss_index,
        evidence_documents=evidence_documents,
        top_k=3
    )

    candidate_results = []


    # --------------------------------------------------------
    # Evaluate every retrieved candidate
    # --------------------------------------------------------

    for evidence in retrieved:

        evidence_text = str(
            evidence.get("text", "")
        ).strip()


        # ----------------------------------------------------
        # NLI
        # ----------------------------------------------------

        nli_result = run_qwen_nli_test(
            claim_text,
            evidence_text
        )


        # ----------------------------------------------------
        # Lexical support
        # ----------------------------------------------------

        try:

            lexical_result = lexical_support(
                claim_text,
                evidence_text
            )

            if isinstance(
                lexical_result,
                dict
            ):

                lexical_overlap = float(
                    lexical_result.get(
                        "overlap",
                        lexical_result.get(
                            "lexical_overlap",
                            0.0
                        )
                    )
                )

            else:

                lexical_overlap = float(
                    lexical_result
                )

        except Exception:

            lexical_overlap = 0.0


        # ----------------------------------------------------
        # M10 content classification
        # ----------------------------------------------------

        content_status = classify_r7_candidate(
            nli_result,
            lexical_overlap
        )


        # ----------------------------------------------------
        # Speaker consistency
        # ----------------------------------------------------

        speaker_result = check_speaker_consistency(
            claim_speaker,
            [evidence.get("speaker")]
        )


        # ----------------------------------------------------
        # Event-type consistency
        # ----------------------------------------------------

        event_result = check_event_type_consistency(
            claim_type,
            [evidence_text]
        )


        candidate_results.append({

            "rank":
                evidence.get(
                    "retrieval_rank"
                ),

            "evidence_id":
                evidence.get(
                    "evidence_id"
                ),

            "retrieval_similarity":
                float(
                    evidence.get(
                        "retrieval_similarity",
                        0.0
                    )
                ),

            "speaker":
                evidence.get(
                    "speaker"
                ),

            "start":
                evidence.get(
                    "start"
                ),

            "end":
                evidence.get(
                    "end"
                ),

            "text":
                evidence_text,

            "nli_label":
                nli_result["label"],

            "nli_entailment":
                nli_result["entailment"],

            "nli_contradiction":
                nli_result["contradiction"],

            "nli_neutral":
                nli_result["neutral"],

            "lexical_overlap":
                lexical_overlap,

            "content_status":
                content_status,

            "speaker_consistent":
                speaker_result.get(
                    "passed",
                    False
                ),

            "event_type_consistent":
                event_result.get(
                    "passed",
                    False
                )
        })


    # --------------------------------------------------------
    # Select content-supported evidence
    # --------------------------------------------------------

    supported = [
        item
        for item in candidate_results
        if item["content_status"] == "SUPPORTED"
    ]

    contradicted = [
        item
        for item in candidate_results
        if item["content_status"] == "CONTRADICTED"
    ]


    # --------------------------------------------------------
    # Priority:
    #
    # SUPPORTED > CONTRADICTED > UNCERTAIN
    # --------------------------------------------------------

    if supported:

        best = max(
            supported,
            key=lambda x: (
                x["nli_entailment"],
                x["retrieval_similarity"]
            )
        )

        content_status = "SUPPORTED"

    elif contradicted:

        best = max(
            contradicted,
            key=lambda x: (
                x["nli_contradiction"],
                x["lexical_overlap"]
            )
        )

        content_status = "CONTRADICTED"

    else:

        best = max(
            candidate_results,
            key=lambda x: (
                x["retrieval_similarity"],
                x["lexical_overlap"]
            )
        )

        content_status = "UNCERTAIN"


    # --------------------------------------------------------
    # Overall multi-attribute status
    # --------------------------------------------------------
    #
    # IMPORTANT:
    # Content support alone does NOT make the claim VERIFIED.
    #
    # Speaker and event-type consistency are independent
    # attributes.
    # --------------------------------------------------------

    failed_attributes = []

    if content_status != "SUPPORTED":

        failed_attributes.append(
            "content_consistency"
        )

    if not best["speaker_consistent"]:

        failed_attributes.append(
            "speaker_consistency"
        )

    if not best["event_type_consistent"]:

        failed_attributes.append(
            "event_type_consistency"
        )


    if len(failed_attributes) == 0:

        overall_status = "VERIFIED"

    else:

        overall_status = "FLAGGED"


    result = {

        "claim_index":
            claim_index,

        "claim":
            claim_text,

        "claim_speaker":
            claim_speaker,

        "claim_type":
            claim_type,

        "content_status":
            content_status,

        "overall_status":
            overall_status,

        "failed_attributes":
            failed_attributes,

        "best_candidate":
            best,

        "all_candidates":
            candidate_results
    }


    r7_results.append(
        result
    )


    # --------------------------------------------------------
    # Display candidate-level results
    # --------------------------------------------------------

    for candidate in candidate_results:

        print(
            f"\nRank {candidate['rank']}"
        )

        print(
            "Evidence ID:",
            candidate["evidence_id"]
        )

        print(
            "BGE:",
            round(
                candidate[
                    "retrieval_similarity"
                ],
                4
            )
        )

        print(
            "Speaker:",
            candidate["speaker"]
        )

        print(
            "NLI:",
            candidate["nli_label"],
            "| entailment:",
            round(
                candidate[
                    "nli_entailment"
                ],
                4
            ),
            "| contradiction:",
            round(
                candidate[
                    "nli_contradiction"
                ],
                4
            )
        )

        print(
            "Lexical:",
            round(
                candidate[
                    "lexical_overlap"
                ],
                4
            )
        )

        print(
            "Content:",
            candidate[
                "content_status"
            ]
        )

        print(
            "Speaker:",
            "PASS"
            if candidate[
                "speaker_consistent"
            ]
            else "FAIL"
        )

        print(
            "Event:",
            "PASS"
            if candidate[
                "event_type_consistent"
            ]
            else "FAIL"
        )

        print(
            "Evidence:",
            candidate["text"]
        )


    # --------------------------------------------------------
    # Final claim result
    # --------------------------------------------------------

    print("\n" + "-" * 90)

    print(
        "SELECTED EVIDENCE:"
    )

    print(
        "Rank:",
        best["rank"]
    )

    print(
        "Content:",
        content_status
    )

    print(
        "Overall:",
        overall_status
    )

    print(
        "Failed attributes:",
        failed_attributes
    )


# ------------------------------------------------------------
# 3. Summary
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("STEP 12O-R7 SUMMARY")
print("=" * 90)

for result in r7_results:

    print(
        f"\nClaim {result['claim_index']}: "
        f"{result['overall_status']}"
    )

    print(
        "Content:",
        result["content_status"]
    )

    print(
        "Failed:",
        result["failed_attributes"]
    )

    best = result["best_candidate"]

    print(
        "Selected evidence:",
        f"Rank {best['rank']} | "
        f"Speaker {best['speaker']} | "
        f"BGE {best['retrieval_similarity']:.4f}"
    )

print("\n" + "=" * 90)
print("STEP 12O-R7 COMPLETE")
print("=" * 90)
````

````python
# ============================================================
# STEP 12O-R10 — NLI PAIR-ORDER VALIDATION
# ============================================================

print("=" * 90)
print("STEP 12O-R10 — NLI PAIR-ORDER VALIDATION")
print("=" * 90)

test_pairs = [
    {
        "name": "25 EURO",
        "claim": "The remote control is expected to sell for 25 euros.",
        "evidence": "The selling price is at 25 euros."
    },
    {
        "name": "12.50 PRODUCTION COST",
        "claim": "The remote control's production costs are estimated at 12.50 euros.",
        "evidence": "The production costs will be 12.50 euros."
    },
    {
        "name": "INTERNATIONAL MARKET",
        "claim": "The remote control is intended for an international market.",
        "evidence": "It is going to be in the international market like Australia, America."
    }
]


def run_nli_pair(text_a, text_b):
    inputs = nli_tokenizer(
        text_a,
        text_b,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    inputs = {
        key: value.to(nli_device)
        for key, value in inputs.items()
    }

    with torch.no_grad():
        outputs = nli_model(**inputs)

        probabilities = torch.softmax(
            outputs.logits,
            dim=-1
        )[0]

    scores = {
        "contradiction": float(probabilities[0]),
        "entailment": float(probabilities[1]),
        "neutral": float(probabilities[2])
    }

    label = max(
        scores,
        key=scores.get
    )

    return {
        "label": label,
        **scores
    }


for pair in test_pairs:

    print("\n" + "-" * 90)
    print(pair["name"])
    print("-" * 90)

    print("\nCLAIM:")
    print(pair["claim"])

    print("\nEVIDENCE:")
    print(pair["evidence"])

    # Evidence as premise, claim as hypothesis
    evidence_first = run_nli_pair(
        pair["evidence"],
        pair["claim"]
    )

    # Claim as premise, evidence as hypothesis
    claim_first = run_nli_pair(
        pair["claim"],
        pair["evidence"]
    )

    print("\n[Evidence → Claim]")
    print(evidence_first)

    print("\n[Claim → Evidence]")
    print(claim_first)


print("\n" + "=" * 90)
print("STEP 12O-R10 COMPLETE")
print("=" * 90)
````

````python
# ============================================================
# STEP 12O-R11 — VALIDATE NLI ON ACTUAL RETRIEVED EVIDENCE
# ============================================================

print("=" * 90)
print("STEP 12O-R11 — VALIDATE NLI ON ACTUAL RETRIEVED EVIDENCE")
print("=" * 90)


# Actual claim/evidence pairs observed during R7
test_pairs = [
    {
        "name": "25 EURO",
        "claim": "The remote control is expected to sell for 25 euros.",
        "evidence": "selling price at 25 euros"
    },
    {
        "name": "12.50 PRODUCTION COST",
        "claim": "The remote control's production costs are estimated at 12.50 euros.",
        "evidence": "production costs at 12.50"
    },
    {
        "name": "INTERNATIONAL MARKET",
        "claim": "The remote control is intended for an international market.",
        "evidence": "going to be in the international market like Australia, America"
    }
]


def run_nli_corrected(
    claim_text,
    evidence_text
):
    """
    Corrected NLI test:
    claim = first sequence
    evidence = second sequence
    """

    inputs = nli_tokenizer(
        claim_text,
        evidence_text,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    inputs = {
        key: value.to(nli_device)
        for key, value in inputs.items()
    }

    with torch.no_grad():

        outputs = nli_model(**inputs)

        probabilities = torch.softmax(
            outputs.logits,
            dim=-1
        )[0]

    scores = {
        "contradiction": float(probabilities[0]),
        "entailment": float(probabilities[1]),
        "neutral": float(probabilities[2])
    }

    label = max(
        scores,
        key=scores.get
    )

    return {
        "label": label,
        **scores
    }


for pair in test_pairs:

    result = run_nli_corrected(
        claim_text=pair["claim"],
        evidence_text=pair["evidence"]
    )

    print("\n" + "-" * 90)
    print(pair["name"])
    print("-" * 90)

    print("\nCLAIM:")
    print(pair["claim"])

    print("\nACTUAL RETRIEVED EVIDENCE:")
    print(pair["evidence"])

    print("\nNLI RESULT:")
    print(result)


print("\n" + "=" * 90)
print("STEP 12O-R11 COMPLETE")
print("=" * 90)
````

````python
# STEP 12O-R18 — INTEGRATE VALIDATED M10 CLASSIFIER
# This adds only the M10 atomic decision rule.
# No existing functions/files are deleted.

from pathlib import Path

PIPELINE_PATH = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project/"
    "07_dashboard/meeting_pipeline.py"
)

source = PIPELINE_PATH.read_text(encoding="utf-8")

if "def classify_atomic_candidate(" in source:
    print("✓ M10 classifier already exists in meeting_pipeline.py")
else:

    m10_code = '''

# ============================================================
# M10 — MULTI-ATTRIBUTE EVIDENCE CONSISTENCY
# Atomic evidence classification
# ============================================================

def classify_atomic_candidate(
    candidate,
    entailment_threshold=0.80,
    contradiction_threshold=0.80,
    relevance_threshold=0.20,
):
    """
    Classify one retrieved evidence candidate.

    Rules:
      1. Strong entailment -> SUPPORTED
      2. Strong contradiction requires relevance
      3. Irrelevant contradiction -> UNCERTAIN
      4. Otherwise -> UNCERTAIN

    Retrieval similarity is not treated as verification.
    """

    nli_label = str(
        candidate.get("nli_label", "")
    ).lower()

    entailment = float(
        candidate.get("nli_entailment", 0.0)
    )

    contradiction = float(
        candidate.get("nli_contradiction", 0.0)
    )

    lexical_overlap = float(
        candidate.get("lexical_overlap", 0.0)
    )

    # --------------------------------------------------------
    # Strong semantic entailment
    # --------------------------------------------------------

    if (
        nli_label == "entailment"
        and entailment >= entailment_threshold
    ):
        return "SUPPORTED"

    # --------------------------------------------------------
    # Relevant contradiction
    # --------------------------------------------------------

    if (
        nli_label == "contradiction"
        and contradiction >= contradiction_threshold
        and lexical_overlap >= relevance_threshold
    ):
        return "CONTRADICTED"

    # --------------------------------------------------------
    # Everything else
    # --------------------------------------------------------

    return "UNCERTAIN"


'''

    # Append to the clean pipeline rather than altering
    # existing validated functions.
    PIPELINE_PATH.write_text(
        source.rstrip() + "\n" + m10_code,
        encoding="utf-8"
    )

    print("✓ M10 classifier integrated.")

print("\nPipeline:", PIPELINE_PATH)
print("✓ Existing code preserved.")
print("✓ No files/cells deleted.")
print("✓ M10 thresholds: 0.80 / 0.80 / 0.20")
````

````python
# STEP 12O-R20 — INTEGRATE EVIDENCE ANALYSIS INTO PIPELINE

from pathlib import Path

PIPELINE_PATH = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project/"
    "07_dashboard/meeting_pipeline.py"
)

source = PIPELINE_PATH.read_text(encoding="utf-8")

if "def analyze_evidence_candidate(" in source:
    print("✓ Evidence analysis already exists in meeting_pipeline.py")

else:

    analysis_code = '''

# ============================================================
# M10 — EVIDENCE ANALYSIS
# BGE retrieval + DeBERTa NLI + lexical support
# ============================================================

def analyze_evidence_candidate(
    proposition,
    evidence,
    claim_speaker=None,
):
    """
    Analyze one retrieved evidence candidate.

    Combines:
      - retrieval metadata
      - DeBERTa NLI
      - lexical support
      - speaker consistency

    This function does NOT make the final
    SUPPORTED / CONTRADICTED / UNCERTAIN decision.
    """

    evidence_text = str(
        evidence.get("text", "")
    ).strip()

    # --------------------------------------------------------
    # NLI
    # --------------------------------------------------------

    nli_result = run_nli(
        evidence_text,
        proposition
    )

    # --------------------------------------------------------
    # Lexical support
    # --------------------------------------------------------

    lexical_result = lexical_support(
        proposition,
        evidence_text
    )

    # Support both known lexical-result schemas.
    lexical_overlap = float(
        lexical_result.get(
            "lexical_overlap",
            lexical_result.get(
                "overlap_ratio",
                0.0
            )
        )
    )

    # --------------------------------------------------------
    # Speaker consistency
    # --------------------------------------------------------

    evidence_speaker = evidence.get(
        "speaker"
    )

    speaker_match = None

    if claim_speaker is not None:
        speaker_match = (
            evidence_speaker == claim_speaker
            or claim_speaker == "MULTIPLE"
        )

    # --------------------------------------------------------
    # Standardized candidate record
    # --------------------------------------------------------

    return {
        "evidence_id":
            evidence.get("evidence_id"),

        "speaker":
            evidence_speaker,

        "start":
            evidence.get("start"),

        "end":
            evidence.get("end"),

        "text":
            evidence_text,

        "retrieval_similarity":
            float(
                evidence.get(
                    "retrieval_similarity",
                    0.0
                )
            ),

        "nli_label":
            str(
                nli_result.get(
                    "nli_label",
                    nli_result.get(
                        "predicted_label",
                        ""
                    )
                )
            ).lower(),

        "nli_entailment":
            float(
                nli_result.get(
                    "entailment",
                    nli_result.get(
                        "entailment_probability",
                        0.0
                    )
                )
            ),

        "nli_contradiction":
            float(
                nli_result.get(
                    "contradiction",
                    nli_result.get(
                        "contradiction_probability",
                        0.0
                    )
                )
            ),

        "nli_neutral":
            float(
                nli_result.get(
                    "neutral",
                    nli_result.get(
                        "neutral_probability",
                        0.0
                    )
                )
            ),

        "lexical_overlap":
            lexical_overlap,

        "matched_terms":
            lexical_result.get(
                "matched_terms",
                []
            ),

        "missing_terms":
            lexical_result.get(
                "missing_terms",
                []
            ),

        "speaker_match":
            speaker_match,
    }


'''

    PIPELINE_PATH.write_text(
        source.rstrip() + "\n" + analysis_code,
        encoding="utf-8"
    )

    print("✓ Evidence analysis integrated.")

print("\nPipeline:", PIPELINE_PATH)
print("✓ Existing code preserved.")
print("✓ No files/cells deleted.")
````

````python
# STEP 12O-R22 — INTEGRATE DEBERTA NLI

from pathlib import Path

PIPELINE_PATH = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project/"
    "07_dashboard/meeting_pipeline.py"
)

source = PIPELINE_PATH.read_text(encoding="utf-8")

if "def load_nli_model(" in source and "def run_nli(" in source:
    print("✓ NLI functions already exist in meeting_pipeline.py")

else:

    nli_code = '''

# ============================================================
# M10 — DEBERTA NLI
# ============================================================

NLI_MODEL_NAME = "cross-encoder/nli-deberta-v3-small"


def load_nli_model(device=None):
    """
    Load the DeBERTa-v3 NLI model used for
    evidence consistency verification.
    """

    from transformers import (
        AutoTokenizer,
        AutoModelForSequenceClassification,
    )

    if device is None:
        device = (
            "cuda"
            if torch.cuda.is_available()
            else "cpu"
        )

    tokenizer = AutoTokenizer.from_pretrained(
        NLI_MODEL_NAME
    )

    model = AutoModelForSequenceClassification.from_pretrained(
        NLI_MODEL_NAME
    )

    model = model.to(device)
    model.eval()

    return tokenizer, model, device


def run_nli(
    evidence_text,
    proposition,
    nli_tokenizer,
    nli_model,
    device,
):
    """
    Run DeBERTa-v3 NLI.

    Ordering follows the validated project implementation:
        premise    = evidence
        hypothesis = proposition

    Returns contradiction, entailment and neutral
    probabilities.
    """

    inputs = nli_tokenizer(
        evidence_text,
        proposition,
        return_tensors="pt",
        truncation=True,
        max_length=512,
    )

    inputs = {
        key: value.to(device)
        for key, value in inputs.items()
    }

    with torch.no_grad():
        outputs = nli_model(**inputs)

    probabilities = (
        torch.softmax(
            outputs.logits,
            dim=-1
        )[0]
        .cpu()
        .tolist()
    )

    label_map = {
        0: "contradiction",
        1: "entailment",
        2: "neutral",
    }

    predicted_id = int(
        torch.argmax(
            outputs.logits,
            dim=-1
        )[0]
        .cpu()
    )

    return {
        "nli_label": label_map[predicted_id],
        "contradiction": float(probabilities[0]),
        "entailment": float(probabilities[1]),
        "neutral": float(probabilities[2]),
    }

'''

    PIPELINE_PATH.write_text(
        source.rstrip() + "\n" + nli_code,
        encoding="utf-8"
    )

    print("✓ DeBERTa NLI integrated.")

print("\nPipeline:", PIPELINE_PATH)
print("✓ Existing code preserved.")
print("✓ No files/cells deleted.")
print("✓ NLI model:", "cross-encoder/nli-deberta-v3-small")
````

````python
# STEP 12O-R23 — INTEGRATE LEXICAL SUPPORT

from pathlib import Path
import re

PIPELINE_PATH = Path(
    "/content/drive/MyDrive/MTechIndProj/MTechIndProj/"
    "MoM_Project/07_dashboard/meeting_pipeline.py"
)

# Correct path if the duplicated folder segment does not exist
if not PIPELINE_PATH.exists():
    PIPELINE_PATH = Path(
        "/content/drive/MyDrive/MTechIndProj/MoM_Project/"
        "07_dashboard/meeting_pipeline.py"
    )

source = PIPELINE_PATH.read_text(encoding="utf-8")

if "def lexical_support(" in source:
    print("✓ lexical_support already exists in meeting_pipeline.py")

else:

    lexical_code = '''

# ============================================================
# M10 — LEXICAL SUPPORT
# ============================================================

def lexical_support(
    proposition,
    evidence_text,
):
    """
    Measure content-word overlap between a proposition
    and an evidence passage.

    This is a supporting relevance signal.
    It is NOT a standalone verification decision.
    """

    proposition_text = str(
        proposition or ""
    ).lower()

    evidence_text = str(
        evidence_text or ""
    ).lower()

    proposition_terms = set(
        re.findall(
            r"\\\\b[a-zA-Z0-9]+\\\\b",
            proposition_text,
        )
    )

    evidence_terms = set(
        re.findall(
            r"\\\\b[a-zA-Z0-9]+\\\\b",
            evidence_text,
        )
    )

    # Remove very common function words.
    stopwords = {
        "the", "a", "an", "is", "are",
        "was", "were", "be", "been",
        "being", "to", "of", "and",
        "or", "in", "on", "at",
        "for", "with", "by", "from",
        "this", "that", "these", "those",
        "it", "its", "as",
    }

    proposition_terms = {
        term
        for term in proposition_terms
        if term not in stopwords
    }

    evidence_terms = {
        term
        for term in evidence_terms
        if term not in stopwords
    }

    if not proposition_terms:
        return {
            "lexical_overlap": 0.0,
            "matched_terms": [],
            "missing_terms": [],
        }

    matched_terms = sorted(
        proposition_terms.intersection(
            evidence_terms
        )
    )

    missing_terms = sorted(
        proposition_terms.difference(
            evidence_terms
        )
    )

    overlap = (
        len(matched_terms)
        / len(proposition_terms)
    )

    return {
        "lexical_overlap": float(overlap),
        "matched_terms": matched_terms,
        "missing_terms": missing_terms,
    }

'''

    PIPELINE_PATH.write_text(
        source.rstrip() + "\n" + lexical_code,
        encoding="utf-8"
    )

    print("✓ Lexical support integrated.")

print("\nPipeline:", PIPELINE_PATH)
print("✓ Existing code preserved.")
print("✓ No files/cells deleted.")
````

````python
# STEP 12O-R27 — INTEGRATE MULTI-EVIDENCE ATOMIC VERIFICATION

from pathlib import Path

PIPELINE_PATH = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project/"
    "07_dashboard/meeting_pipeline.py"
)

source = PIPELINE_PATH.read_text(encoding="utf-8")

if "def verify_atomic_proposition(" in source:
    print("✓ verify_atomic_proposition already exists in meeting_pipeline.py")

else:

    verification_code = '''

# ============================================================
# M10 — MULTI-EVIDENCE ATOMIC VERIFICATION
# ============================================================

def verify_atomic_proposition(
    proposition,
    bge_model,
    faiss_index,
    evidence_documents,
    claim_speaker=None,
    top_k=10,
):
    """
    Verify one atomic proposition against multiple
    retrieved evidence candidates.

    Decision priority:
        SUPPORTED > CONTRADICTED > UNCERTAIN

    The validated M10 thresholds are preserved:
        entailment       >= 0.80
        contradiction    >= 0.80
        lexical relevance >= 0.20
    """

    proposition = str(
        proposition or ""
    ).strip()

    if not proposition:
        return {
            "proposition": proposition,
            "status": "UNCERTAIN",
            "selected_evidence": None,
            "candidates": [],
        }

    # --------------------------------------------------------
    # Retrieve Top-K evidence
    # --------------------------------------------------------

    retrieved = retrieve_evidence(
        proposition,
        bge_model=bge_model,
        faiss_index=faiss_index,
        evidence_documents=evidence_documents,
        top_k=top_k,
    )

    analyzed_candidates = []

    # --------------------------------------------------------
    # Analyze every retrieved candidate
    # --------------------------------------------------------

    for evidence in retrieved:

        analyzed = analyze_evidence_candidate(
            proposition=proposition,
            evidence=evidence,
            claim_speaker=claim_speaker,
        )

        status = classify_atomic_candidate(
            analyzed
        )

        analyzed["status"] = status

        analyzed_candidates.append(
            analyzed
        )

    # --------------------------------------------------------
    # Select strongest candidate
    #
    # Priority:
    #   1. SUPPORTED
    #   2. CONTRADICTED
    #   3. UNCERTAIN
    # --------------------------------------------------------

    supported = [
        candidate
        for candidate in analyzed_candidates
        if candidate["status"] == "SUPPORTED"
    ]

    contradicted = [
        candidate
        for candidate in analyzed_candidates
        if candidate["status"] == "CONTRADICTED"
    ]

    uncertain = [
        candidate
        for candidate in analyzed_candidates
        if candidate["status"] == "UNCERTAIN"
    ]

    selected = None
    overall_status = "UNCERTAIN"

    if supported:

        selected = max(
            supported,
            key=lambda candidate: (
                candidate["nli_entailment"],
                candidate["retrieval_similarity"],
            ),
        )

        overall_status = "SUPPORTED"

    elif contradicted:

        selected = max(
            contradicted,
            key=lambda candidate: (
                candidate["nli_contradiction"],
                candidate["lexical_overlap"],
                candidate["retrieval_similarity"],
            ),
        )

        overall_status = "CONTRADICTED"

    elif uncertain:

        selected = max(
            uncertain,
            key=lambda candidate: (
                candidate["retrieval_similarity"],
                candidate["lexical_overlap"],
            ),
        )

    # --------------------------------------------------------
    # Return complete verification record
    # --------------------------------------------------------

    return {
        "proposition": proposition,
        "status": overall_status,
        "selected_evidence": selected,
        "candidates": analyzed_candidates,
        "retrieved_count": len(retrieved),
        "supported_count": len(supported),
        "contradicted_count": len(contradicted),
        "uncertain_count": len(uncertain),
    }

'''

    PIPELINE_PATH.write_text(
        source.rstrip() + "\n" + verification_code,
        encoding="utf-8"
    )

    print("✓ Multi-evidence atomic verification integrated.")

print("\nPipeline:", PIPELINE_PATH)
print("✓ Existing code preserved.")
print("✓ No files/cells deleted.")
print("✓ M10 priority preserved:")
print("  SUPPORTED > CONTRADICTED > UNCERTAIN")
````

````python
# STEP 12O-R28 — INTEGRATE SPEAKER CONSISTENCY

from pathlib import Path

PIPELINE_PATH = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project/"
    "07_dashboard/meeting_pipeline.py"
)

source = PIPELINE_PATH.read_text(encoding="utf-8")

if "def check_speaker_consistency_pipeline(" in source:
    print("✓ Speaker consistency already exists in meeting_pipeline.py")

else:

    speaker_code = '''

# ============================================================
# MULTI-ATTRIBUTE CONSISTENCY — SPEAKER
# ============================================================

def check_speaker_consistency_pipeline(
    claim_speaker,
    evidence_speaker,
):
    """
    Check whether the speaker attributed to a claim
    matches the speaker of the selected evidence.

    MULTIPLE is treated as a valid multi-speaker claim.
    """

    if claim_speaker is None:
        return {
            "status": "NOT_EVALUATED",
            "claim_speaker": None,
            "evidence_speaker": evidence_speaker,
        }

    if evidence_speaker is None:
        return {
            "status": "NOT_EVALUATED",
            "claim_speaker": claim_speaker,
            "evidence_speaker": None,
        }

    if claim_speaker == "MULTIPLE":
        status = "PASS"

    elif claim_speaker == evidence_speaker:
        status = "PASS"

    else:
        status = "FAIL"

    return {
        "status": status,
        "claim_speaker": claim_speaker,
        "evidence_speaker": evidence_speaker,
    }

'''

    PIPELINE_PATH.write_text(
        source.rstrip() + "\n" + speaker_code,
        encoding="utf-8"
    )

    print("✓ Speaker consistency integrated.")

print("\nPipeline:", PIPELINE_PATH)
print("✓ Existing code preserved.")
print("✓ No files/cells deleted.")
````

````python
# STEP 12O-R29 — INTEGRATE EVENT-TYPE CONSISTENCY

from pathlib import Path

PIPELINE_PATH = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project/"
    "07_dashboard/meeting_pipeline.py"
)

source = PIPELINE_PATH.read_text(encoding="utf-8")

if "def check_event_type_consistency_pipeline(" in source:
    print("✓ Event-type consistency already exists in meeting_pipeline.py")

else:

    event_type_code = '''

# ============================================================
# MULTI-ATTRIBUTE CONSISTENCY — EVENT TYPE
# ============================================================

def check_event_type_consistency_pipeline(
    claim_event_type,
    evidence_event_type,
):
    """
    Check whether the event type attributed to a claim
    is consistent with the event type of the evidence.

    Supported current schemas:
        - event_type
        - candidate_type

    Event types are compared as normalized labels.
    """

    if claim_event_type is None:
        return {
            "status": "NOT_EVALUATED",
            "claim_event_type": None,
            "evidence_event_type": evidence_event_type,
        }

    if evidence_event_type is None:
        return {
            "status": "NOT_EVALUATED",
            "claim_event_type": claim_event_type,
            "evidence_event_type": None,
        }

    claim_type = str(
        claim_event_type
    ).strip().upper()

    evidence_type = str(
        evidence_event_type
    ).strip().upper()

    if claim_type == evidence_type:
        status = "PASS"
    else:
        status = "FAIL"

    return {
        "status": status,
        "claim_event_type": claim_type,
        "evidence_event_type": evidence_type,
    }


def get_claim_event_type(claim):
    """
    Resolve the event type from the current claim schemas.

    Priority:
        event_type → candidate_type → type

    The legacy `type` field is retained only as a
    compatibility fallback.
    """

    if not isinstance(claim, dict):
        return None

    if claim.get("event_type") is not None:
        return claim.get("event_type")

    if claim.get("candidate_type") is not None:
        return claim.get("candidate_type")

    if claim.get("type") is not None:
        return claim.get("type")

    return None


def get_evidence_event_type(evidence):
    """
    Resolve event type from an evidence record.

    Current evidence records may carry `event_type`
    or `candidate_type`.
    """

    if not isinstance(evidence, dict):
        return None

    if evidence.get("event_type") is not None:
        return evidence.get("event_type")

    if evidence.get("candidate_type") is not None:
        return evidence.get("candidate_type")

    return None

'''

    PIPELINE_PATH.write_text(
        source.rstrip() + "\n" + event_type_code,
        encoding="utf-8"
    )

    print("✓ Event-type consistency integrated.")

print("\nPipeline:", PIPELINE_PATH)
print("✓ Existing code preserved.")
print("✓ No files/cells deleted.")
print("✓ Current schema support:")
print("  event_type → candidate_type → legacy type")
````

````python
# STEP 12O-R30 — INTEGRATE TEMPORAL CONSISTENCY

from pathlib import Path

PIPELINE_PATH = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project/"
    "07_dashboard/meeting_pipeline.py"
)

source = PIPELINE_PATH.read_text(encoding="utf-8")

if "def check_timestamp_consistency_pipeline(" in source:
    print("✓ Temporal consistency already exists in meeting_pipeline.py")

else:

    temporal_code = '''

# ============================================================
# MULTI-ATTRIBUTE CONSISTENCY — TEMPORAL
# ============================================================

def check_timestamp_consistency_pipeline(
    claim_start,
    claim_end,
    evidence_start,
    evidence_end,
    tolerance_seconds=5.0,
):
    """
    Check whether a claim timestamp is consistent with
    the timestamp of its selected evidence.

    The evidence is considered temporally consistent when
    the claim interval and evidence interval overlap, or
    when their boundaries are within the configured tolerance.

    If usable timestamps are unavailable, return
    NOT_EVALUATED rather than forcing a decision.
    """

    # --------------------------------------------------------
    # Validate timestamps
    # --------------------------------------------------------

    if (
        claim_start is None
        or claim_end is None
        or evidence_start is None
        or evidence_end is None
    ):
        return {
            "status": "NOT_EVALUATED",
            "claim_start": claim_start,
            "claim_end": claim_end,
            "evidence_start": evidence_start,
            "evidence_end": evidence_end,
            "temporal_distance": None,
        }

    try:
        claim_start = float(claim_start)
        claim_end = float(claim_end)
        evidence_start = float(evidence_start)
        evidence_end = float(evidence_end)
        tolerance_seconds = float(tolerance_seconds)
    except (TypeError, ValueError):
        return {
            "status": "NOT_EVALUATED",
            "claim_start": claim_start,
            "claim_end": claim_end,
            "evidence_start": evidence_start,
            "evidence_end": evidence_end,
            "temporal_distance": None,
        }

    # --------------------------------------------------------
    # Normalize reversed intervals
    # --------------------------------------------------------

    if claim_end < claim_start:
        claim_start, claim_end = claim_end, claim_start

    if evidence_end < evidence_start:
        evidence_start, evidence_end = (
            evidence_end,
            evidence_start,
        )

    # --------------------------------------------------------
    # Calculate interval relationship
    # --------------------------------------------------------

    overlap = max(
        0.0,
        min(claim_end, evidence_end)
        - max(claim_start, evidence_start),
    )

    if overlap > 0.0:
        return {
            "status": "PASS",
            "claim_start": claim_start,
            "claim_end": claim_end,
            "evidence_start": evidence_start,
            "evidence_end": evidence_end,
            "temporal_distance": 0.0,
            "overlap_seconds": overlap,
        }

    # No overlap — calculate the gap between intervals.
    if claim_end < evidence_start:
        temporal_distance = evidence_start - claim_end
    else:
        temporal_distance = claim_start - evidence_end

    if temporal_distance <= tolerance_seconds:
        status = "PASS"
    else:
        status = "FAIL"

    return {
        "status": status,
        "claim_start": claim_start,
        "claim_end": claim_end,
        "evidence_start": evidence_start,
        "evidence_end": evidence_end,
        "temporal_distance": temporal_distance,
        "overlap_seconds": 0.0,
    }

'''

    PIPELINE_PATH.write_text(
        source.rstrip() + "\n" + temporal_code,
        encoding="utf-8"
    )

    print("✓ Temporal consistency integrated.")

print("\nPipeline:", PIPELINE_PATH)
print("✓ Existing code preserved.")
print("✓ No files/cells deleted.")
print("✓ Missing timestamps remain NOT_EVALUATED.")
print("✓ Default tolerance:", "5.0 seconds")
````

````python
# STEP 12O-R31 — INTEGRATE MULTI-ATTRIBUTE EVIDENCE CONSISTENCY

from pathlib import Path

PIPELINE_PATH = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project/"
    "07_dashboard/meeting_pipeline.py"
)

source = PIPELINE_PATH.read_text(encoding="utf-8")

if "def verify_multi_attribute_claim(" in source:
    print("✓ Multi-attribute verification already exists in meeting_pipeline.py")

else:

    multi_attribute_code = '''

# ============================================================
# MULTI-ATTRIBUTE EVIDENCE CONSISTENCY
# ============================================================

def verify_multi_attribute_claim(
    claim,
    bge_model,
    faiss_index,
    evidence_documents,
    top_k=10,
    temporal_tolerance_seconds=5.0,
):
    """
    Perform multi-attribute evidence consistency verification.

    Attributes:
        1. Content
        2. Speaker
        3. Event type
        4. Temporal consistency

    Final decision:
        VERIFIED  -> all applicable attributes pass
        FLAGGED   -> one or more applicable attributes fail

    NOT_EVALUATED attributes are not treated as failures.
    """

    if not isinstance(claim, dict):
        raise TypeError(
            "claim must be a dictionary"
        )

    claim_text = str(
        claim.get(
            "claim",
            claim.get("claim_text", "")
        )
    ).strip()

    claim_speaker = claim.get(
        "speaker"
    )

    claim_event_type = get_claim_event_type(
        claim
    )

    claim_start = claim.get(
        "start"
    )

    claim_end = claim.get(
        "end"
    )

    # --------------------------------------------------------
    # 1. CONTENT / EVIDENCE VERIFICATION
    # --------------------------------------------------------

    content_result = verify_atomic_proposition(
        proposition=claim_text,
        bge_model=bge_model,
        faiss_index=faiss_index,
        evidence_documents=evidence_documents,
        claim_speaker=claim_speaker,
        top_k=top_k,
    )

    selected_evidence = content_result.get(
        "selected_evidence"
    )

    content_status = content_result.get(
        "status",
        "UNCERTAIN"
    )

    # --------------------------------------------------------
    # 2. SPEAKER CONSISTENCY
    # --------------------------------------------------------

    if selected_evidence is not None:

        speaker_result = (
            check_speaker_consistency_pipeline(
                claim_speaker=claim_speaker,
                evidence_speaker=selected_evidence.get(
                    "speaker"
                ),
            )
        )

    else:

        speaker_result = {
            "status": "NOT_EVALUATED",
            "claim_speaker": claim_speaker,
            "evidence_speaker": None,
        }

    # --------------------------------------------------------
    # 3. EVENT-TYPE CONSISTENCY
    # --------------------------------------------------------

    if selected_evidence is not None:

        evidence_event_type = (
            get_evidence_event_type(
                selected_evidence
            )
        )

        event_result = (
            check_event_type_consistency_pipeline(
                claim_event_type=claim_event_type,
                evidence_event_type=evidence_event_type,
            )
        )

    else:

        event_result = {
            "status": "NOT_EVALUATED",
            "claim_event_type": claim_event_type,
            "evidence_event_type": None,
        }

    # --------------------------------------------------------
    # 4. TEMPORAL CONSISTENCY
    # --------------------------------------------------------

    if selected_evidence is not None:

        temporal_result = (
            check_timestamp_consistency_pipeline(
                claim_start=claim_start,
                claim_end=claim_end,
                evidence_start=selected_evidence.get(
                    "start"
                ),
                evidence_end=selected_evidence.get(
                    "end"
                ),
                tolerance_seconds=(
                    temporal_tolerance_seconds
                ),
            )
        )

    else:

        temporal_result = {
            "status": "NOT_EVALUATED",
            "claim_start": claim_start,
            "claim_end": claim_end,
            "evidence_start": None,
            "evidence_end": None,
            "temporal_distance": None,
        }

    # --------------------------------------------------------
    # ATTRIBUTE STATUS COLLECTION
    # --------------------------------------------------------

    attribute_results = {
        "content": content_status,
        "speaker": speaker_result["status"],
        "event_type": event_result["status"],
        "temporal": temporal_result["status"],
    }

    # --------------------------------------------------------
    # FINAL MULTI-ATTRIBUTE DECISION
    #
    # Content must be SUPPORTED.
    # Every applicable additional attribute must PASS.
    # NOT_EVALUATED attributes do not cause failure.
    # --------------------------------------------------------

    content_pass = (
        content_status == "SUPPORTED"
    )

    additional_failures = []

    for attribute_name in [
        "speaker",
        "event_type",
        "temporal",
    ]:

        attribute_status = attribute_results[
            attribute_name
        ]

        if attribute_status == "FAIL":
            additional_failures.append(
                attribute_name
            )

    if (
        content_pass
        and not additional_failures
    ):
        overall_status = "VERIFIED"
    else:
        overall_status = "FLAGGED"

    # --------------------------------------------------------
    # RETURN COMPLETE VERIFICATION RECORD
    # --------------------------------------------------------

    return {
        "claim": claim,
        "claim_text": claim_text,

        "overall_status": overall_status,

        "content": {
            "status": content_status,
            "verification": content_result,
        },

        "speaker": speaker_result,

        "event_type": event_result,

        "temporal": temporal_result,

        "attribute_results": attribute_results,

        "failed_attributes": additional_failures,

        "selected_evidence": selected_evidence,
    }

'''

    PIPELINE_PATH.write_text(
        source.rstrip() + "\n" + multi_attribute_code,
        encoding="utf-8"
    )

    print("✓ Multi-attribute verification integrated.")

print("\nPipeline:", PIPELINE_PATH)
print("✓ Existing code preserved.")
print("✓ No files/cells deleted.")
print("✓ Attributes combined:")
print("  Content")
print("  Speaker")
print("  Event type")
print("  Temporal")
print("✓ Final statuses: VERIFIED / FLAGGED")
````

````python
# STEP 12O-R31-FIX — RESOLVE EVENT TYPE FROM V3 CANDIDATES

from pathlib import Path

PIPELINE_PATH = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project/"
    "07_dashboard/meeting_pipeline.py"
)

source = PIPELINE_PATH.read_text(encoding="utf-8")

if "def resolve_evidence_event_type_pipeline(" in source:
    print("✓ Evidence event-type resolution already exists.")

else:

    event_resolution_code = '''

# ============================================================
# EVENT-TYPE RESOLUTION FOR EVIDENCE
# ============================================================

def resolve_evidence_event_type_pipeline(
    evidence,
    source_candidates=None,
):
    """
    Resolve the event type associated with a retrieved
    evidence utterance using temporal overlap with the
    V3 MoM candidate set.

    The retrieved evidence itself is still the source of
    content verification. V3 candidates provide the
    event-type metadata.
    """

    if not source_candidates:
        return None

    evidence_start = evidence.get("start")
    evidence_end = evidence.get("end")

    if (
        evidence_start is None
        or evidence_end is None
    ):
        return None

    try:
        evidence_start = float(evidence_start)
        evidence_end = float(evidence_end)
    except (TypeError, ValueError):
        return None

    best_candidate = None
    best_overlap = 0.0

    for candidate in source_candidates:

        candidate_start = candidate.get("start")
        candidate_end = candidate.get("end")

        if (
            candidate_start is None
            or candidate_end is None
        ):
            continue

        try:
            candidate_start = float(candidate_start)
            candidate_end = float(candidate_end)
        except (TypeError, ValueError):
            continue

        overlap = max(
            0.0,
            min(
                evidence_end,
                candidate_end
            )
            - max(
                evidence_start,
                candidate_start
            )
        )

        if overlap > best_overlap:
            best_overlap = overlap
            best_candidate = candidate

    if best_candidate is None:
        return None

    return {
        "event_type": (
            best_candidate.get("candidate_type")
            or best_candidate.get("event_type")
            or best_candidate.get("type")
        ),
        "candidate_overlap": best_overlap,
        "candidate": best_candidate,
    }

'''

    PIPELINE_PATH.write_text(
        source.rstrip() + "\n" + event_resolution_code,
        encoding="utf-8"
    )

    print("✓ Evidence event-type resolution integrated.")

print("\nPipeline:", PIPELINE_PATH)
print("✓ Existing code preserved.")
print("✓ No files/cells deleted.")
print("✓ Uses V3 candidate metadata.")
````

````python
# STEP 12O-R32 — FULL ES2004a BACKEND VALIDATION

import importlib.util
import json
from pathlib import Path

print("=" * 90)
print("STEP 12O-R32 — FULL ES2004a BACKEND VALIDATION")
print("=" * 90)


# ============================================================
# 1. LOAD CLEAN PIPELINE
# ============================================================

PIPELINE_PATH = (
    "/content/drive/MyDrive/MTechIndProj/MoM_Project/"
    "07_dashboard/meeting_pipeline.py"
)

spec = importlib.util.spec_from_file_location(
    "meeting_pipeline_r32",
    PIPELINE_PATH
)

pipeline_r32 = importlib.util.module_from_spec(spec)
spec.loader.exec_module(pipeline_r32)

print("\n✓ meeting_pipeline.py loaded")


# ============================================================
# 2. LOAD ES2004a SPEAKER TRANSCRIPT
# ============================================================

TRANSCRIPT_PATH = (
    "/content/drive/MyDrive/MTechIndProj/MoM_Project/"
    "data/transcripts/"
    "ES2004a_pipeline_speaker_transcript.json"
)

with open(
    TRANSCRIPT_PATH,
    "r",
    encoding="utf-8"
) as f:
    transcript_data = json.load(f)

speaker_utterances = transcript_data["utterances"]

print(
    f"✓ Speaker utterances loaded: "
    f"{len(speaker_utterances)}"
)


# ============================================================
# 3. BUILD V3 MOM CANDIDATES
# ============================================================

v3_candidates = pipeline_r32.filter_mom_candidates_v3(
    speaker_utterances
)

print(
    f"✓ V3 MoM candidates: "
    f"{len(v3_candidates)}"
)


# ============================================================
# 4. BUILD BGE + FAISS EVIDENCE RESOURCES
# ============================================================

retrieval_resources = (
    pipeline_r32.build_evidence_retrieval_resources(
        speaker_utterances
    )
)

bge_model_r32 = retrieval_resources[
    "bge_model"
]

faiss_index_r32 = retrieval_resources[
    "faiss_index"
]

evidence_documents_r32 = retrieval_resources[
    "evidence_documents"
]

print(
    f"✓ Evidence documents: "
    f"{len(evidence_documents_r32)}"
)

print(
    f"✓ FAISS index size: "
    f"{faiss_index_r32.ntotal}"
)


# ============================================================
# 5. PREPARE TEST CLAIMS
#
# These are actual structured claims derived from
# validated ES2004a V3 candidates.
#
# We deliberately use source-grounded claims here rather
# than inventing new meeting content.
# ============================================================

test_claims = []

for candidate in v3_candidates[:5]:

    claim_text = str(
        candidate.get("text", "")
    ).strip()

    if not claim_text:
        continue

    test_claims.append({
        "claim": claim_text,
        "speaker": candidate.get("speaker"),
        "event_type": candidate.get("candidate_type"),
        "start": candidate.get("start"),
        "end": candidate.get("end"),
        "source_candidate": candidate,
    })


print(
    f"✓ Claims selected for backend validation: "
    f"{len(test_claims)}"
)


# ============================================================
# 6. VERIFY EACH CLAIM
# ============================================================

verification_results = []

for index, claim in enumerate(
    test_claims,
    start=1
):

    print("\n" + "-" * 90)
    print(
        f"CLAIM {index}/{len(test_claims)}"
    )
    print("-" * 90)

    print(
        "Claim:",
        claim["claim"][:250]
    )

    print(
        "Speaker:",
        claim["speaker"]
    )

    print(
        "Event type:",
        claim["event_type"]
    )

    print(
        "Timestamp:",
        claim["start"],
        "→",
        claim["end"]
    )

    result = (
        pipeline_r32.verify_multi_attribute_claim(
            claim=claim,
            bge_model=bge_model_r32,
            faiss_index=faiss_index_r32,
            evidence_documents=evidence_documents_r32,
            source_candidates=v3_candidates,
            top_k=5,
        )
    )

    verification_results.append(result)

    print(
        "\nOverall:",
        result["overall_status"]
    )

    print(
        "Content:",
        result["attribute_results"]["content"]
    )

    print(
        "Speaker:",
        result["attribute_results"]["speaker"]
    )

    print(
        "Event type:",
        result["attribute_results"]["event_type"]
    )

    print(
        "Temporal:",
        result["attribute_results"]["temporal"]
    )

    print(
        "Failed attributes:",
        result["failed_attributes"]
    )

    selected = result.get(
        "selected_evidence"
    )

    if selected is not None:

        print(
            "\nSelected evidence ID:",
            selected["evidence_id"]
        )

        print(
            "Evidence speaker:",
            selected["speaker"]
        )

        print(
            "Retrieval similarity:",
            round(
                selected["retrieval_similarity"],
                4
            )
        )

        print(
            "NLI:",
            selected["nli_label"]
        )

        print(
            "Entailment:",
            round(
                selected["nli_entailment"],
                4
            )
        )


# ============================================================
# 7. SUMMARY
# ============================================================

verified_count = sum(
    result["overall_status"] == "VERIFIED"
    for result in verification_results
)

flagged_count = sum(
    result["overall_status"] == "FLAGGED"
    for result in verification_results
)

print("\n" + "=" * 90)
print("R32 SUMMARY")
print("=" * 90)

print(
    "Claims tested:",
    len(verification_results)
)

print(
    "VERIFIED:",
    verified_count
)

print(
    "FLAGGED:",
    flagged_count
)

print("\nAttribute summary:")

for attribute in [
    "content",
    "speaker",
    "event_type",
    "temporal",
]:

    statuses = [
        result["attribute_results"][attribute]
        for result in verification_results
    ]

    counts = {}

    for status in statuses:
        counts[status] = (
            counts.get(status, 0) + 1
        )

    print(
        f"  {attribute:12s}: {counts}"
    )


# ============================================================
# 8. STRUCTURE VALIDATION
# ============================================================

required_keys = {
    "claim_text",
    "overall_status",
    "content",
    "speaker",
    "event_type",
    "temporal",
    "attribute_results",
    "failed_attributes",
    "selected_evidence",
}

structure_passed = all(
    required_keys.issubset(
        result.keys()
    )
    for result in verification_results
)

print("\nStructure validation:")

if structure_passed:
    print(
        "✓ All verification records have "
        "the required dashboard-ready fields."
    )
else:
    print(
        "❌ Verification record structure "
        "is incomplete."
    )


# ============================================================
# 9. FINAL R32 STATUS
# ============================================================

print("\n" + "=" * 90)

if (
    len(verification_results) > 0
    and structure_passed
):
    print(
        "✓ R32 PASSED — "
        "full ES2004a backend verification pipeline is operational."
    )
    print(
        "✓ Ready for final output-schema stabilization "
        "and Streamlit dashboard integration."
    )
else:
    print(
        "❌ R32 FAILED — "
        "backend pipeline needs inspection before dashboard integration."
    )

print("=" * 90)
````

````python
# ============================================================
# R32.5 — FINAL DASHBOARD OUTPUT SCHEMA
# ============================================================

def format_verified_claim_result(result):
    """
    Convert the internal verification result into a stable,
    dashboard-ready schema.

    Internal verification logic is unchanged.
    """

    if result is None:
        return {
            "status": "ERROR",
            "claim": None,
            "speaker": None,
            "start": None,
            "end": None,
            "event_type": None,
            "verification": {},
            "evidence": [],
        }

    selected_evidence = result.get("selected_evidence") or {}

    verification = result.get("verification") or {}

    output = {
        "claim": result.get("claim") or result.get("proposition"),

        "speaker": result.get("speaker"),

        "start": result.get("start"),

        "end": result.get("end"),

        "event_type": (
            result.get("event_type")
            or result.get("candidate_type")
            or result.get("type")
        ),

        "status": result.get("status", "UNKNOWN"),

        "verification": {
            "content": verification.get(
                "content",
                result.get("content_status", "NOT_EVALUATED")
            ),

            "speaker": verification.get(
                "speaker",
                result.get("speaker_status", "NOT_EVALUATED")
            ),

            "event_type": verification.get(
                "event_type",
                result.get("event_type_status", "NOT_EVALUATED")
            ),

            "temporal": verification.get(
                "temporal",
                result.get("temporal_status", "NOT_EVALUATED")
            ),
        },

        "evidence": {
            "evidence_id": selected_evidence.get("evidence_id"),
            "speaker": selected_evidence.get("speaker"),
            "start": selected_evidence.get("start"),
            "end": selected_evidence.get("end"),
            "text": selected_evidence.get("text"),
            "retrieval_similarity": selected_evidence.get(
                "retrieval_similarity"
            ),
            "nli_label": selected_evidence.get("nli_label"),
            "nli_entailment": selected_evidence.get(
                "nli_entailment"
            ),
            "nli_contradiction": selected_evidence.get(
                "nli_contradiction"
            ),
            "lexical_overlap": selected_evidence.get(
                "lexical_overlap"
            ),
        },
    }

    return output


print("R32.5 output-schema function loaded successfully.")
````

````python
# ============================================================
# R32.6 — FIX FINAL STATUS MAPPING
# ============================================================

def format_verified_claim_result(result):
    """
    Convert the internal verification result into a stable,
    dashboard-ready schema.
    """

    if result is None:
        return {
            "claim": None,
            "speaker": None,
            "start": None,
            "end": None,
            "event_type": None,
            "status": "ERROR",
            "verification": {},
            "evidence": {},
        }

    # Handle current verification_results schema
    claim_data = result.get("claim", {})

    if isinstance(claim_data, dict):
        claim_text = claim_data.get("claim", "")
        speaker = claim_data.get("speaker")
        start = claim_data.get("start")
        end = claim_data.get("end")
        event_type = claim_data.get("event_type")
    else:
        claim_text = claim_data
        speaker = result.get("speaker")
        start = result.get("start")
        end = result.get("end")
        event_type = result.get("event_type")

    selected_evidence = result.get("selected_evidence") or {}

    status = (
        result.get("overall_status")
        or result.get("status")
        or "UNKNOWN"
    )

    return {
        "claim": claim_text,

        "speaker": speaker,

        "start": start,

        "end": end,

        "event_type": event_type,

        "status": status,

        "verification": {
            "content": result.get(
                "content",
                "NOT_EVALUATED"
            ),
            "speaker": (
                result.get("speaker", {}).get("status")
                if isinstance(result.get("speaker"), dict)
                else "NOT_EVALUATED"
            ),
            "event_type": (
                result.get("event_type", {}).get("status")
                if isinstance(result.get("event_type"), dict)
                else "NOT_EVALUATED"
            ),
            "temporal": (
                result.get("temporal", {}).get("status")
                if isinstance(result.get("temporal"), dict)
                else "NOT_EVALUATED"
            ),
        },

        "evidence": {
            "evidence_id": selected_evidence.get("evidence_id"),
            "speaker": selected_evidence.get("speaker"),
            "start": selected_evidence.get("start"),
            "end": selected_evidence.get("end"),
            "text": selected_evidence.get("text"),
            "retrieval_similarity": selected_evidence.get(
                "retrieval_similarity"
            ),
            "nli_label": selected_evidence.get("nli_label"),
            "nli_entailment": selected_evidence.get(
                "nli_entailment"
            ),
            "nli_contradiction": selected_evidence.get(
                "nli_contradiction"
            ),
            "lexical_overlap": selected_evidence.get(
                "lexical_overlap"
            ),
        },
    }


print("R32.6 status mapping loaded successfully.")
````

````python
# ============================================================
# R33 — INTEGRATE FINAL RESULT FORMATTER
# ============================================================

from pathlib import Path

PROJECT_ROOT = Path("/content/drive/MyDrive/MTechIndProj/MoM_Project")

pipeline_path = PROJECT_ROOT / "07_dashboard" / "meeting_pipeline.py"

assert pipeline_path.exists(), (
    f"Pipeline file not found: {pipeline_path}"
)

pipeline_text = pipeline_path.read_text(encoding="utf-8")

formatter_code = r'''

# ============================================================
# Dashboard Result Formatting
# ============================================================

def format_verified_claim_result(result):
    """
    Convert the internal verification result into a stable
    dashboard-ready schema.
    """

    if result is None:
        return {
            "claim": None,
            "speaker": None,
            "start": None,
            "end": None,
            "event_type": None,
            "status": "ERROR",
            "verification": {},
            "evidence": {},
        }

    claim_data = result.get("claim", {})

    if isinstance(claim_data, dict):
        claim_text = claim_data.get("claim", "")
        speaker = claim_data.get("speaker")
        start = claim_data.get("start")
        end = claim_data.get("end")
        event_type = claim_data.get("event_type")
    else:
        claim_text = claim_data
        speaker = result.get("speaker")
        start = result.get("start")
        end = result.get("end")
        event_type = result.get("event_type")

    selected_evidence = result.get("selected_evidence") or {}

    status = (
        result.get("overall_status")
        or result.get("status")
        or "UNKNOWN"
    )

    content_value = result.get("content")
    content_status = (
        content_value.get("status", "NOT_EVALUATED")
        if isinstance(content_value, dict)
        else content_value or "NOT_EVALUATED"
    )

    speaker_value = result.get("speaker")
    speaker_status = (
        speaker_value.get("status", "NOT_EVALUATED")
        if isinstance(speaker_value, dict)
        else "NOT_EVALUATED"
    )

    event_value = result.get("event_type")
    event_status = (
        event_value.get("status", "NOT_EVALUATED")
        if isinstance(event_value, dict)
        else "NOT_EVALUATED"
    )

    temporal_value = result.get("temporal")
    temporal_status = (
        temporal_value.get("status", "NOT_EVALUATED")
        if isinstance(temporal_value, dict)
        else temporal_value or "NOT_EVALUATED"
    )

    return {
        "claim": claim_text,
        "speaker": speaker,
        "start": start,
        "end": end,
        "event_type": event_type,
        "status": status,
        "verification": {
            "content": content_status,
            "speaker": speaker_status,
            "event_type": event_status,
            "temporal": temporal_status,
        },
        "evidence": {
            "evidence_id": selected_evidence.get("evidence_id"),
            "speaker": selected_evidence.get("speaker"),
            "start": selected_evidence.get("start"),
            "end": selected_evidence.get("end"),
            "text": selected_evidence.get("text"),
            "retrieval_similarity": selected_evidence.get(
                "retrieval_similarity"
            ),
            "nli_label": selected_evidence.get("nli_label"),
            "nli_entailment": selected_evidence.get(
                "nli_entailment"
            ),
            "nli_contradiction": selected_evidence.get(
                "nli_contradiction"
            ),
            "lexical_overlap": selected_evidence.get(
                "lexical_overlap"
            ),
        },
    }
'''

marker = "# Dashboard Result Formatting"

if marker in pipeline_text:
    print("Formatter already exists in meeting_pipeline.py")
else:
    pipeline_path.write_text(
        pipeline_text.rstrip() + "\n" + formatter_code + "\n",
        encoding="utf-8"
    )
    print("R33 formatter integrated successfully.")

print("Pipeline:", pipeline_path)
````

````python
# ============================================================
# DASHBOARD D3 — BUILD CLAIM / VERIFICATION DATA MODEL
# ============================================================

# ------------------------------------------------------------
# Extract claims
# ------------------------------------------------------------

dashboard_claim_list = dashboard_claims.get(
    "claims",
    []
)

dashboard_verification_results = dashboard_verification.get(
    "results",
    []
)


# ------------------------------------------------------------
# Create verification lookup
# ------------------------------------------------------------

verification_by_claim = {}

for result in dashboard_verification_results:

    claim_text = result.get("claim", "")

    if claim_text:
        verification_by_claim[claim_text] = result


# ------------------------------------------------------------
# Combine claim + verification information
# ------------------------------------------------------------

dashboard_rows = []

for claim in dashboard_claim_list:

    claim_text = claim.get(
        "claim_text",
        ""
    )

    verification_result = verification_by_claim.get(
        claim_text,
        {}
    )

    dashboard_rows.append({
        "claim_id": claim.get("claim_id"),
        "topic": claim.get("topic"),
        "claim_text": claim_text,
        "speaker": claim.get("speaker"),
        "start": claim.get("start"),
        "end": claim.get("end"),
        "event_type": claim.get("event_type"),

        "status": verification_result.get(
            "status",
            "UNKNOWN"
        ),

        "reason": verification_result.get(
            "reason",
            ""
        ),

        "checks": verification_result.get(
            "checks",
            {}
        ),

        "evidence_ids": verification_result.get(
            "evidence_ids",
            []
        ),

        "evidence_start": verification_result.get(
            "evidence_start"
        ),

        "evidence_end": verification_result.get(
            "evidence_end"
        ),
    })


# ------------------------------------------------------------
# Validate dashboard model
# ------------------------------------------------------------

print("Dashboard rows:", len(dashboard_rows))

assert len(dashboard_rows) == len(
    dashboard_claim_list
)

assert len(dashboard_rows) == 12

print("\nFirst dashboard row:")
print(json.dumps(
    dashboard_rows[0],
    indent=2,
    ensure_ascii=False
))

print("\nD3 PASSED")
````

**# Verifiable Meeting Minutes — Streamlit Dashboard**

````python
from pathlib import Path

APP_PATH = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project/07_dashboard/app.py"
)

APP_CODE = r'''
# ============================================================
# Verifiable Meeting Minutes — Streamlit Dashboard
# Phase 1: Basic Dashboard
# ============================================================

import json
from pathlib import Path

import streamlit as st


# ============================================================
# PAGE CONFIGURATION
# ============================================================

st.set_page_config(
    page_title="Verifiable Meeting Minutes",
    page_icon="📝",
    layout="wide",
)


# ============================================================
# PROJECT PATHS
# ============================================================

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project"
)

CLAIMS_DIR = PROJECT_ROOT / "data" / "transcripts"
VERIFICATION_DIR = PROJECT_ROOT / "outputs" / "verified"

MEETINGS = {
    "ES2004a": {
        "claims": CLAIMS_DIR / "ES2004a_structured_mom_claims.json",
        "verification": (
            VERIFICATION_DIR
            / "ES2004a_M9_full_multi_attribute_verification.json"
        ),
    }
}


# ============================================================
# LOAD JSON
# ============================================================

@st.cache_data
def load_json(path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)


def load_meeting(meeting_id):

    paths = MEETINGS[meeting_id]

    claims_data = load_json(paths["claims"])
    verification_data = load_json(paths["verification"])

    return claims_data, verification_data


# ============================================================
# HEADER
# ============================================================

st.title("📝 Verifiable Meeting Minutes")

st.markdown(
    """
    **AI-Powered Meeting Intelligence & Evidence Verification**

    Generate structured meeting minutes and verify each claim
    against the original meeting evidence.
    """
)

st.divider()


# ============================================================
# MEETING SELECTION
# ============================================================

st.subheader("Meeting")

selected_meeting = st.selectbox(
    "Select an implemented meeting",
    options=list(MEETINGS.keys()),
)

claims_data, verification_data = load_meeting(
    selected_meeting
)


# ============================================================
# SUMMARY
# ============================================================

claims = claims_data.get("claims", [])

total_claims = verification_data.get(
    "total_claims",
    len(claims),
)

verified_claims = verification_data.get(
    "verified_claims",
    0,
)

flagged_claims = verification_data.get(
    "flagged_claims",
    0,
)

verification_rate = verification_data.get(
    "verification_rate",
    0,
)


# ============================================================
# METRICS
# ============================================================

st.subheader("Meeting Summary")

col1, col2, col3, col4 = st.columns(4)

with col1:
    st.metric("Total Claims", total_claims)

with col2:
    st.metric("Verified", verified_claims)

with col3:
    st.metric("Flagged", flagged_claims)

with col4:
    st.metric(
        "Verification Rate",
        f"{verification_rate:.2f}%"
    )


st.divider()


# ============================================================
# CLAIMS
# ============================================================

st.subheader("Meeting Minutes")

verification_results = verification_data.get(
    "results",
    []
)

verification_map = {}

for result in verification_results:

    claim = result.get("claim", "")

    if claim:
        verification_map[claim] = result


if not claims:

    st.info("No structured meeting claims were found.")

else:

    for index, claim_data in enumerate(
        claims,
        start=1
    ):

        if isinstance(claim_data, dict):

            claim_text = (
                claim_data.get("claim")
                or claim_data.get("text")
                or claim_data.get("content")
                or ""
            )

            speaker = (
                claim_data.get("speaker")
                or "Unknown"
            )

            event_type = (
                claim_data.get("event_type")
                or claim_data.get("event")
                or "Unknown"
            )

        else:

            claim_text = str(claim_data)
            speaker = "Unknown"
            event_type = "Unknown"


        result = verification_map.get(
            claim_text,
            {}
        )

        status = result.get(
            "status",
            "UNKNOWN"
        )

        verification = result.get(
            "verification",
            {}
        )


        if status == "VERIFIED":

            status_icon = "✅"
            status_label = "VERIFIED"

        elif status == "FLAGGED":

            status_icon = "⚠️"
            status_label = "FLAGGED"

        else:

            status_icon = "ℹ️"
            status_label = status


        with st.container(border=True):

            header_col1, header_col2 = st.columns(
                [5, 1]
            )

            with header_col1:

                st.markdown(
                    f"### {index}. {claim_text}"
                )

            with header_col2:

                st.markdown(
                    f"### {status_icon} {status_label}"
                )


            info_col1, info_col2 = st.columns(2)

            with info_col1:

                st.markdown(
                    f"**Speaker:** `{speaker}`"
                )

            with info_col2:

                st.markdown(
                    f"**Event Type:** `{event_type}`"
                )


            if verification:

                st.markdown("**Verification**")

                vcol1, vcol2, vcol3, vcol4 = st.columns(4)

                with vcol1:
                    st.write(
                        "Content: "
                        f"`{verification.get('content', 'N/A')}`"
                    )

                with vcol2:
                    st.write(
                        "Speaker: "
                        f"`{verification.get('speaker', 'N/A')}`"
                    )

                with vcol3:
                    st.write(
                        "Event Type: "
                        f"`{verification.get('event_type', 'N/A')}`"
                    )

                with vcol4:
                    st.write(
                        "Temporal: "
                        f"`{verification.get('temporal', 'N/A')}`"
                    )


# ============================================================
# FOOTER
# ============================================================

st.divider()

st.caption(
    "Verifiable Meeting Minutes • Phase 1 Dashboard"
)
'''

APP_PATH.write_text(APP_CODE, encoding="utf-8")

print("Created:", APP_PATH)
print("Exists:", APP_PATH.exists())
````

````python
# ============================================================
# DASHBOARD D4 — GENERATE STREAMLIT APP
# ============================================================

from pathlib import Path

APP_PATH = (
    PROJECT_ROOT
    / "07_dashboard"
    / "app.py"
)

app_code = r'''
import json
from pathlib import Path
import streamlit as st


# ============================================================
# PROJECT PATH
# ============================================================

# Google Colab / Linux
if Path("/content/drive/MyDrive").exists():
    PROJECT_ROOT = Path(
        "/content/drive/MyDrive/MTechIndProj/MoM_Project"
    )

# Windows Google Drive for Desktop
elif Path(r"G:\My Drive\MTechIndProj\MoM_Project").exists():
    PROJECT_ROOT = Path(
        r"G:\My Drive\MTechIndProj\MoM_Project"
    )

else:
    PROJECT_ROOT = Path.cwd().parent


# ============================================================
# DATA PATHS
# ============================================================

CLAIMS_PATH = (
    PROJECT_ROOT
    / "data"
    / "transcripts"
    / "ES2004a_structured_mom_claims.json"
)

VERIFICATION_PATH = (
    PROJECT_ROOT
    / "outputs"
    / "verified"
    / "ES2004a_M9_full_multi_attribute_verification.json"
)

PDF_PATH = (
    PROJECT_ROOT
    / "outputs"
    / "pdf"
    / "ES2004a_Verifiable_Meeting_Minutes.pdf"
)


# ============================================================
# LOAD DATA
# ============================================================

with open(CLAIMS_PATH, "r", encoding="utf-8") as f:
    claims_data = json.load(f)

with open(VERIFICATION_PATH, "r", encoding="utf-8") as f:
    verification_data = json.load(f)


claims = claims_data.get("claims", [])
verification_results = verification_data.get("results", [])


# ============================================================
# BUILD VERIFICATION LOOKUP
# ============================================================

verification_by_claim = {}

for result in verification_results:

    claim_text = result.get("claim", "")

    if claim_text:
        verification_by_claim[claim_text] = result


# ============================================================
# BUILD DASHBOARD ROWS
# ============================================================

dashboard_rows = []

for claim in claims:

    claim_text = claim.get(
        "claim_text",
        ""
    )

    verification_result = verification_by_claim.get(
        claim_text,
        {}
    )

    dashboard_rows.append({
        "claim_id": claim.get("claim_id"),
        "topic": claim.get("topic"),
        "claim_text": claim_text,
        "speaker": claim.get("speaker"),
        "start": claim.get("start"),
        "end": claim.get("end"),
        "event_type": claim.get("event_type"),

        "status": verification_result.get(
            "status",
            "UNKNOWN"
        ),

        "reason": verification_result.get(
            "reason",
            ""
        ),

        "checks": verification_result.get(
            "checks",
            {}
        ),

        "evidence_ids": verification_result.get(
            "evidence_ids",
            []
        ),

        "evidence_start": verification_result.get(
            "evidence_start"
        ),

        "evidence_end": verification_result.get(
            "evidence_end"
        ),
    })


# ============================================================
# PAGE CONFIG
# ============================================================

st.set_page_config(
    page_title="Verifiable Meeting Minutes",
    page_icon="📝",
    layout="wide"
)


# ============================================================
# HEADER
# ============================================================

st.title("📝 Verifiable Meeting Minutes")

st.caption(
    "AI-generated meeting minutes with evidence retrieval "
    "and multi-level verification"
)


# ============================================================
# MEETING INFORMATION
# ============================================================

st.subheader("Meeting")

col1, col2 = st.columns(2)

with col1:
    st.metric(
        "Meeting ID",
        claims_data.get("meeting_id", "Unknown")
    )

with col2:
    st.metric(
        "Total Claims",
        len(dashboard_rows)
    )


# ============================================================
# SUMMARY METRICS
# ============================================================

verified_count = sum(
    1
    for row in dashboard_rows
    if row["status"] == "VERIFIED"
)

flagged_count = sum(
    1
    for row in dashboard_rows
    if row["status"] != "VERIFIED"
)

verification_rate = (
    verified_count / len(dashboard_rows) * 100
    if dashboard_rows
    else 0
)


st.subheader("Verification Summary")

m1, m2, m3 = st.columns(3)

with m1:
    st.metric(
        "Total Claims",
        len(dashboard_rows)
    )

with m2:
    st.metric(
        "Verified",
        verified_count
    )

with m3:
    st.metric(
        "Verification Rate",
        f"{verification_rate:.2f}%"
    )


# ============================================================
# MEETING MINUTES
# ============================================================

st.subheader("Meeting Minutes")

for row in dashboard_rows:

    if row["status"] == "VERIFIED":
        status_label = "✅ VERIFIED"
    else:
        status_label = "⚠️ FLAGGED"

    with st.container(border=True):

        st.markdown(
            f"### {row['claim_id']}. {row['claim_text']}"
        )

        c1, c2, c3 = st.columns(3)

        with c1:
            st.write(
                f"**Speaker:** {row['speaker']}"
            )

        with c2:
            st.write(
                f"**Event Type:** {row['event_type']}"
            )

        with c3:
            st.write(
                f"**Status:** {status_label}"
            )

        st.caption(
            f"Timestamp: "
            f"{row['start']:.2f}s – "
            f"{row['end']:.2f}s"
        )

        if row["reason"]:
            st.write(
                f"**Verification:** {row['reason']}"
            )

        # ----------------------------------------------------
        # Verification details
        # ----------------------------------------------------

        checks = row["checks"]

        if checks:

            with st.expander("View verification details"):

                content_check = checks.get(
                    "content_consistency",
                    {}
                )

                speaker_check = checks.get(
                    "speaker_consistency",
                    {}
                )

                timestamp_check = checks.get(
                    "timestamp_consistency",
                    {}
                )

                event_check = checks.get(
                    "event_type_consistency",
                    {}
                )

                st.write(
                    "**Content consistency:**",
                    "PASS"
                    if content_check.get("passed")
                    else "FAIL"
                )

                st.write(
                    "**Speaker consistency:**",
                    "PASS"
                    if speaker_check.get("passed")
                    else "FAIL"
                )

                st.write(
                    "**Timestamp consistency:**",
                    "PASS"
                    if timestamp_check.get("passed")
                    else "FAIL"
                )

                st.write(
                    "**Event-type consistency:**",
                    "PASS"
                    if event_check.get("passed")
                    else "FAIL"
                )

                if row["evidence_ids"]:
                    st.write(
                        "**Evidence IDs:**",
                        ", ".join(
                            str(x)
                            for x in row["evidence_ids"]
                        )
                    )


# ============================================================
# PDF
# ============================================================

st.subheader("Generated PDF")

if PDF_PATH.exists():

    with open(PDF_PATH, "rb") as pdf_file:

        st.download_button(
            label="📄 Download Verified Meeting Minutes PDF",
            data=pdf_file,
            file_name="ES2004a_Verifiable_Meeting_Minutes.pdf",
            mime="application/pdf"
        )

else:

    st.info(
        "Verified PDF is not available."
    )
'''


# ------------------------------------------------------------
# WRITE APP
# ------------------------------------------------------------

APP_PATH.write_text(
    app_code,
    encoding="utf-8"
)

print("app.py generated successfully:")
print(APP_PATH)

print("\nD4 PASSED")
````

````python
# ============================================================
# DASHBOARD D6-A — BUILD EVIDENCE LOOKUP
# ============================================================

# ------------------------------------------------------------
# Evidence structure
# ------------------------------------------------------------

print("Evidence object type:", type(dashboard_evidence).__name__)
print("Number of evidence records:", len(dashboard_evidence))

# ------------------------------------------------------------
# Create evidence lookup by evidence_id
# ------------------------------------------------------------

evidence_by_id = {}

for evidence in dashboard_evidence:

    evidence_id = evidence.get("evidence_id")

    if evidence_id is not None:
        evidence_by_id[evidence_id] = evidence

# ------------------------------------------------------------
# Validate lookup
# ------------------------------------------------------------

print(
    "Evidence lookup entries:",
    len(evidence_by_id)
)

assert len(evidence_by_id) == len(
    dashboard_evidence
)

# ------------------------------------------------------------
# Test evidence IDs from first verified claim
# ------------------------------------------------------------

first_row = dashboard_rows[0]

print("\nFirst claim:")
print(first_row["claim_text"])

print(
    "\nEvidence IDs:",
    first_row["evidence_ids"]
)

print("\nRetrieved evidence:")

for evidence_id in first_row["evidence_ids"]:

    evidence = evidence_by_id.get(
        evidence_id
    )

    assert evidence is not None

    print("\nEvidence ID:", evidence_id)
    print("Speaker:", evidence.get("speaker"))
    print(
        "Timestamp:",
        evidence.get("start"),
        "→",
        evidence.get("end")
    )
    print(
        "Similarity:",
        evidence.get("retrieval_similarity")
    )
    print(
        "NLI:",
        evidence.get("nli_label")
    )
    print(
        "Text:",
        evidence.get("text", "")[:150]
    )

print("\nD6-A PASSED")
````

````python
# ============================================================
# DASHBOARD D6-B — BUILD VERIFIED EVIDENCE LOOKUP
# ============================================================

verified_evidence_by_id = {}

for row in dashboard_rows:

    checks = row.get(
        "checks",
        {}
    )

    content_check = checks.get(
        "content_consistency",
        {}
    )

    supporting_candidates = content_check.get(
        "details",
        {}
    ).get(
        "supporting_candidates",
        []
    )

    for candidate in supporting_candidates:

        evidence_id = candidate.get(
            "evidence_id"
        )

        if evidence_id is not None:

            verified_evidence_by_id[
                evidence_id
            ] = candidate


# ------------------------------------------------------------
# Validate
# ------------------------------------------------------------

print(
    "Verified evidence lookup entries:",
    len(verified_evidence_by_id)
)

first_row = dashboard_rows[0]

print("\nFirst claim evidence:")

for evidence_id in first_row["evidence_ids"]:

    evidence = verified_evidence_by_id.get(
        evidence_id
    )

    print("\nEvidence ID:", evidence_id)

    if evidence is None:
        print("  Not available in supporting candidates.")
        continue

    print(
        "  Speaker:",
        evidence.get("speaker")
    )

    print(
        "  Timestamp:",
        evidence.get("start"),
        "→",
        evidence.get("end")
    )

    print(
        "  Retrieval similarity:",
        evidence.get("retrieval_similarity")
    )

    print(
        "  NLI label:",
        evidence.get("nli_label")
    )

    print(
        "  Speaker match:",
        evidence.get("speaker_match")
    )

    print(
        "  Text:",
        evidence.get("text", "")[:150]
    )


print("\nD6-B PASSED")
````

````python
# ============================================================
# DASHBOARD D6-C — ADD EVIDENCE DISPLAY TO STREAMLIT APP
# ============================================================

evidence_ui_block = '''

                # ------------------------------------------------
                # Retrieved Evidence
                # ------------------------------------------------

                st.markdown("#### 📚 Retrieved Evidence")

                evidence_ids = row["evidence_ids"]

                if evidence_ids:

                    for evidence_id in evidence_ids:

                        evidence = verified_evidence_by_id.get(
                            evidence_id
                        )

                        if evidence is None:
                            continue

                        with st.container(border=True):

                            st.markdown(
                                f"**Evidence ID: {evidence_id}**"
                            )

                            st.write(
                                evidence.get(
                                    "text",
                                    "Evidence text unavailable."
                                )
                            )

                            e1, e2, e3 = st.columns(3)

                            with e1:
                                st.write(
                                    f"**Speaker:** "
                                    f"{evidence.get('speaker', 'Unknown')}"
                                )

                            with e2:
                                start = evidence.get("start")
                                end = evidence.get("end")

                                if (
                                    start is not None
                                    and end is not None
                                ):
                                    st.write(
                                        f"**Timestamp:** "
                                        f"{start:.2f}s – {end:.2f}s"
                                    )
                                else:
                                    st.write(
                                        "**Timestamp:** Not available"
                                    )

                            with e3:
                                similarity = evidence.get(
                                    "retrieval_similarity"
                                )

                                if similarity is not None:
                                    st.write(
                                        f"**Similarity:** "
                                        f"{similarity:.4f}"
                                    )
                                else:
                                    st.write(
                                        "**Similarity:** Not available"
                                    )

                            e4, e5 = st.columns(2)

                            with e4:
                                nli_label = evidence.get(
                                    "nli_label",
                                    "unknown"
                                )

                                st.write(
                                    f"**NLI:** {nli_label}"
                                )

                            with e5:
                                speaker_match = evidence.get(
                                    "speaker_match"
                                )

                                st.write(
                                    f"**Speaker Match:** "
                                    f"{speaker_match}"
                                )

                else:

                    st.info(
                        "No supporting evidence was recorded."
                    )
'''

# ------------------------------------------------------------
# Insert the evidence block immediately before the PDF section
# ------------------------------------------------------------

pdf_marker = '''
# ============================================================
# PDF
# ============================================================
'''

assert pdf_marker in app_code

app_code = app_code.replace(
    pdf_marker,
    evidence_ui_block + "\n\n" + pdf_marker
)

# ------------------------------------------------------------
# Write updated app.py
# ------------------------------------------------------------

APP_PATH.write_text(
    app_code,
    encoding="utf-8"
)

print("D6-C applied successfully.")
print("Updated:", APP_PATH)

print("\nD6-C PASSED")
````

````python
# ============================================================
# DASHBOARD D9-F — OPTIMIZED END-TO-END MEETING PROCESSOR
# ============================================================

from pathlib import Path


def process_uploaded_meeting(
    audio_path,
    generator="qwen",
    generator_model=None,
    generator_tokenizer=None,
    hf_token=None,
    top_k=10,
    temporal_tolerance_seconds=5.0,
):
    """
    End-to-end processing for one uploaded meeting.

    Pipeline:
        Audio
        -> VAD
        -> WhisperX
        -> Pyannote
        -> Speaker-word alignment
        -> MoM candidate filtering
        -> Structured MoM generation
        -> BGE + FAISS retrieval resources
        -> Multi-attribute verification
        -> Dashboard-ready results
    """

    audio_path = Path(audio_path)

    if not audio_path.exists():
        raise FileNotFoundError(
            f"Meeting file not found: {audio_path}"
        )

    # ========================================================
    # 1. VAD
    # ========================================================

    print("1/8 Running Silero VAD...")

    vad_result = mp.run_silero_vad(
        audio_path
    )

    # ========================================================
    # 2. WHISPERX
    # ========================================================

    print("2/8 Running WhisperX...")

    whisper_result = mp.run_whisperx_pipeline(
        audio_path
    )

    # ========================================================
    # 3. PYANNOTE
    # ========================================================

    print("3/8 Running Pyannote diarization...")

    diarization_result = (
        mp.run_pyannote_diarization(
            audio_path,
            hf_token=hf_token
        )
    )

    # ========================================================
    # 4. SPEAKER-WORD ALIGNMENT
    # ========================================================

    print("4/8 Aligning speakers with words...")

    alignment_result = (
        mp.run_speaker_word_alignment(
            whisper_result,
            diarization_result
        )
    )

    utterances = alignment_result.get(
        "utterances",
        []
    )

    # ========================================================
    # 5. MoM CANDIDATE FILTERING
    # ========================================================

    print("5/8 Detecting MoM candidates...")

    candidate_result = (
        mp.filter_mom_candidates_v3(
            utterances,
            return_all=True
        )
    )

    candidates = candidate_result.get(
        "candidates",
        []
    )

    # ========================================================
    # 6. STRUCTURED MoM GENERATION
    # ========================================================

    print("6/8 Generating structured MoM...")

    generation_result = (
        mp.generate_structured_mom(
            candidates,
            generator=generator,
            model=generator_model,
            tokenizer=generator_tokenizer,
        )
    )

    generated_claims = []

    if isinstance(
        generation_result,
        dict
    ):
        generated_claims = (
            generation_result.get(
                "claims",
                []
            )
        )

    # ========================================================
    # 7. BUILD RETRIEVAL RESOURCES ONCE
    # ========================================================

    print("7/8 Building BGE + FAISS resources...")

    retrieval_resources = (
        mp.build_evidence_retrieval_resources(
            utterances
        )
    )

    bge_model = retrieval_resources[
        "bge_model"
    ]

    faiss_index = retrieval_resources[
        "faiss_index"
    ]

    evidence_documents = retrieval_resources[
        "evidence_documents"
    ]

    # ========================================================
    # 8. VERIFY EVERY GENERATED CLAIM
    # ========================================================

    print("8/8 Verifying generated claims...")

    verification_results = []

    for claim in generated_claims:

        verification_result = (
            mp.verify_multi_attribute_claim(
                claim=claim,
                bge_model=bge_model,
                faiss_index=faiss_index,
                evidence_documents=(
                    evidence_documents
                ),
                source_candidates=candidates,
                top_k=top_k,
                temporal_tolerance_seconds=(
                    temporal_tolerance_seconds
                ),
            )
        )

        formatted_result = (
            mp.format_verified_claim_result(
                verification_result
            )
        )

        verification_results.append(
            formatted_result
        )

    # ========================================================
    # SUMMARY
    # ========================================================

    verified_count = sum(
        1
        for result in verification_results
        if result.get("status") == "VERIFIED"
    )

    flagged_count = sum(
        1
        for result in verification_results
        if result.get("status") == "FLAGGED"
    )

    total_claims = len(
        verification_results
    )

    verification_rate = (
        verified_count / total_claims * 100
        if total_claims
        else 0.0
    )

    # ========================================================
    # FINAL RESULT
    # ========================================================

    return {
        "status": "completed",

        "audio_path": str(
            audio_path
        ),

        "vad": vad_result,

        "whisper": whisper_result,

        "diarization": diarization_result,

        "alignment": alignment_result,

        "candidate_filtering": {
            "candidate_count": len(
                candidates
            ),
            "candidates": candidates,
            "all_items": candidate_result.get(
                "all_items",
                []
            ),
        },

        "generation": generation_result,

        "retrieval": {
            "model": mp.BGE_MODEL_NAME,
            "top_k": int(top_k),
            "evidence_count": len(
                evidence_documents
            ),
        },

        "verification": {
            "results": verification_results,
            "total_claims": total_claims,
            "verified_claims": verified_count,
            "flagged_claims": flagged_count,
            "verification_rate": (
                verification_rate
            ),
        },
    }


print(
    "D9-F COMPLETE"
)

print(
    "Optimized process_uploaded_meeting() "
    "defined successfully."
)
````

````python
# ============================================================
# DASHBOARD D9-J — UPLOAD FILE STORAGE LOGIC
# ============================================================

RAW_UPLOAD_DIR = PROJECT_ROOT / "data" / "raw"

RAW_UPLOAD_DIR.mkdir(
    parents=True,
    exist_ok=True
)


def save_uploaded_meeting(
    uploaded_file
):
    """
    Save a Streamlit UploadedFile into
    the project's data/raw directory.

    Returns
    -------
    Path
        Path to the saved meeting file.
    """

    if uploaded_file is None:
        raise ValueError(
            "No meeting file was supplied."
        )

    filename = Path(
        uploaded_file.name
    ).name

    output_path = (
        RAW_UPLOAD_DIR / filename
    )

    with open(
        output_path,
        "wb"
    ) as f:

        f.write(
            uploaded_file.getbuffer()
        )

    return output_path


print("=" * 80)
print("D9-J COMPLETE")
print("=" * 80)

print(
    "Upload directory:",
    RAW_UPLOAD_DIR
)

print(
    "save_uploaded_meeting() defined successfully."
)
````

````python
# ============================================================
# DASHBOARD D9-L — SAVED AUDIO COMPATIBILITY CHECK
# ============================================================

SUPPORTED_AUDIO_EXTENSIONS = {
    ".wav",
    ".mp3",
    ".m4a",
    ".mp4",
    ".mov",
    ".avi",
    ".mkv",
}

def validate_uploaded_meeting_path(audio_path):
    """
    Validate that a saved meeting file is suitable
    for the existing meeting processing pipeline.
    """
    audio_path = Path(audio_path)

    if not audio_path.exists():
        raise FileNotFoundError(
            f"Meeting file does not exist: {audio_path}"
        )

    if not audio_path.is_file():
        raise ValueError(
            f"Meeting path is not a file: {audio_path}"
        )

    extension = audio_path.suffix.lower()

    if extension not in SUPPORTED_AUDIO_EXTENSIONS:
        raise ValueError(
            f"Unsupported meeting format: {extension}"
        )

    return {
        "valid": True,
        "path": str(audio_path),
        "filename": audio_path.name,
        "extension": extension,
        "size_bytes": audio_path.stat().st_size,
    }


d9l_result = validate_uploaded_meeting_path(saved_test_path)

print("=" * 80)
print("D9-L COMPLETE")
print("=" * 80)
print("Validation result:")
print(d9l_result)

assert d9l_result["valid"] is True
assert d9l_result["extension"] == ".wav"

print("D9-L validation PASSED")
````

````python
# ============================================================
# DASHBOARD D9-M — UPLOAD HANDOFF VALIDATION
# ============================================================

def prepare_uploaded_meeting(uploaded_file):
    """
    Complete pre-processing handoff:

    Streamlit UploadedFile
        ↓
    Save to data/raw
        ↓
    Validate saved path
        ↓
    Return processing-ready metadata
    """
    if uploaded_file is None:
        raise ValueError("No meeting file was supplied.")

    saved_path = save_uploaded_meeting(uploaded_file)

    validation = validate_uploaded_meeting_path(saved_path)

    return {
        "saved_path": saved_path,
        "validation": validation,
    }


# Test using the same mock upload object
d9m_result = prepare_uploaded_meeting(test_file)

print("=" * 80)
print("D9-M COMPLETE")
print("=" * 80)
print("Saved path:", d9m_result["saved_path"])
print("Validation:", d9m_result["validation"])

assert d9m_result["saved_path"].exists()
assert d9m_result["validation"]["valid"] is True

print("D9-M validation PASSED")
````

````python
# ============================================================
# DASHBOARD D9-N — BACKEND HANDOFF VALIDATION
# ============================================================

import inspect

# Confirm the orchestration function is available
assert callable(process_uploaded_meeting), (
    "process_uploaded_meeting() is not available."
)

# Inspect its signature without executing it
d9n_signature = inspect.signature(process_uploaded_meeting)

print("=" * 80)
print("D9-N COMPLETE")
print("=" * 80)
print("process_uploaded_meeting() signature:")
print(d9n_signature)

# Confirm the first required parameter is audio_path
parameter_names = list(d9n_signature.parameters.keys())

print("\nParameters:", parameter_names)

assert "audio_path" in parameter_names

# Confirm our saved path is a Path object
assert isinstance(d9m_result["saved_path"], Path)

print("\nProcessing-ready path:")
print(d9m_result["saved_path"])

print("\nD9-N validation PASSED")
print("Backend handoff interface is compatible.")
````

````python
# ============================================================
# DASHBOARD D9-O — REAL MEETING FILE DISCOVERY
# ============================================================

real_meeting_files = sorted(
    [
        p for p in RAW_UPLOAD_DIR.iterdir()
        if p.is_file()
        and p.suffix.lower() in SUPPORTED_AUDIO_EXTENSIONS
        and p.name != "D9K_test_meeting.wav"
    ]
)

print("=" * 80)
print("D9-O COMPLETE")
print("=" * 80)

print("Upload directory:")
print(RAW_UPLOAD_DIR)

print("\nReal meeting files found:", len(real_meeting_files))

for i, path in enumerate(real_meeting_files, start=1):
    size_mb = path.stat().st_size / (1024 * 1024)
    print(f"{i}. {path.name} | {size_mb:.2f} MB")

if real_meeting_files:
    d9o_real_meeting_path = real_meeting_files[0]
    print("\nSelected meeting file:")
    print(d9o_real_meeting_path)
else:
    d9o_real_meeting_path = None
    print("\nNo real meeting recording is currently present.")

print("=" * 80)
````

````python
# ============================================================
# DASHBOARD D9-P — STREAMLIT UPLOAD SAVE INTEGRATION
# ============================================================

app_path = DASHBOARD_DIR / "app.py"

assert app_path.exists(), f"app.py not found: {app_path}"

app_text = app_path.read_text(encoding="utf-8")

# Add upload-save helper if it is not already present.
upload_helper_block = '''
RAW_UPLOAD_DIR = PROJECT_ROOT / "data" / "raw"
RAW_UPLOAD_DIR.mkdir(parents=True, exist_ok=True)

SUPPORTED_AUDIO_EXTENSIONS = {
    ".wav",
    ".mp3",
    ".m4a",
    ".mp4",
    ".mov",
    ".avi",
    ".mkv",
}

def save_uploaded_meeting(uploaded_file):
    """Save an uploaded meeting recording to data/raw."""
    if uploaded_file is None:
        raise ValueError("No meeting file was supplied.")

    filename = Path(uploaded_file.name).name
    output_path = RAW_UPLOAD_DIR / filename

    with open(output_path, "wb") as f:
        f.write(uploaded_file.getbuffer())

    return output_path
'''

if "def save_uploaded_meeting(uploaded_file):" not in app_text:
    marker = "st.set_page_config("
    marker_pos = app_text.find(marker)

    if marker_pos == -1:
        raise RuntimeError(
            "Could not find Streamlit page configuration in app.py."
        )

    app_text = (
        app_text[:marker_pos]
        + upload_helper_block
        + "\n\n"
        + app_text[marker_pos:]
    )

# Replace the current placeholder processing block.
old_processing = '''    if process_meeting:
        st.info("Meeting processing pipeline will be connected in the next stage.")
'''

new_processing = '''    if process_meeting:
        try:
            saved_path = save_uploaded_meeting(uploaded_meeting)

            st.success(
                f"Meeting uploaded successfully: {saved_path.name}"
            )

            st.write(
                f"Saved to: `{saved_path}`"
            )

            st.write(
                f"File size: "
                f"{saved_path.stat().st_size / (1024 * 1024):.2f} MB"
            )

            st.info(
                "Upload saved successfully. "
                "The AI processing pipeline will be connected next."
            )

        except Exception as e:
            st.error(f"Upload failed: {e}")
'''

if old_processing in app_text:
    app_text = app_text.replace(
        old_processing,
        new_processing,
        1
    )
else:
    print(
        "WARNING: Existing placeholder processing block was not found."
    )

app_path.write_text(app_text, encoding="utf-8")

print("=" * 80)
print("D9-P COMPLETE")
print("=" * 80)
print("Updated:", app_path)
print("Upload saving integrated into app.py.")
print()
print("Next:")
print("1. Restart Streamlit.")
print("2. Upload a REAL meeting recording.")
print("3. Click Process Meeting.")
print("4. Confirm the success message shows the saved path.")
````

````python
# ============================================================
# DASHBOARD D9-S — CONNECT PROCESS BUTTON TO FILE SAVING
# ============================================================

app_path = DASHBOARD_DIR / "app.py"

app_text = app_path.read_text(encoding="utf-8")

old_block = '''    if process_meeting:

        st.info(
            "Meeting processing pipeline will be connected "
            "in the next stage."
        )
'''

new_block = '''    if process_meeting:

        try:

            saved_path = save_uploaded_meeting(
                uploaded_meeting
            )

            st.success(
                f"Meeting uploaded successfully: "
                f"{saved_path.name}"
            )

            st.write(
                f"Saved to: `{saved_path}`"
            )

            st.write(
                f"File size: "
                f"{saved_path.stat().st_size / (1024 * 1024):.2f} MB"
            )

            st.info(
                "Upload saved successfully. "
                "AI processing will be connected next."
            )

        except Exception as e:

            st.error(
                f"Upload failed: {e}"
            )
'''

if old_block not in app_text:
    raise RuntimeError(
        "The expected placeholder upload block was not found. "
        "No changes were made."
    )

app_text = app_text.replace(
    old_block,
    new_block,
    1
)

app_path.write_text(
    app_text,
    encoding="utf-8"
)

print("=" * 80)
print("D9-S COMPLETE")
print("=" * 80)
print("Updated:", app_path)
print()
print("The Process Meeting button now calls:")
print("    save_uploaded_meeting(uploaded_meeting)")
print()
print("No AI processing has been connected yet.")
````
