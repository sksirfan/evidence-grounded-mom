```python
import os

# ============================================================
# CREATE A COMPLETELY NEW DASHBOARD APP
# ============================================================

PROJECT_ROOT = "/content/drive/MyDrive/MTechIndProj/MoM_Project"

DASHBOARD_DIR = os.path.join(
    PROJECT_ROOT,
    "07_dashboard"
)

APP_PATH = os.path.join(
    DASHBOARD_DIR,
    "app.py"
)

os.makedirs(
    DASHBOARD_DIR,
    exist_ok=True
)


APP_CODE = r'''
# ============================================================
# VERIFIABLE MEETING MINUTES
# Multi-Attribute Evidence Consistency
#
# Clean Streamlit Dashboard
# ============================================================

import os
import json
import html

import streamlit as st

from reportlab.lib.pagesizes import A4
from reportlab.platypus import (
    SimpleDocTemplate,
    Paragraph,
    Spacer,
    Table,
    TableStyle
)
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet


# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

PROJECT_ROOT = (
    "/content/drive/MyDrive/MTechIndProj/MoM_Project"
)

MEETING_ID = "ES2004a"


# ============================================================
# 2. DATA PATHS
# ============================================================

CLAIMS_PATH = os.path.join(
    PROJECT_ROOT,
    "data",
    "transcripts",
    f"{MEETING_ID}_structured_mom_claims.json"
)

TRANSCRIPT_PATH = os.path.join(
    PROJECT_ROOT,
    "data",
    "transcripts",
    f"{MEETING_ID}_speaker_transcript.json"
)

EVIDENCE_PATH = os.path.join(
    PROJECT_ROOT,
    "data",
    "evidence",
    f"{MEETING_ID}_evidence_documents.json"
)

VERIFICATION_PATH = os.path.join(
    PROJECT_ROOT,
    "outputs",
    "verified",
    f"{MEETING_ID}_M9_full_multi_attribute_verification.json"
)

PDF_DIR = os.path.join(
    PROJECT_ROOT,
    "outputs",
    "pdf"
)

PDF_PATH = os.path.join(
    PDF_DIR,
    f"{MEETING_ID}_Verifiable_Meeting_Minutes.pdf"
)


# ============================================================
# 3. HELPER FUNCTIONS
# ============================================================

def load_json(path):
    """Load a JSON file."""

    with open(
        path,
        "r",
        encoding="utf-8"
    ) as file:

        return json.load(file)


def format_timestamp(seconds):
    """
    Convert seconds to MM:SS.

    Example:
        84.66 -> 01:24
        105.24 -> 01:45
    """

    if seconds is None:
        return "N/A"

    seconds = float(seconds)

    minutes = int(seconds // 60)
    secs = int(seconds % 60)

    return f"{minutes:02d}:{secs:02d}"


def safe_html(text):
    """Escape text before inserting into HTML/PDF."""

    if text is None:
        return ""

    return html.escape(
        str(text)
    )


# ============================================================
# 4. LOAD ES2004a DATA
# ============================================================

claims_data = load_json(
    CLAIMS_PATH
)

transcript_data = load_json(
    TRANSCRIPT_PATH
)

evidence_data = load_json(
    EVIDENCE_PATH
)

verification_data = load_json(
    VERIFICATION_PATH
)

results = verification_data[
    "results"
]


# ============================================================
# 5. STREAMLIT CONFIGURATION
# ============================================================

st.set_page_config(
    page_title="Verifiable Meeting Minutes",
    page_icon="📝",
    layout="wide"
)


# ============================================================
# 6. CUSTOM UI STYLE
# ============================================================

st.markdown(
    """
    <style>

    /* -------------------------------------------------------
       Page
       ------------------------------------------------------- */

    .stApp {
        background:
            linear-gradient(
                135deg,
                #f8faff 0%,
                #f4f3ff 100%
            );
    }


    /* -------------------------------------------------------
       Main header
       ------------------------------------------------------- */

    .hero {
        background:
            linear-gradient(
                135deg,
                #312e81,
                #4f46e5,
                #7c3aed
            );

        padding: 30px 34px;

        border-radius: 22px;

        color: white;

        margin-bottom: 25px;

        box-shadow:
            0 12px 30px
            rgba(79, 70, 229, 0.22);
    }

    .hero-title {
        font-size: 36px;
        font-weight: 800;
        margin-bottom: 8px;
    }

    .hero-subtitle {
        font-size: 16px;
        opacity: 0.92;
    }


    /* -------------------------------------------------------
       Mode cards
       ------------------------------------------------------- */

    .mode-card {
        background: white;

        border-radius: 20px;

        padding: 24px;

        min-height: 245px;

        border: 1px solid #e5e7eb;

        box-shadow:
            0 8px 24px
            rgba(15, 23, 42, 0.07);
    }

    .mode-title {
        font-size: 22px;
        font-weight: 750;
        color: #1e293b;
    }

    .mode-description {
        color: #64748b;
        font-size: 14px;
        margin-top: 5px;
        margin-bottom: 18px;
    }


    /* -------------------------------------------------------
       Metrics
       ------------------------------------------------------- */

    .metric-card {
        background: white;

        border-radius: 16px;

        padding: 16px;

        text-align: center;

        border: 1px solid #e5e7eb;

        box-shadow:
            0 5px 16px
            rgba(15, 23, 42, 0.06);
    }

    .metric-value {
        font-size: 27px;
        font-weight: 800;
    }

    .metric-label {
        color: #64748b;
        font-size: 12px;
        margin-top: 3px;
    }


    /* -------------------------------------------------------
       Upload information
       ------------------------------------------------------- */

    .upload-box {
        background: #eef2ff;

        border: 2px dashed #6366f1;

        border-radius: 16px;

        padding: 18px;

        color: #3730a3;

        margin-bottom: 15px;
    }


    /* -------------------------------------------------------
       Section titles
       ------------------------------------------------------- */

    .section-title {
        font-size: 25px;

        font-weight: 800;

        color: #1e293b;

        margin-top: 10px;

        margin-bottom: 15px;
    }


    /* -------------------------------------------------------
       Evidence card
       ------------------------------------------------------- */

    .evidence-card {
        background: #f8fafc;

        border-left: 4px solid #6366f1;

        border-radius: 10px;

        padding: 13px 16px;

        margin: 8px 0;
    }


    /* -------------------------------------------------------
       Status
       ------------------------------------------------------- */

    .status-verified {
        color: #166534;

        background: #dcfce7;

        padding: 6px 13px;

        border-radius: 20px;

        font-weight: 700;
    }

    .status-flagged {
        color: #991b1b;

        background: #fee2e2;

        padding: 6px 13px;

        border-radius: 20px;

        font-weight: 700;
    }

    </style>
    """,
    unsafe_allow_html=True
)


# ============================================================
# 7. HERO HEADER
# ============================================================

st.markdown(
    """
    <div class="hero">

        <div class="hero-title">
            📝 Verifiable Meeting Minutes
        </div>

        <div class="hero-subtitle">
            Multi-Attribute Evidence Consistency
            &nbsp; • &nbsp;
            Evidence-Grounded Meeting Intelligence
        </div>

    </div>
    """,
    unsafe_allow_html=True
)


# ============================================================
# 8. TWO OPERATING MODES
# ============================================================

implemented_col, upload_col = st.columns(
    2,
    gap="large"
)


# ============================================================
# LEFT MODE
# IMPLEMENTED MEETINGS
# ============================================================

with implemented_col:

    st.markdown(
        """
        <div class="mode-card">

            <div class="mode-title">
                📚 Implemented Meetings
            </div>

            <div class="mode-description">
                Explore meetings that have already been
                processed through the evidence verification
                pipeline.
            </div>

        </div>
        """,
        unsafe_allow_html=True
    )


    implemented_meetings = [
        "ES2004a"
    ]


    selected_meeting = st.selectbox(
        "Select Meeting",
        implemented_meetings,
        key="implemented_meeting"
    )


    st.markdown(
        f"**Selected:** `{selected_meeting}`"
    )


    # --------------------------------------------------------
    # Summary metrics
    # --------------------------------------------------------

    total_claims = verification_data[
        "total_claims"
    ]

    verified_claims = verification_data[
        "verified_claims"
    ]

    flagged_claims = verification_data[
        "flagged_claims"
    ]

    verification_rate = verification_data[
        "verification_rate"
    ]


    metric1, metric2 = st.columns(2)

    with metric1:

        st.markdown(
            f"""
            <div class="metric-card">

                <div class="metric-value">
                    {total_claims}
                </div>

                <div class="metric-label">
                    Total MoM Claims
                </div>

            </div>
            """,
            unsafe_allow_html=True
        )


    with metric2:

        st.markdown(
            f"""
            <div class="metric-card">

                <div
                    class="metric-value"
                    style="color:#16a34a;"
                >
                    {verified_claims}
                </div>

                <div class="metric-label">
                    Verified
                </div>

            </div>
            """,
            unsafe_allow_html=True
        )


    metric3, metric4 = st.columns(2)

    with metric3:

        st.markdown(
            f"""
            <div class="metric-card">

                <div
                    class="metric-value"
                    style="color:#dc2626;"
                >
                    {flagged_claims}
                </div>

                <div class="metric-label">
                    Flagged
                </div>

            </div>
            """,
            unsafe_allow_html=True
        )


    with metric4:

        st.markdown(
            f"""
            <div class="metric-card">

                <div
                    class="metric-value"
                    style="color:#7c3aed;"
                >
                    {verification_rate:.2f}%
                </div>

                <div class="metric-label">
                    Verification Rate
                </div>

            </div>
            """,
            unsafe_allow_html=True
        )


# ============================================================
# RIGHT MODE
# UPLOAD NEW MEETING
# ============================================================

with upload_col:

    st.markdown(
        """
        <div class="mode-card">

            <div class="mode-title">
                📤 Upload New Meeting
            </div>

            <div class="mode-description">
                Upload a meeting recording and run the
                evidence-grounded MoM pipeline.
            </div>

            <div class="upload-box">

                🎙️ <b>Meeting Recording</b>

                <br><br>

                Supported formats:
                WAV • MP3 • M4A • MP4

                <br><br>

                Processing will include:
                VAD • ASR • Diarization •
                MoM Generation • Evidence Retrieval •
                Verification

            </div>

        </div>
        """,
        unsafe_allow_html=True
    )


    uploaded_file = st.file_uploader(
        "Choose meeting recording",
        type=[
            "wav",
            "mp3",
            "m4a",
            "mp4"
        ],
        key="meeting_upload"
    )


    if uploaded_file is not None:

        st.success(
            f"Uploaded: {uploaded_file.name}"
        )


        process_clicked = st.button(
            "🚀 Process Meeting",
            use_container_width=True,
            key="process_meeting"
        )


        if process_clicked:

            st.info(
                "Upload processing pipeline will be "
                "connected in the next implementation stage."
            )


# ============================================================
# 9. RESULTS SECTION
# ============================================================

st.divider()

st.markdown(
    """
    <div class="section-title">
        📊 Meeting Results
    </div>
    """,
    unsafe_allow_html=True
)


# ============================================================
# 10. CLAIM DISPLAY
# ============================================================

for idx, result in enumerate(
    results,
    start=1
):

    status = result.get(
        "status",
        "UNKNOWN"
    )

    claim = result.get(
        "claim",
        ""
    )

    claim_speaker = result.get(
        "claim_speaker",
        "Unknown"
    )

    claim_type = result.get(
        "claim_type",
        "Unknown"
    )

    reason = result.get(
        "reason",
        ""
    )


    with st.expander(
        f"Claim {idx} — {status}",
        expanded=(idx == 1)
    ):

        st.markdown(
            f"### {claim}"
        )


        # ----------------------------------------------------
        # Metadata
        # ----------------------------------------------------

        meta1, meta2, meta3 = st.columns(3)


        with meta1:

            st.markdown(
                f"**👤 Speaker:** {claim_speaker}"
            )


        with meta2:

            st.markdown(
                f"**🏷 Event Type:** {claim_type}"
            )


        with meta3:

            if result.get(
                "evidence_start"
            ) is not None:

                start = result[
                    "evidence_start"
                ]

                end = result.get(
                    "evidence_end",
                    start
                )

                st.markdown(
                    f"**⏱ Evidence Time:** "
                    f"{format_timestamp(start)} – "
                    f"{format_timestamp(end)}"
                )

            else:

                st.markdown(
                    "**⏱ Evidence Time:** N/A"
                )


        st.divider()


        # ----------------------------------------------------
        # Verification status
        # ----------------------------------------------------

        if status == "VERIFIED":

            st.success(
                "✓ VERIFIED"
            )

        else:

            st.error(
                "✗ FLAGGED"
            )


        st.markdown(
            f"**Reason:** {reason}"
        )


        # ====================================================
        # CONSISTENCY CHECKS
        # ====================================================

        st.markdown(
            "#### 🔍 Multi-Attribute Consistency"
        )


        checks = result.get(
            "checks",
            {}
        )


        check_names = [
            (
                "content_consistency",
                "Content"
            ),
            (
                "speaker_consistency",
                "Speaker"
            ),
            (
                "timestamp_consistency",
                "Timestamp"
            ),
            (
                "event_type_consistency",
                "Event Type"
            )
        ]


        check_columns = st.columns(4)


        for col, (
            check_key,
            display_name
        ) in zip(
            check_columns,
            check_names
        ):

            check = checks.get(
                check_key,
                {}
            )


            with col:

                if check_key == (
                    "timestamp_consistency"
                ):

                    st.info(
                        f"ℹ **{display_name}**\n\n"
                        "Not independently evaluated"
                    )


                elif not check.get(
                    "applicable",
                    True
                ):

                    st.info(
                        f"— **{display_name}**\n\n"
                        "Not applicable"
                    )


                elif check.get(
                    "passed",
                    False
                ):

                    st.success(
                        f"✓ **{display_name}**\n\n"
                        "Passed"
                    )


                else:

                    st.error(
                        f"✗ **{display_name}**\n\n"
                        "Failed"
                    )


        # ====================================================
        # SUPPORTING EVIDENCE
        # ====================================================

        st.markdown(
            "#### 🔎 Supporting Evidence"
        )


        content_check = checks.get(
            "content_consistency",
            {}
        )


        details = content_check.get(
            "details",
            {}
        )


        supporting_candidates = details.get(
            "supporting_candidates",
            []
        )


        if not supporting_candidates:

            st.info(
                "No supporting evidence available."
            )


        else:

            for evidence in supporting_candidates:

                evidence_id = evidence.get(
                    "evidence_id",
                    "N/A"
                )

                speaker = evidence.get(
                    "speaker",
                    "Unknown"
                )

                start = evidence.get(
                    "start",
                    0
                )

                end = evidence.get(
                    "end",
                    0
                )

                text = evidence.get(
                    "text",
                    ""
                )


                st.markdown(
                    f"""
                    <div class="evidence-card">

                        <b>
                            Evidence {evidence_id}
                        </b>

                        &nbsp; • &nbsp;

                        {speaker}

                        &nbsp; • &nbsp;

                        ⏱ {format_timestamp(start)}
                        –
                        {format_timestamp(end)}

                        <br><br>

                        {safe_html(text)}

                    </div>
                    """,
                    unsafe_allow_html=True
                )


                st.caption(
                    f"NLI: "
                    f"{evidence.get('nli_label', 'N/A')}"
                    f"  |  "
                    f"Retrieval similarity: "
                    f"{evidence.get('retrieval_similarity', 0):.3f}"
                )


# ============================================================
# 11. PDF GENERATION
# ============================================================

def create_mom_pdf(
    results,
    verification_data
):
    """
    Create the clean project-deliverable PDF.

    The PDF intentionally excludes the detailed
    four-attribute consistency table because those
    diagnostics are already visible in the dashboard.
    """

    os.makedirs(
        PDF_DIR,
        exist_ok=True
    )


    styles = getSampleStyleSheet()

    title_style = styles[
        "Title"
    ]

    heading_style = styles[
        "Heading2"
    ]

    body_style = styles[
        "BodyText"
    ]


    document = SimpleDocTemplate(
        PDF_PATH,
        pagesize=A4,
        rightMargin=40,
        leftMargin=40,
        topMargin=40,
        bottomMargin=40
    )


    story = []


    # --------------------------------------------------------
    # PDF header
    # --------------------------------------------------------

    story.append(
        Paragraph(
            "Verifiable Meeting Minutes",
            title_style
        )
    )


    story.append(
        Paragraph(
            "Multi-Attribute Evidence Consistency",
            heading_style
        )
    )


    story.append(
        Paragraph(
            f"<b>Meeting:</b> {MEETING_ID}",
            body_style
        )
    )


    story.append(
        Spacer(
            1,
            15
        )
    )


    # --------------------------------------------------------
    # Summary table
    # --------------------------------------------------------

    summary_data = [
        [
            "Total Claims",
            "Verified",
            "Flagged",
            "Verification Rate"
        ],
        [
            str(
                verification_data[
                    "total_claims"
                ]
            ),
            str(
                verification_data[
                    "verified_claims"
                ]
            ),
            str(
                verification_data[
                    "flagged_claims"
                ]
            ),
            (
                f"{verification_data['verification_rate']:.2f}%"
            )
        ]
    ]


    summary_table = Table(
        summary_data
    )


    summary_table.setStyle(
        TableStyle([
            (
                "BACKGROUND",
                (0, 0),
                (-1, 0),
                colors.lightgrey
            ),
            (
                "GRID",
                (0, 0),
                (-1, -1),
                0.5,
                colors.grey
            ),
            (
                "ALIGN",
                (0, 0),
                (-1, -1),
                "CENTER"
            ),
            (
                "FONTNAME",
                (0, 0),
                (-1, 0),
                "Helvetica-Bold"
            ),
            (
                "TOPPADDING",
                (0, 0),
                (-1, 0),
                8
            ),
            (
                "BOTTOMPADDING",
                (0, 0),
                (-1, 0),
                8
            )
        ])
    )


    story.append(
        summary_table
    )


    story.append(
        Spacer(
            1,
            20
        )
    )


    # --------------------------------------------------------
    # Claims
    # --------------------------------------------------------

    for idx, result in enumerate(
        results,
        start=1
    ):

        status = result.get(
            "status",
            "UNKNOWN"
        )

        claim = safe_html(
            result.get(
                "claim",
                ""
            )
        )

        speaker = safe_html(
            result.get(
                "claim_speaker",
                "Unknown"
            )
        )

        event_type = safe_html(
            result.get(
                "claim_type",
                "Unknown"
            )
        )

        reason = safe_html(
            result.get(
                "reason",
                ""
            )
        )


        story.append(
            Paragraph(
                f"Claim {idx} — "
                f"{safe_html(status)}",
                heading_style
            )
        )


        story.append(
            Paragraph(
                f"<b>Claim:</b> {claim}",
                body_style
            )
        )


        story.append(
            Paragraph(
                f"<b>Speaker:</b> {speaker}",
                body_style
            )
        )


        story.append(
            Paragraph(
                f"<b>Event Type:</b> {event_type}",
                body_style
            )
        )


        if result.get(
            "evidence_start"
        ) is not None:

            start = result[
                "evidence_start"
            ]

            end = result.get(
                "evidence_end",
                start
            )

            story.append(
                Paragraph(
                    f"<b>Evidence Time:</b> "
                    f"{format_timestamp(start)} – "
                    f"{format_timestamp(end)}",
                    body_style
                )
            )


        story.append(
            Paragraph(
                f"<b>Verification:</b> "
                f"{safe_html(status)}",
                body_style
            )
        )


        story.append(
            Paragraph(
                f"<b>Reason:</b> {reason}",
                body_style
            )
        )


        story.append(
            Spacer(
                1,
                10
            )
        )


        # ----------------------------------------------------
        # Supporting evidence
        # ----------------------------------------------------

        story.append(
            Paragraph(
                "<b>Supporting Evidence</b>",
                body_style
            )
        )


        checks = result.get(
            "checks",
            {}
        )


        content_check = checks.get(
            "content_consistency",
            {}
        )


        details = content_check.get(
            "details",
            {}
        )


        supporting_candidates = details.get(
            "supporting_candidates",
            []
        )


        for evidence in supporting_candidates:

            evidence_id = evidence.get(
                "evidence_id",
                "N/A"
            )

            evidence_speaker = safe_html(
                evidence.get(
                    "speaker",
                    "Unknown"
                )
            )

            evidence_start = evidence.get(
                "start",
                0
            )

            evidence_end = evidence.get(
                "end",
                0
            )

            evidence_text = safe_html(
                evidence.get(
                    "text",
                    ""
                )
            )


            story.append(
                Paragraph(
                    f"<b>Evidence "
                    f"{safe_html(evidence_id)}</b> "
                    f"| {evidence_speaker} | "
                    f"{format_timestamp(evidence_start)}"
                    f"–"
                    f"{format_timestamp(evidence_end)}",
                    body_style
                )
            )


            story.append(
                Paragraph(
                    evidence_text,
                    body_style
                )
            )


            story.append(
                Spacer(
                    1,
                    6
                )
            )


        story.append(
            Spacer(
                1,
                15
            )
        )


    document.build(
        story
    )


    return PDF_PATH


# ============================================================
# 12. PDF EXPORT
# ============================================================

st.divider()

st.subheader(
    "📥 Export"
)


if st.button(
    "📥 Export Verified MoM as PDF",
    use_container_width=True
):

    try:

        generated_pdf = create_mom_pdf(
            results,
            verification_data
        )


        with open(
            generated_pdf,
            "rb"
        ) as pdf_file:

            pdf_bytes = pdf_file.read()


        st.download_button(
            label="Download PDF",
            data=pdf_bytes,
            file_name=(
                f"{MEETING_ID}_"
                "Verifiable_Meeting_Minutes.pdf"
            ),
            mime="application/pdf",
            use_container_width=True
        )


        st.success(
            "PDF generated successfully."
        )


    except Exception as error:

        st.error(
            f"PDF generation failed: {error}"
        )

'''


# ============================================================
# WRITE NEW APP
# ============================================================

with open(
    APP_PATH,
    "w",
    encoding="utf-8"
) as file:

    file.write(
        APP_CODE
    )


print("✓ New app.py created from scratch")
print(APP_PATH)
print(
    "Size:",
    os.path.getsize(APP_PATH),
    "bytes"
)
```

```python
# Generalized Pyannote function:
# ============================================================
# Add reusable Pyannote speaker diarization functions
# Uses in-memory audio loading to avoid TorchCodec dependency
# and supports Pyannote Audio 4.x DiarizeOutput format.
# ============================================================
from pathlib import Path
import os
import json
import torch
import soundfile as sf

def load_pyannote_pipeline(hf_token=None):
    """Load Pyannote speaker diarization pipeline."""
    from pyannote.audio import Pipeline

    if hf_token is None:
        hf_token = os.environ.get("HF_TOKEN")

    pipeline = Pipeline.from_pretrained(
        "pyannote/speaker-diarization-3.1",
        token=hf_token
    )

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    pipeline.to(device)

    return pipeline


def run_pyannote_diarization(
    audio_path,
    output_path=None,
    hf_token=None
):
    """
    Run Pyannote diarization using an in-memory waveform.

    Returns:
        list of dictionaries containing start, end and speaker.
    """

    pipeline = load_pyannote_pipeline(hf_token)

    waveform, sample_rate = sf.read(str(audio_path))

    if waveform.ndim == 1:
        waveform = waveform[None, :]
    else:
        waveform = waveform.T

    waveform = torch.tensor(
        waveform,
        dtype=torch.float32
    )

    diarization_output = pipeline({
        "waveform": waveform,
        "sample_rate": sample_rate
    })

    # Pyannote 4.x returns DiarizeOutput.
    # Exclusive diarization gives one speaker per time region.
    annotation = diarization_output.exclusive_speaker_diarization

    segments = []

    for turn, _, speaker in annotation.itertracks(
        yield_label=True
    ):
        segments.append({
            "start": round(float(turn.start), 3),
            "end": round(float(turn.end), 3),
            "speaker": speaker
        })

    result = {
        "audio_path": str(audio_path),
        "sample_rate": sample_rate,
        "num_segments": len(segments),
        "speakers": sorted(
            list(set(x["speaker"] for x in segments))
        ),
        "segments": segments
    }

    if output_path is not None:
        output_path = Path(output_path)
        output_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        with open(output_path, "w", encoding="utf-8") as f:
            json.dump(
                result,
                f,
                indent=2,
                ensure_ascii=False
            )

    return result
```

```python
# ============================================================
# Add reusable speaker-word alignment functions
# Aligns WhisperX word timestamps with Pyannote speaker
# segments using temporal overlap and groups words into
# speaker-attributed utterances.
# ============================================================

from pathlib import Path

pipeline_file = Path(
    "/content/drive/MyDrive/"
    "MTechIndProj/MoM_Project/07_dashboard/"
    "meeting_pipeline.py"
)

speaker_alignment_code = r'''

# ============================================================
# Speaker–Word Alignment
# ============================================================

def calculate_temporal_overlap(
    word_start,
    word_end,
    speaker_start,
    speaker_end
):
    """Calculate temporal overlap between a word and speaker segment."""

    overlap_start = max(word_start, speaker_start)
    overlap_end = min(word_end, speaker_end)

    return max(0.0, overlap_end - overlap_start)


def assign_speaker_to_word(
    word_start,
    word_end,
    diarization_segments
):
    """Assign the speaker with the largest temporal overlap."""

    best_speaker = None
    best_overlap = 0.0

    for segment in diarization_segments:

        overlap = calculate_temporal_overlap(
            word_start,
            word_end,
            segment["start"],
            segment["end"]
        )

        if overlap > best_overlap:
            best_overlap = overlap
            best_speaker = segment["speaker"]

    return best_speaker, round(best_overlap, 3)


def align_words_with_speakers(
    whisper_data,
    pyannote_data
):
    """
    Align WhisperX word-level timestamps with Pyannote
    speaker diarization.
    """

    words = whisper_data.get("word_segments", [])

    if not words:
        for segment in whisper_data.get("segments", []):
            words.extend(segment.get("words", []))

    diarization_segments = pyannote_data.get(
        "segments",
        []
    )

    aligned_words = []

    for word in words:

        if "start" not in word or "end" not in word:
            continue

        speaker, overlap = assign_speaker_to_word(
            float(word["start"]),
            float(word["end"]),
            diarization_segments
        )

        aligned_words.append({
            "word": word["word"],
            "start": round(float(word["start"]), 3),
            "end": round(float(word["end"]), 3),
            "score": word.get("score"),
            "speaker": speaker,
            "speaker_overlap": overlap
        })

    return aligned_words


def group_words_into_speaker_utterances(
    aligned_words
):
    """Group consecutive speaker-attributed words into utterances."""

    if not aligned_words:
        return []

    utterances = []

    current_speaker = aligned_words[0]["speaker"]
    current_words = [aligned_words[0]]

    for word in aligned_words[1:]:

        if word["speaker"] == current_speaker:
            current_words.append(word)

        else:
            utterances.append({
                "speaker": current_speaker,
                "start": current_words[0]["start"],
                "end": current_words[-1]["end"],
                "text": " ".join(
                    item["word"]
                    for item in current_words
                ),
                "words": current_words
            })

            current_speaker = word["speaker"]
            current_words = [word]

    utterances.append({
        "speaker": current_speaker,
        "start": current_words[0]["start"],
        "end": current_words[-1]["end"],
        "text": " ".join(
            item["word"]
            for item in current_words
        ),
        "words": current_words
    })

    return utterances


def run_speaker_word_alignment(
    whisper_data,
    pyannote_data,
    output_path=None
):
    """
    Run complete speaker-word alignment.

    Returns:
        Dictionary containing aligned words and
        speaker-attributed utterances.
    """

    import json

    aligned_words = align_words_with_speakers(
        whisper_data,
        pyannote_data
    )

    utterances = group_words_into_speaker_utterances(
        aligned_words
    )

    result = {
        "num_words": len(aligned_words),
        "num_utterances": len(utterances),
        "speakers": sorted(
            list(set(
                word["speaker"]
                for word in aligned_words
                if word["speaker"] is not None
            ))
        ),
        "words": aligned_words,
        "utterances": utterances
    }

    if output_path is not None:

        output_path = Path(output_path)
        output_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        with open(
            output_path,
            "w",
            encoding="utf-8"
        ) as f:
            json.dump(
                result,
                f,
                indent=2,
                ensure_ascii=False
            )

        print(
            f"Speaker-word alignment saved to: {output_path}"
        )

    return result
'''

with open(
    pipeline_file,
    "a",
    encoding="utf-8"
) as f:
    f.write("\n" + speaker_alignment_code)

print("Speaker-word alignment functions appended successfully.")
print("File size:", pipeline_file.stat().st_size, "bytes")

text = pipeline_file.read_text(encoding="utf-8")

print(
    "run_speaker_word_alignment saved:",
    "def run_speaker_word_alignment" in text
)
```
