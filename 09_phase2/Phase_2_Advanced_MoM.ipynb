```python
import json
from pathlib import Path

TRANSCRIPT_FILE = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project/"
    "data/raw/ami/ES2006b/transcript/"
    "ES2006b_timestamped_transcript.json"
)

with open(TRANSCRIPT_FILE, "r", encoding="utf-8") as f:
    words = json.load(f)


def group_words_into_utterances(words, max_gap=0.8):

    utterances = []
    current = []

    for word in words:

        if not current:
            current.append(word)
            continue

        previous = current[-1]

        speaker_changed = word["speaker"] != previous["speaker"]
        pause_detected = word["start"] - previous["end"] > max_gap

        if speaker_changed or pause_detected:
            utterances.append(current)
            current = []

        current.append(word)

    if current:
        utterances.append(current)

    results = []

    for group in utterances:
        results.append({
            "speaker": group[0]["speaker"],
            "start": group[0]["start"],
            "end": group[-1]["end"],
            "text": " ".join(w["text"] for w in group)
        })

    return results


utterances = group_words_into_utterances(words)

print("Total words:", len(words))
print("Total utterances:", len(utterances))

for i, item in enumerate(utterances[:20], 1):
    print(
        f"\n{i}. [{item['start']:.2f}-{item['end']:.2f}] "
        f"{item['speaker']}: {item['text']}"
    )
```

```python
import asyncio
import edge_tts
import nest_asyncio
from pydub import AudioSegment
from google.colab import files

nest_asyncio.apply()

speakers = {
    "Elena": "en-US-JennyNeural",
    "Rohan": "en-US-GuyNeural",
    "Maya": "en-US-AriaNeural",
    "Devika": "en-GB-SoniaNeural"
}

conversation = [
    ("Elena", "Okay, let's hit record so folks out sick can catch up. Morning everyone! Let's do a quick status check on the Q4 Checkout Launch. Everything is due Friday, so I want to make sure our task board is clear. Rohan, you're up first. How's the new payment API looking?"),

    ("Rohan", "Uh, yeah, so I'm actually blocked right now. I can't finalize the payment endpoints until Maya locks in the frontend state payload. I need to know the exact JSON structure her checkout screen is sending over before I write the validation logic."),

    ("Elena", "Got it. So Rohan is waiting on Maya. Maya, over to you. Where are we on that checkout UI payload?"),

    ("Maya", "Um, well, that's the thing. I haven't built the form state yet because I'm waiting on Devika. She's still working on the final design specs for the promo-code field, and I don't know if that's going to be a separate object or nested inside the cart array."),

    ("Elena", "Okay, so Maya is waiting on Devika. Devika, no pressure, but looks like the whole chain starts with you! How close are those promo-code design specs?"),

    ("Devika", "Wait, what? I thought I was waiting on Rohan."),

    ("Rohan", "Wait, me? Why are you waiting on me?"),

    ("Devika", "Because yesterday you said the third-party payment gateway has strict limits on how many discount lines we can pass! I can't finalize the UI design until you test the API and tell me if users can stack multiple promo codes or just use one."),

    ("Maya", "Hold on. Let me get this straight. Rohan is waiting on me."),

    ("Rohan", "And you're waiting on Devika."),

    ("Devika", "And I'm waiting on Rohan."),

    ("Elena", "Aha. We have a classic circular dependency on our hands. Nobody can start because everyone's task is downstream from someone else's. If we leave the project graph like this, we're going to hit Friday with zero progress."),

    ("Rohan", "Yeah, we're completely deadlocked. How do we break the loop?"),

    ("Elena", "Rohan, you're the root dependency because the third-party gateway rules are a hard technical constraint. Can you run a quick spike today? Just test the gateway sandbox and give Devika a simple yes or no on stacked promo codes by noon. You don't need Maya's frontend payload to test that."),

    ("Rohan", "Oh, yeah, totally. I can test the sandbox with dummy data in like an hour."),

    ("Elena", "Awesome. Once Devika gets that answer at noon, she can finish the wireframe by three PM. Maya, that unblocks your JSON payload structure for tomorrow morning, which then unblocks Rohan's full API build."),

    ("Maya", "Makes complete sense. That turns the circle back into a straight line."),

    ("Elena", "Exactly. I'll update the task dependencies on the board right now so we don't get stuck in a loop again. Alright, let's get to it. I'll stop the recording here.")
]

async def create_audio():
    final_audio = AudioSegment.empty()

    for i, (speaker, text) in enumerate(conversation):

        voice = speakers[speaker]

        tts = edge_tts.Communicate(
            text=text,
            voice=voice,
            rate="-5%"
        )

        filename = f"speaker_{i}.mp3"

        await tts.save(filename)

        segment = AudioSegment.from_mp3(filename)

        # Small natural pause between turns
        final_audio += segment
        final_audio += AudioSegment.silent(duration=500)

    # Export WhisperX-compatible WAV
    final_audio = final_audio.set_frame_rate(16000)
    final_audio = final_audio.set_channels(1)
    final_audio = final_audio.set_sample_width(2)

    final_audio.export(
        "Q4_Checkout_DAG_Cycle_Test.wav",
        format="wav"
    )

    print("Audio generation completed!")

await create_audio()

files.download("Q4_Checkout_DAG_Cycle_Test.wav")
```

```python
import json
from pathlib import Path

PROJECT = Path(
    "/content/drive/MyDrive/MTechIndProj/MoM_Project"
)

RESULT_FILE = (
    PROJECT / "outputs/live/"
    "Q4_Checkout_DAG_Cycle_Test_4_phase2_rerun.json"
)

with open(RESULT_FILE, "r", encoding="utf-8") as f:
    result = json.load(f)

p2 = result["phase2_dependency_analysis"]

# Manually annotated ground-truth dependency
ground_truth = {
    ("T1", "T2", "MUST_PRECEDE")
}

# Model-predicted, NLI-verified dependencies
predicted = {
    (
        edge["prerequisite_id"],
        edge["dependent_id"],
        edge["relation"]
    )
    for edge in p2.get("dependencies", [])
    if edge.get("verification_status") == "SUPPORTED"
}

TP = len(ground_truth & predicted)
FP = len(predicted - ground_truth)
FN = len(ground_truth - predicted)

precision = TP / (TP + FP) if (TP + FP) else 0.0
recall = TP / (TP + FN) if (TP + FN) else 0.0
f1 = (
    2 * precision * recall / (precision + recall)
    if (precision + recall) else 0.0
)

print("=" * 60)
print("PHASE 2 DEPENDENCY EVALUATION — Q4 TEST 4")
print("=" * 60)

print("Ground-truth edges:", ground_truth)
print("Predicted verified edges:", predicted)

print("\nConfusion counts")
print("True Positives:", TP)
print("False Positives:", FP)
print("False Negatives:", FN)

print("\nEvaluation metrics")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-score:  {f1:.4f}")

print("\nEvaluation scope: One meeting, one ground-truth dependency.")
```

```python
import meeting_pipeline as mp

resources = getattr(mp, "_NLI_RESOURCES", None)
if resources is None:
    resources = mp.load_nli_model()
    mp._NLI_RESOURCES = resources

tokenizer, model, device = resources

evidence = (
    "Because yesterday you said the third party payment gateway has strict "
    "limits on how many discount lines we can pass. I can't finalize the UI "
    "design until you test the API and tell me if users can stack multiple "
    "promo codes or just use one."
)

hypotheses = [
    '"Test the API and tell me if users can stack multiple promo codes or just use one" '
    'must happen before "Finalize the UI design".',

    "The API must be tested before the UI design can be finalized.",

    "The UI design cannot be finalized until the API test determines whether "
    "users can stack multiple promo codes."
]

for i, hypothesis in enumerate(hypotheses, 1):
    result = mp.run_nli(
        evidence, hypothesis, tokenizer, model, device
    )
    print(f"\nTEST {i}")
    print("Hypothesis:", hypothesis)
    print("Result:", result)
```
